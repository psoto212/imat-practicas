```
import pandas as pd
```

```
def airbnb_read(csv):
    df = pd.read_csv(csv)
    return df

df = airbnb_read("airbnb.csv")
```

```
def airbnb_size(df):
    return df.shape

airbnb_size(df)
```

```
def airbnb_columns(df):
    return df.columns

airbnb_columns(df)
```

```
def airbnb_countries(df):
    numero_alojamientos = df["country"].value_counts()
    return numero_alojamientos



airbnb_countries(df)
```

```
def airbnb_boroughs(df):
    df["neighbourhood group"] = df["neighbourhood group"].replace({"manhatan" : "Manhattan", "brookln": "Brooklyn"})
    barrio = df["neighbourhood group"].value_counts()
    return barrio

airbnb_boroughs(df)
```

```
def airbnb_aggregate(df):
    resultado = df.groupby("neighbourhood group")["price"].agg(["mean", "median", "min", "max"])
    return resultado

airbnb_aggregate(df)
```

```
mediana = df["salary"].median()
df["salary"] = df["salary"].fillna(mediana)
```

```
df["price"] = df["price"].fillna(df.groupby("city")["price"].transform("median"))
```
