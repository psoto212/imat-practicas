```
def clean_students(df):
    df["score"] = df["score"].astype(str)
    df["score"] = df["score"].str.extract(r"(\d+)")
    df["score"] = pd.to_numeric(df["score"], errors = "coerce")


    df["study_time"] = df["study_time"].astype(str).str.extract(r"(\d+)")
    df["study_time"] = pd.to_numeric(df["study_time"], errors = "coerce")
    df["study_time"] = df["study_time"].fillna(
        df.groupby("group")["study_time"].transform("median")
    )

    q1 = df["score"].quantile(0.25)
    q3 = df["score"].quantile(0.75)
    iqr = q3 - q1
    lim_inf = q1 -2.5*iqr
    lim_sup = q3 +2.5*iqr
    anomalos = df[df["score"] < lim_inf | df["score"] > lim_sup]

    for anomalo in anomalos:
        df.drop(anomalo)

    return df

```

```
def generate_universities_database(soup):
    tabla = soup.find("table", class_ = "ranking-table")
    rows = tabla.find("tr")
    filas = []

    for row in rows[1:]:
        celda = row.find("td")
        c1 = celda[0].get_text(strip = True)
        c2 = celda[1].get_text(strip = True)
        c3 = celda[2].get_text(strip = True)
        c4 = celda[3].get_text(strip = True)

        filas.append({
            "University" : c1,
            "Country" : c2,
            "Rank" : c3,
            "Tuition" : c4
        })
    df = pd.DataFrame(filas)

    return df

```

```
def clean_universities(df):
    df = df.copy()
    df["University"] = df["University"].astype(str).str.lower()
    df["Rank"] = df["Rank"].astype(str).str.replace(r"[^0-9]", "", regex = True)
    df["Rank"] = pd.to_numeric(df["Rank"], errors = "coerce")
    df["Tuition"] = df["Tuition"].astype(str).str.extract(r"(\d.+)")
    df["Tuition"] = df["Tuition"].str.replace(".", "", regex = False)
    df["Tuition"] = pd.to_numeric(df["Tuition"], errors = "coerce")


    return df

```

```
def get_employment_database(universities_selected):
    for university in universities_selected:
        BASE_URL = "https://api.example.com"
        url = BASE_URL + "/api/universities/" + university
        response = requests.get(url)
        filas = []
        if response.status_code == 200:
            data = response.json()

            employment = data.get("employment_rate", np.nan)
            salary = data.get("average_salary", np.nan)
            students = data.get("students", np.nan)
        
        else:
            employment = np.nan
            salary = np.nan
            students = np.nan

        filas.append({
            "University" : university,
            "Employment Rate" : employment,
            "Average Salary" : salary,
            "Students" : students
            })
        

    df = pd.DataFrame(filas)

    return df
```

```
def clean_employment_database(df):
    df["Employment Rate"] = df["Employment Rate"].astype(str).str.extract(r"\d+")
    df["Employment Rate"] = pd.to_numeric(df["Employment Rate"], errors = "coerce")

    df["Average Salary"] = df["Average Salary"].astype(str).str.extract(r"\d+")
    df["Average Salary"] = pd.to_numeric(df["Average Salary"], errors = "coerce")

    df["Students"] = pd.to_numeric(df["Students"], errors = "coerce")

    return df
```
