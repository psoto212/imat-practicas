```
import pandas as pd
import numpy as np
import requests
from bs4 import BeautifulSoup
import re
import matplotlib.pyplot as plt
```

```
def get_soup(url):
    response = requests.get(url)
    soup = BeautifulSoup(response.text, "html.parser")
    return soup
```

```
def generate_restaurants_database(soup):
    data = soup.find("table", id="restaurants-table")
    rows = data.find_all("tr")

    filas = []
    for row in rows[1:]:
        celdas = row.find_all("td")
        c1 = celdas[0].get_text(strip=True)
        c2 = celdas[1].get_Text(strip=True)
        c3 = celdas[2].get_text(strip=True)
        c4 = celdas[3].get_text(strip=True)

        filas.append({
            "Restaurant" : c1,
            "City" : c2,
            "Food Type" : c3,
            "Average Price" : c4
        })


    df = pd.DataFrame(filas)
    df["Restaurants"] = df["Restaurants"].astype(str).str.lower()
    df["Restaurants"] = df["Restaurants"].str.replace(" ", "-", regex = False)
    df["Food Type"] = df["Food Type"].astype(str).str.extract(r"()")
    return df
```
