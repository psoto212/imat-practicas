## 1️⃣ Lectura y exploración de datos
**Objetivo:** Cargar un archivo CSV y explorar sus características básicas.

```
import pandas as pd

# TODO: Cambia el nombre del archivo por el tuyo si es necesario
df = pd.read_csv('drivers.csv')

# Muestra las primeras filas
display(df.head())

# Muestra información general del DataFrame
df.info()

# Muestra estadísticas básicas
display(df.describe())
```

## 2️⃣ Selección y filtrado
**Objetivo:** Practicar cómo seleccionar columnas y filtrar filas según condiciones.

```
# Selecciona la columna 'name'
display(df['name'].head())

# Filtra las filas donde los puntos sean mayores que 6
df_filtrado = df[df['points'] > 6]
display(df_filtrado)
```

## 3️⃣ Limpieza de datos
**Objetivo:** Eliminar duplicados, detectar y rellenar valores nulos.

```
# Elimina duplicados
df = df.drop_duplicates()

# Detecta valores nulos
display(df.isnull().sum())

# Rellena valores nulos en 'points' con la media de la columna
df['points'].fillna(df['points'].mean(), inplace=True)
display(df.head())
```

## 4️⃣ Análisis básico
**Objetivo:** Calcular medidas estadísticas simples y conteos.

```
# Número de valores únicos por columna
print(df.nunique())

# Valor máximo de puntos
print('Máximo de puntos:', df['points'].max())

# Media de puntos
print('Media de puntos:', df['points'].mean())
```

## 5️⃣ Visualización básica
**Objetivo:** Usar boxplots e histogramas para detectar anomalías.

```
import matplotlib.pyplot as plt

# Dibuja un boxplot de la columna 'points'
df['points'].plot.box()
plt.title('Distribución de puntos')
plt.show()

# Dibuja un histograma
df['points'].hist()
plt.title('Histograma de puntos')
plt.show()
```

## 6️⃣ Reto final: detección de outliers con IQR
**Objetivo:** Detectar y eliminar valores atípicos usando el rango intercuartil.

```
Q1 = df['points'].quantile(0.25)
Q3 = df['points'].quantile(0.75)
IQR = Q3 - Q1

lim_inf = Q1 - 1.5 * IQR
lim_sup = Q3 + 1.5 * IQR

print('Rango aceptable:', lim_inf, '-', lim_sup)

df_limpio = df[(df['points'] >= lim_inf) & (df['points'] <= lim_sup)]
print('Filas antes:', len(df))
print('Filas después:', len(df_limpio))

df_limpio.to_csv('drivers_limpios.csv', index=False)
print('Archivo guardado como drivers_limpios.csv')
```
