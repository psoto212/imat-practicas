# Language Detector

```
import requests
from urllib.parse import urlunsplit, urlencode, urljoin
```

Example of an API in which yo send a text, and it looks for grammar/spell errors. </br> <a href='https://languagetool.org/http-api/'> Link to API documentation</a>

```
SCHEME = "https"
BASE = "api.languagetoolplus.com/v2"


def build_api_url(path, query_dict=None):
    if query_dict is None:
        query_dict = {}
    query = urlencode(query_dict)
    return urlunsplit((SCHEME, BASE, path, query, ""))
```

```
url_languages = build_api_url("languages")
response_languages = requests.get(url_languages)
```

```
response_languages.json()
```

```
text = "This is just an example txt, with a mistake here."
url_checker = build_api_url("check")
body = {'text': text, 'language': 'en-US'}
response_text = requests.post(url_checker, data=body)
```

```
response_text
```
