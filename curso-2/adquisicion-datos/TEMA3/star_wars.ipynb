```
import pandas as pd
import requests
```

This is just an example practice, in which we obtain data form the <a href='https://swapi.dev/documentation'>SW API</a>.

# Simple Request for just 1 person

```
people_request = requests.get("https://swapi.dev/api/people/1")
people_json = people_request.json()
```

```
people_json
```

# Loop to get name and url of all the people available in the API

```
names = {}
links = {}
i = '1'
while True:
    url = f"https://swapi.dev/api/people/{i}"
    people_request = requests.get(url)
    if people_request.status_code != 200:
        break
    else:
        people_json = people_request.json()
        names[i] = people_json['name']
        links[i] = url
        i = str(int(i) + 1)
        print(i)
```

```
people_urls_df = pd.DataFrame(data=links.items(), columns=['id', 'url'])
people_names_df = pd.DataFrame(data=names.items(), columns=['id', 'name'])
```

```
people_df = people_names_df.merge(people_urls_df, on='id')
```

```
people_df
```

# Loop to obtain the same information than before, but additionally the vehicles of each character

```
names = {}
links = {}
vehicles_dict = {}
i = '1'
while True:
    url = f"https://swapi.dev/api/people/{i}"
    people_request = requests.get(url)
    if people_request.status_code != 200:
        break
    else:
        people_json = people_request.json()
        names[i] = people_json['name']
        links[i] = url
        if people_json['vehicles'] != None:
            vehicles = []
            for vehicle_url in people_json['vehicles']:
                vehicle_name = requests.get(vehicle_url).json()['name']
                vehicles.append(vehicle_name)
        else:
            vehicles = []
        vehicles_dict[i] = vehicles
        i = str(int(i) + 1)
```

```
people_urls_df = pd.DataFrame(data=links.items(), columns=['id', 'url'])
people_names_df = pd.DataFrame(data=names.items(), columns=['id', 'name'])
people_vehicles_df = pd.DataFrame(data=vehicles_dict.items(), columns=['id', 'vehicle'])
```

```
people_df = people_names_df.merge(people_urls_df, on='id')
people_df = people_df.merge(people_vehicles_df, on='id')
```
