```python
import requests
from urllib.parse import urlunsplit, urlencode, urljoin
```

```python
SCHEME = "https"
BASE = "api.schiphol.nl/public-flights"


def build_api_url(path, query_dict=None):
    if query_dict is None:
        query_dict = {}
    query = urlencode(query_dict)
    return urlunsplit((SCHEME, BASE, path, query, ""))
```

https://api.schiphol.nl/public-flights/flights?includedelays=false&page=0&sort=%2BscheduleTime

```python
query_dict = {'includedelays': 'false', 'page': '0', "sort": "+scheduleTime"}
```

```python
airport_url = build_api_url("flights", query_dict)
airport_url
```

```python
response = requests.get(airport_url)
```

```python
print(response.status_code)
print(response.text)
```
