# EJERCICIOS TEORÍA TEMA 9 - Limpieza y Tratamiento de Datos - Solución

El archivo CSV 'transacciones.csv' contiene la información de transacciones de una tienda en línea. 

1. Identificar y manejar los valores nulos en el DataFrame, rellenándolos con un valor predeterminado (en caso de que el nulo esté en el campo cliente o el campo fecha) o eliminando las filas con valores nulos (en el resto de casos).
2. Filtrar el DataFrame para obtener transacciones de pantalones.
3. Ordenar: Ordenar el DataFrame por fecha para tener las transacciones en orden cronológico.
4. Eliminar duplicados: Eliminar transacciones duplicadas basadas en el ID de la transacción.
5. Añadir y eliminar registros: Agregar 3 nuevas transacciones al DataFrame y eliminar 3 transacciones existentes.
6. Conversión de tipos de datos: Convertir el tipo de datos de la columna 'Fecha' a objetos de fecha y tiempo.
7. Datos categóricos: Convertir la columna 'Producto' a datos categóricos.
8. Fechas y tiempo: Calcular la duración entre transacciones y filtrar transacciones para el primer día del mes.

```
# Importar la librería pandas
import pandas as pd

# Cargar el archivo CSV
df = pd.read_csv('transacciones.csv')

# 1. Identificar y manejar los valores nulos
# Rellenar valores nulos en 'Cliente' y 'Fecha' con valores predeterminados
# df['Cliente'].fillna('Cliente_Desconocido', inplace=True)
# df['Fecha'].fillna('2024-01-01', inplace=True)  # Rellenar con una fecha predeterminada
df.fillna({'Cliente': 'Cliente_Desconocido'}, inplace=True)
df.fillna({'Fecha': '2024-01-01'}, inplace=True)  # Rellenar con una fecha predeterminada

# Eliminar filas con valores nulos en el resto de columnas
df.dropna(inplace=True)

# 2. Filtrar el DataFrame para obtener transacciones de pantalones
transacciones_pantalones = df[df['Producto'] == 'Pantalones']

# 3. Ordenar el DataFrame por fecha para tener las transacciones en orden cronológico
df['Fecha'] = pd.to_datetime(df['Fecha'])
df = df.sort_values(by='Fecha')

# 4. Eliminar duplicados basados en el ID de la transacción
df.drop_duplicates(subset='ID_Transaccion', inplace=True)

# 5. Añadir y eliminar registros
# Agregar 3 nuevas transacciones al DataFrame
nuevas_transacciones = [
    {'ID_Transaccion': 101, 'Fecha': '2022-01-15', 'Producto': 'Camiseta', 'Precio': 19.99, 'Cantidad': 2, 'Cliente': 105},
    {'ID_Transaccion': 102, 'Fecha': '2022-01-16', 'Producto': 'Zapatos', 'Precio': 39.99, 'Cantidad': 1, 'Cliente': 106},
    {'ID_Transaccion': 103, 'Fecha': '2022-01-17', 'Producto': 'Sombrero', 'Precio': 14.99, 'Cantidad': 3, 'Cliente': 107}
]

# df = df.append(nuevas_transacciones, ignore_index=True) #método deprecado

nuevas_transacciones_df =  pd.DataFrame(nuevas_transacciones)
df = pd.concat([df, nuevas_transacciones_df], ignore_index=True)

# Eliminar 3 transacciones existentes
transacciones_a_eliminar = [2, 4, 6]
df = df[~df['ID_Transaccion'].isin(transacciones_a_eliminar)]

# 6. Conversión de tipos de datos: Convertir el tipo de datos de la columna 'Fecha' a objetos de fecha y tiempo
df['Fecha'] = pd.to_datetime(df['Fecha'])

# 7. Datos categóricos: Convertir la columna 'Producto' a datos categóricos
df['Producto'] = df['Producto'].astype('category')

# 8. Fechas y tiempo: Calcular la duración entre transacciones y filtrar transacciones para el primer día del mes
# Calcular la duración entre transacciones
df['Duracion_entre_transacciones'] = df['Fecha'].diff().fillna(pd.Timedelta(seconds=0))

# Filtrar transacciones para el primer día del mes
transacciones_primer_dia_mes = df[df['Fecha'].dt.day == 1]

# Mostrar el DataFrame resultante
print(df)
```
