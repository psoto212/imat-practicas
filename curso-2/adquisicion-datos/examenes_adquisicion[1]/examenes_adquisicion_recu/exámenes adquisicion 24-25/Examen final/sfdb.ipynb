```
def obtain_numeric_salary(best_jobs_1c: pd.DataFrame) -> pd.DataFrame:
    best_jobs_1c["salary_average"] = best_jobs_1c["salary_average"].str.replace("$", "", regex = False)
    best_jobs_1c["salary_average"] = best_jobs_1c["salary_average"].str.replace(",", "", regex = False)
    best_jobs_1c["salary_average"] = pd.to_numeric(best_jobs_1c["salary_average"], errors = "coerce")

    return best_jobs_1c
```
