```
import pandas as pd
import re
import matplotlib.pyplot as plt
import numpy as np
```

```
df_drivers=pd.read_csv("drivers.csv")
df_radars=pd.read_csv("radars.csv")
df_fines=pd.read_csv("fines.csv")
```

```
df_radars_noduplicates = df_radars.drop_duplicates(["id"])
df_radars_noduplicates.shape[0]
```

Pregunta 2

```
pattern=re.compile(r"([0-9]{1,2})º([0-9]{1,2})'([0-9]{1,2}.[0-9]{1,2})\"\s([A-Z])")
re.findall(pattern, "42º50'28.00\" N,\n 1º14'20.36\" E")


df_radars_noduplicates["location"]=df_radars_noduplicates["location"].apply(lambda localizacion: re.findall(pattern,localizacion))

df_radars_noduplicates
```

Pregunta 3

```
df_fines["speed"].plot.box()
plt.show()
```

```
m=1.5
IQR= df_fines.speed.quantile(0.75)-df_fines.speed.quantile(0.25)
speed_thresholds = [df_fines.speed.quantile(0.25) - m*IQR, df_fines.speed.quantile(0.75) + m*IQR]
speed_thresholds
```

```
df_fines.loc[(df_fines.speed < speed_thresholds[0]), "speed"] = np.nan
df_fines.dropna
df_fines.head()
```
