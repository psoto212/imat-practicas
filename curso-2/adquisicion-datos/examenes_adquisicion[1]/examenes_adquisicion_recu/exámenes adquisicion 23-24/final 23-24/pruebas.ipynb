```
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup
import requests
import re
from urllib.parse import urlunsplit, urlencode
import io
```

```
SCHEME = "https"
BASE = "date.nager.at/"
PATH = "/api/v3/PublicHolidays/2024/ES"
QUERY = ""
url = urlunsplit((SCHEME, BASE, PATH, QUERY, ""))
response = requests.get(url)
```

```
df_fiestas = pd.DataFrame.from_dict(response.json())
df_fiestas_final = df_fiestas[["counties","localName"]].explode("counties").value_counts("counties")
df_fiestas_final.reset_index()
df_fiestas_final
```

```
response = requests.get("https://es.wikipedia.org/wiki/ISO_3166-2:ES")
soup = BeautifulSoup(response.content)
table = soup.find("table")
df_comunidades = pd.read_html(io.StringIO(str(table)))
df_comunidades
```
