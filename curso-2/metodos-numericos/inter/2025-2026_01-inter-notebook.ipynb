# Intercuatrimestral

```
import numpy as np
import matplotlib.pyplot as plt

def foo(f, a, b, N):
    h = (b-a)/(N-1)
    xx = np.linspace(a,b,N)
    yy = f(xx)
    dd = np.zeros(N)
    dd[0] = -3*yy[0] + 4*yy[1] - yy[2]
    dd[1:-1] = yy[2:] - yy[:-2]
    dd[-1] = yy[-3] - 4*yy[-2] + 3*yy[-1]
    return xx, yy, dd/2/h
```

## Ejercicio 2

Determinamos el mínimo número de nodos `N` que garantizan un error relativo máximo inferior a $10^{-5}$:

```
f = lambda x: np.exp(4*x)
a, b = -1, 1

Nmin = 3    # Necesitamos 3 datos al menos
Er = 1      # Garantiza el acceso al bucle
while Er >= 1e-5:
    Nmin += 1
    xx, yy, dd = foo(f, a, b, N=Nmin)
    ww = 4*yy
    Er = np.max(np.abs( (dd-ww)/ww ))
    
print("Número mínimo de puntos para error relativo máximo < 1e-5:", Nmin)
```

**CONCLUIMOS** que hacen falta $1465$ nodos.

## Ejercicio 3

Verificamos la condición inicial.

```
f = lambda x: 1/(1+np.exp(x**2/2))
a, b = 0, 2

print( np.abs(f(0)-0.5) )
```

**CONCLUIMOS** que la función cumple la condición inicial.

Ahora comprobamos la ecuación diferencial, y determinamos el número mínimo de nodos a mano:

```
f = lambda x: 1/(1+np.exp(x**2/2))
a, b = 0, 2
N = 19

xx, yy, dd = foo(f, a, b, N)
vv = xx*yy*(yy-1)
Ea = np.max( np.abs(vv - dd) )
print("Err. abs. máx. con", N, "nodos:", Ea)

N = N+1
xx, yy, dd = foo(f, a, b, N)
vv = xx*yy*(yy-1)
Ea = np.max( np.abs(vv - dd) )
print("Err. abs. máx. con", N, "nodos:", Ea)
```

**CONCLUIMOS** que hacen falta $20$ nodos.

Barrido sobre `h` y gráfica:

```
f = lambda x: np.cos(x**2)
df = lambda x: -2*x*np.sin(x**2)
x = 0.7

hh = 10**(np.linspace(-15, 1, 100))
EaEa = np.zeros(len(hh))
for i in range(len(hh)):
    h = hh[i]
    xx, yy, dd = foo(f, x-h, x+h, 3)
    EaEa[i] = np.abs(dd[1]-df(x))

plt.loglog(hh,EaEa)
plt.show()
```
