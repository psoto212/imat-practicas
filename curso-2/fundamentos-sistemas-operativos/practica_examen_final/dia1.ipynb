```
import os
from multiprocessing import Process
from time import sleep


def bisnieto():
    print(f"Soy el bisnieto {os.getpid()}, mi padre es {os.getppid()}")


def nieto_1():
    print(f"Soy el nieto {os.getpid()}, mi padre es {os.getppid()}")
    sleep(2)


def nieto_2():
    print(f"Soy el nieto {os.getpid()}, mi padre es {os.getppid()}")

    p = Process(target=bisnieto)
    p.start()
    p.join()


def hijo_1():
    print(f"Soy el hijo 1 {os.getpid()}, mi padre es {os.getppid()}")

    procesos = []

    for i in range(2):
        p = Process(target=nieto_1)
        procesos.append(p)

    for p in procesos:
        p.start()

    for p in procesos:
        p.join()


def hijo_2():
    print(f"Soy el hijo 2 {os.getpid()}, mi padre es {os.getppid()}")

    p = Process(target=nieto_2)
    p.start()
    p.join()


def hijo_3():
    print(f"Soy el hijo 3 {os.getpid()}, mi padre es {os.getppid()}")
    sleep(3)


def main():
    print(f"Soy el proceso principal {os.getpid()}")

    procesos = [
        Process(target=hijo_1),
        Process(target=hijo_2),
        Process(target=hijo_3)
    ]

    for p in procesos:
        p.start()

    for p in procesos:
        p.join()


if __name__ == "__main__":
    main()
```

```
from threading import Thread, Semaphore

N = 4
terminados = 0
mutex = Semaphore(1)
controlador = Semaphore(0)

def trabajador(id):
    global terminados

    print(f"Hilo {id} trabaja")

    mutex.acquire()
    terminados += 1

    if terminados == N:
        controlador.release()

    mutex.release()

def jefe():
    controlador.acquire()
    print("Todos los hilos han terminado")

hilos = [Thread(target=trabajador, args=(i,)) for i in range(N)]
hilos.append(Thread(target=jefe))

for h in hilos:
    h.start()

for h in hilos:
    h.join()
```
