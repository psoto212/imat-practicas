Un viaje en coche entre un garaje próximo a ICAI y la playa de la Concha en San Sebastián, siguiendo el mismo recorrido, implica recorrer 420 Km y puede durar más o menos en función de diferentes factores, entre los que destacan la velocidad media a la que se circule y el tiempo dedicado a los descansos. La velocidad media a la que se circula, aunque sea siguiendo el mismo recorrido, está afectada por numerosas variables como la densidad del tráfico encontrado en la carretera, las condiciones de visibilidad, el estado del asfalto, etc. Por otro lado, el tiempo dedicado a descansar depende del nº de paradas que se realicen y la duración de cada una de ellas. 

Para realizar este viaje con seguridad y comodidad se han previsto dos paradas estratégicamente colocadas para descansar. De esta forma, el viaje se descompone en tres tramos de conducción, separados por esas dos paradas (C1 - P1 - C2 - P2 - C3). En el tramo C1 se recorren 200 km, en el tramo C2, después del primer descanso, se recorren 140 km y, finalmente en el tramo C3, el más complicado, 80 km.

En la primera parada P1, el tiempo invertido para descansar (en minutos) se puede modelar con una distribución contínua, sabiendo que siempre es superior a 15 minutos e inferior a 70 minutos, siendo el valor más probable 35 minutos. La segunda parada P2 es normalmente un 20% más larga que la primera, y se puede modelar como una variable aleatoria TP2 = 1.2 * TP1, siendo TP1 y TP2 las vv.aa. que representan el tiempo de cada una de las paradas (en minutos).

En los tres tramos de conducción (C1, C2 y C3) el tiempo empleado para recorrerlos (en minutos) se puede modelar según una distribución normal, pero de parámetros diferentes según el tramo:
- Tramo C1 (200km) TC1 ~ normal (mu=100, sigma = 10)
- Tramo C2 (140km) TC2 ~ normal (mu= 93, sigma =  5)
- Tramo C3 (80km)  TC3 ~ normal (mu= 70, sigma = 15)

Estas tres vv.aa. se pueden considerar independientes.

### a) Calcular, utilizando el operador esperanza matemática, el tiempo medio de descanso en la segunda parada, justificando la respuesta (1 punto)

```
# respuesta

print('Se pide calcular el valor esperado E[TP2]')
print('E[TP2] = 1.2 * E[TP1] (TP2 es una transf. lineal de TP1)')

# creamos la va TP1
print('TP1: v.a. tiempo de descanso en segunda parada (en min)')
print('TP1 ~ triangular(a = 15, b = 70, c= 35)')

a = 15 # min
c = 35 # moda 
b = 70 # max

loc = a
escalado = b-a
forma = (c-a)/escalado
TP1 = stats.triang(loc=loc, scale=escalado, c = forma)

# calculamos el valor medio de TP1 y el de TP2
TP1_mean = TP1.mean()
TP2_mean = 1.2 * TP1_mean
print('E[TP1] = ', TP1_mean,'min')
print('Por tanto, el tiempo medio en P2 es E[TP2] = 1.2 * E[TP1] = ', TP2_mean,'min')
```

### b) Representar sobre la misma gráfica la función de densidad del tiempo de descanso en la primera y en la segunda parada. Indica, justificando la respuesta, en qué parada hay más incertidumbre sobre el tiempo dedicado a descansar. (1.5 puntos)

```
# respuesta

# TP2
print('Primero hay que crear la v.a. TP2')
print('TP2 es simplemente un escalado directo de la pdf de TP1 ya que TP2 = 1.2 * TP1')

a = 1.2*15 # min
c = 1.2*35 # moda 
b = 1.2*70 # max

loc = a
escalado = b-a
forma = (c-a)/escalado
TP2 = stats.triang(loc=loc, scale=escalado, c = forma)

# comprobamos que está bien calculando a partir de TP2 la media
TP2_mean2 = TP2.mean()
print('Comprobación TP2 está bien creada, se obtiene la misma media que en el apartado anterior:', TP2_mean2)

# Representamos lo pedido
xs = np.linspace(TP1.ppf(0.001)-3, TP2.ppf(0.999)+3, 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(15, 4))

plt.fill_between(xs, TP1.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, TP1.pdf(xs), color="orange", label = 'pdf(TP1)')
plt.fill_between(xs, TP2.pdf(xs), color="green", alpha=0.3)
plt.plot( xs, TP2.pdf(xs), color="green", label = 'pdf(TP2)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

# ahora determinamos la incertidumbre usando coef. variación
print('Observando ambas pdfs, parece que en TP2 hay más incertidumbre')
print('El rango b-a de TP2 es 1.2 veces mayor que en TP1, una medida sencilla de incertidumbre')
print('Otra forma de verlo, es Var[TP2] = 1.2**2 * Var[TP1], por lo que hay más varianza en TP2')
print('Sin embargo, las medias son diferentes, aunque no demasiado, conviene calcular el coeficiente de variación:')

TP1_cv = TP1.std() / TP1.mean()
TP2_cv = TP2.std() / TP2.mean()

print('  - Coef. variación TP1:', TP1_cv)
print('  - Coef. variación TP2:', TP2_cv)
print('Es lógico que sea exactamente el mismo, ya que la std y la mean de TP2 son las de TP1 * 1.2')
print('Por tanto, en relativo a sus medias, la incertidumbre es igual en ambas paradas')
```

### c) Calcular la probabilidad de que, al mismo tiempo, el primer tramo se recorra en más de 130 minutos, qne en el segundo tramo se circule más rápido de lo esperado y que el tercer tramo se recorra como mínimo un 20% más rápido de lo esperado. Justifica la respuesta (1 punto)

```
# respuesta

print('Se pide calcular P(TC1>130 & TC2<93 & TC3<0.8*70)')
print('Esta probabilidad se obtiene como el producto al ser independientes')
print('P(TC1>130 & TC2<93 & TC3<0.8*70) = P(TC1>130) * P(TC2<93) * P(TC3<0.8*70) = p1*p2*p3')

# - Tramo C1 (200km) TC1 ~ normal (mu=100, sigma = 10)
mu = 100
sigma = 10
TC1 = stats.norm(loc = mu, scale = sigma)
p1 = 1-TC1.cdf(130)

# - Tramo C2 (140km) TC2 ~ normal (mu= 93, sigma =  5)
mu = 93
sigma = 5
TC2 = stats.norm(loc = mu, scale = sigma)
p2 = TC2.cdf(mu) # tiene que salir 0.5 !!

# - Tramo C3 (80km)  TC3 ~ normal (mu= 70, sigma = 15)
mu = 70
sigma = 15
TC3 = stats.norm(loc = mu, scale = sigma)
p3 = TC3.cdf(0.8*mu)

prob = p1*p2*p3
print('Las probabilidades son: p1=',p1, '  p2=',p2, '  p3=',p3)
print('La probabilidad pedida es: p1*p2*p3 = ', prob)
```

### e) Calcular la probabilidad de que el tiempo invertido en recorrer el tercer trayecto (TC3) sea superior al tiempo invertido en el segundo trayecto (TC2). Justificar la respuesta (2 puntos)

```
# respuesta

print('Hay que calcular P(TC3>TC2), es decir, P(TC3-TC2>0)')
print('Definimos DTC32: v.a. diferencia entre TC3 y TC2 (en min)')
print('DTC32 ~ Normal(mu,sigma) ya que es la diferencia de dos normales (reproductividad)')
print('Los param mu y sigma se pueden determinar fácilmente considerando que TC3-TC2 = TC3 + (-1)*TC2')
print('La v.a. (-1)*TC2 (entendida como a+b*X) tiene como media -1*mu(TC2) y '
      'como varianza (-1)^2 * var(TC2) = var(TC2)')
print('Por tanto TC3+(-TC2) tiene:')
print(' > media mu(TC3)-mu(TC2)')
print(' > varianza var(TC3)+var(TC2)')

# calculamos
DTC32_mean = TC3.mean() - TC2.mean()
DTC32_var = TC3.var() + TC2.var()
mu = DTC32_mean
sigma = np.sqrt(DTC32_var) # ojo !!
DTC32 = stats.norm(loc = mu, scale = sigma)
print('DTC32 ~ Normal(mu=',DTC32_mean,', sigma=',sigma,')')

# la probabilidad pedida P(TC3-TC2>0)
prob = 1 - DTC32.cdf(0)
print('La probabilidad pedida P(TC3>TC2) = P(TC3-TC2>0) = 1 - cdf(0)=', prob)

# Representamos lo pedido
xs = np.linspace(DTC32.ppf(0.001), DTC32.ppf(0.999), 1000)
xs_pedido = np.linspace(0, DTC32.ppf(0.999), 1000)

fig = plt.figure(1, figsize=(6, 4))
plt.fill_between(xs, DTC32.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, DTC32.pdf(xs), color="orange", label = 'pdf(TC3-TC2)')
plt.fill_between(xs_pedido, DTC32.pdf(xs_pedido), color="black", alpha=0.3, label = 'P(TC3-TC2>0)')

plt.xlabel('TC3 - TC2 (min)')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```
