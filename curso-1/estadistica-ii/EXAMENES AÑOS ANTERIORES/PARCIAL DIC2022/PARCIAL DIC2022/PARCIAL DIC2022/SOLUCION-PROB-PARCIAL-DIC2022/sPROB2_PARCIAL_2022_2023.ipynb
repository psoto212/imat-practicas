Un pequeño hotel de ciudad cuenta con habitaciones perfectamente equipadas con elementos de calefacción para mantener una temperatura de confort óptima. El nivel de ocupación de un día puede ser bajo, medio o alto. Se han considerado las siguientes probabilidades:
- Probabilidad(ocupación BAJA)  = 0.3
- Probabilidad(ocupación MEDIA) = 0.5
- Probabilidad(ocupación ALTA)  = 0.2

La dirección del mismo, preocupada por la factura de calefacción, han modelado la relación diaria entre el consumo de calefacción y la temperatura mediante una normal bivariante según el nivel de ocupación del día:

Ocupación BAJA:
- T tiene media 5 y varianza 36
- C tiene media 85 y varianza 10
- La correlación lineal entre C y T vale -0.85

Ocupación MEDIA:
- T tiene media 5 y varianza 36
- C tiene media 100 y varianza 12
- La correlación lineal entre C y T vale -0.8

Ocupación ALTA:
- T tiene media 5 y varianza 36
- C tiene media 105 y varianza 18
- La correlación lineal entre C y T vale -0.7

En donde:
- T: Temperatura diaria del exterior del hotel (en ºC)
- C: Consumo diario de calefacción (en kWh)

Se supone que el nivel de ocupación no está condicionado ni por el consumo de calefacción ni por la temperatura y que los días son independientes unos de otros.

### Importamos todos los paquetes aquí por comodidad

```
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import seaborn as sns
```

### A continuación se crean los vectores aleatorios que representan, para cada uno de los tres niveles de ocupación considerados, la distribución conjunta del consumo de calefacción y la temperatura:

```
# Variables con las que hay que trabajar

# Distribución conjunta temporada baja
mu1 = [5, 85]
corr = -0.85
var_X1 = 36
var_X2 = 10
cov_X1_X2 = corr * (np.sqrt(var_X1) * np.sqrt(var_X2))
SIGMA1 = [[var_X1, cov_X1_X2], [cov_X1_X2, var_X2]] 
CT_baja = stats.multivariate_normal(mean=mu1, cov=SIGMA1)

# Distribución conjunta temporada media
mu2 = [5, 100]
corr = -0.8
var_X1 = 36
var_X2 = 12
cov_X1_X2 = corr * (np.sqrt(var_X1) * np.sqrt(var_X2))
SIGMA2 = [[var_X1, cov_X1_X2], [cov_X1_X2, var_X2]] 
CT_media = stats.multivariate_normal(mean=mu2, cov=SIGMA2)

# Distribución conjunta temporada alta
mu3 = [5, 105]
corr = -0.7
var_X1 = 36
var_X2 = 18
cov_X1_X2 = corr * (np.sqrt(var_X1) * np.sqrt(var_X2))
SIGMA3 = [[var_X1, cov_X1_X2], [cov_X1_X2, var_X2]] 
CT_alta = stats.multivariate_normal(mean=mu3, cov=SIGMA3)
```

### a) Representa en la misma gráfica, mediante curvas de nivel, las tres funciones de densidad conjuntas de los vectores aleatorios correspondientes a los tres niveles de ocupación considerados. (1 punto)

```
# Representación
# consejo: para que se vea bien utiliza en contourf alpha=0.3

# genera el mismo grid bidimensional para evaluar las 3 pdfs en 100x100 puntos (6 sigmas)
x = np.linspace(mu2[0]-3*np.sqrt(SIGMA3[0][0]), mu2[0]+3*np.sqrt(SIGMA3[0][0]), num=100) # x1
y = np.linspace(mu1[1]-3*np.sqrt(SIGMA3[1][1]), mu3[1]+3*np.sqrt(SIGMA3[1][1]), num=100) # x2
X, Y = np.meshgrid(x,y)

# evalua las pdfs en cada punto (x_i,y_j)
Z1 = np.zeros(X.shape)
Z2 = np.zeros(X.shape)
Z3 = np.zeros(X.shape)
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
        Z1[i,j] = CT_baja.pdf([X[i,j], Y[i,j]])
        Z2[i,j] = CT_media.pdf([X[i,j], Y[i,j]])
        Z3[i,j] = CT_alta.pdf([X[i,j], Y[i,j]])

fig = plt.figure(1, figsize=(7, 7))
plt.contourf(X, Y, Z1, cmap='viridis', alpha = 0.3) # alpha para controlar la transparencia y poder pintar varias
plt.contourf(X, Y, Z2, cmap='viridis', alpha = 0.3) # alpha para controlar la transparencia y poder pintar varias
plt.contourf(X, Y, Z3, cmap='viridis', alpha = 0.3) # alpha para controlar la transparencia y poder pintar varias
plt.grid()
plt.plot(mu1[0],mu1[1],'sr', label= f'Baja: {mu1}') # cuadrado rojo en el centro de la distribución
plt.plot(mu2[0],mu2[1],'sg', label= f'Media: {mu2}') # cuadrado rojo en el centro de la distribución
plt.plot(mu3[0],mu3[1],'sc', label= f'Alta: {mu3}') # cuadrado rojo en el centro de la distribución

plt.colorbar()
plt.xlabel("T");plt.ylabel("C")
plt.legend()
plt.show()
```

### b) Calcula, justificando la respuesta, el consumo esperado de calefacción en un día cualquiera. (2 puntos)

```
# Respuesta

print('Se pide E[C], siendo C la v.a. consumo de calefacción diario, con independencia del nivel de ocupación.')
print('La probabilidad de que la ocupación de un día sea baja, media o alta vale 0.3, 0.5 y 0.2, respectivamente.')
print('Por tanto: E[C] = 0.3 * E[C|B] +  0.5 * E[C|M] +  0.2 * E[C|A]')

# cálculo de los valores esperados condicionados por el nivel de ocupación
eCcondB = mu1[1]
eCcondM = mu2[1]
eCcondA = mu3[1]

print('\nLas esperanzas condicionales se obtienen directamente a partir de la distribución bivariante de cada nivel de ocupación:')
print(f'  E[C|B] = {eCcondB}')
print(f'  E[C|M] = {eCcondM}')
print(f'  E[C|A] = {eCcondA}')

pB = 0.3
pM = 0.5
pA = 0.2

eC = pB * eCcondB +  pM * eCcondM +  pA * eCcondA
print(f'El consumo esperado en un día cualquiera es: E[C] = 0.3 * E[C|B] +  0.5 * E[C|M] +  0.2 * E[C|A] = {eC} kWh')
```

### c) Calcula la probabilidad de que un día cualquiera se superen los 110 kWh de consumo de calefacción. Justifica la respuesta. (2 puntos)

```
# Respuesta

print('Se pide P(C>110), siendo C la v.a. consumo de calefacción diario, con independencia del nivel de ocupación.')
print('Dado que el nivel de ocupación representa una partición disjunta del espacio muestral, se puede usar el teorema de la probabilidad total:')
print('P(C>110) = P(C>110|B) P(B) +  P(C>110|M) P(M) +  P(C>110|A) P(A)')

# cálculo de las probabilidades condicionadas a partir de las marginales del consumo para cada nivel de ocupación
C1 = stats.norm(loc = mu1[1], scale = np.sqrt(SIGMA1[1][1]))
C2 = stats.norm(loc = mu2[1], scale = np.sqrt(SIGMA2[1][1]))
C3 = stats.norm(loc = mu3[1], scale = np.sqrt(SIGMA3[1][1]))

Prob_C1_sup110 = 1 - C1.cdf(110)
Prob_C2_sup110 = 1 - C2.cdf(110)
Prob_C3_sup110 = 1 - C3.cdf(110)

prob = Prob_C1_sup110 * pB +  Prob_C2_sup110 * pM +  Prob_C3_sup110 * pA

print(f'P(C>110) = {Prob_C1_sup110} * {pB} + {Prob_C2_sup110} * {pM} +  {Prob_C3_sup110} * {pA} = {prob}')
```

### d) Sabiendo que en un día de ocupación media el consumo de calefacción ha sido de 95 kWh, calcula la probabilidad de que la temperatura de dicho día fuese superior a 10 grados centígrados. Justifica la respuesta (1 punto)

```
# Respuesta

print('Se pide calcular P(T>10 | C=95 y ocupación MEDIA)')
print('Como la ocupación es media, se sabe que la distribución conjunta de (T,C) es la correspondiente a ese nivel de ocupación.')
print('Por tanto, para responder hay que determinar la distribución de la v.a. T|C=95, que será una normal ya que (T,C) se distribuye como una normal bivariante:')

# Condicionada X1 | X2 = x2
x2 = 95
mu = mu2
var_X1 = SIGMA2[0][0]
cov_X1_X2 = SIGMA2[0][1]
var_X2 = SIGMA2[1][1]

mu_X1cX2 = mu[0] + ((x2 - mu[1]) * cov_X1_X2 / var_X2)
sigma_X1cX2 = np.sqrt(var_X1 - (cov_X1_X2**2 / var_X2))
TcC95 = stats.norm(loc = mu_X1cX2, scale = sigma_X1cX2)

print(f'T|C={x2} ~ N(mu={mu_X1cX2},sigma={sigma_X1cX2})')
print('A partir de esta distribución se puede calcular P(T>10)')

prob = 1 - TcC95.cdf(10)
print(f'Finalmente, P(T>10 | C=95 y ocupación MEDIA) = {prob}')
```

### e) El coste de calefacción de un cuatrimestre (120 días) se calcula sumando dos términos (uno fijo mas otro variable). El término fíjo son 200€, y el término variable viene dado por el producto de los kWh consumidos en el cuatrimestre por el precio de la tarifa contratada, que es 0.311 €/kWh. Calcula el coste de la factura de la calefacción que tendrá que pagar el hotel como mínimo el 75% de los cuatrimestres en los que todos los días sean de ocupación media. Justifica la respuesta (2 puntos) 

```
# Respuesta

print('Hay que calcular el percentil 25 de la distribución de probabilidad del coste de calefacción de la temporada.')
print('Sea F: el coste de la factura del cuatrimestre.')
print('F = 200 + 0.311 * S, donde S es el consumo de calefacción acumulado de 120 días de ocupación media.')
print('Por tanto, S es una v.a. resultado de sumar de 120 v.a. que son i.i.d.')
print('Según el teorema central del límite, la v.a. S se distribuye como una normal de parámetros:')

mu_S  = 120 * mu2[1]
var_S = (120**2) * SIGMA2[1][1]
print(f'  - La media de S es: mu_S = 120 * {mu2[1]}  = {mu_S} ')
print(f'  - La varianza de S es: var_S = 120**2 * {SIGMA2[1][1]} = {var_S}')

print('\nLa v.a. de interés F también es una normal al ser una transformación lineal de S:')

mu_F  = 200 + 0.311 * mu_S
var_F = (0.311**2) * var_S
print(f'  - La media de F es: mu_F = 200 + 0.311 * mu_S = {mu_F}')
print(f'  - La varianza de F es: var_F = 0.311**2 * var_S = {var_F}')

F = stats.norm(loc = mu_F, scale = np.sqrt(var_F))
p25_F = F.ppf(.25)
print(f'Para calcular el percentil 25 usamos la cdf inversa de F~N(mu={mu_F},std = {np.sqrt(var_F)})')
print(f'Por tanto, la factura de calefacción superará los {p25_F} € el 75% de los cuatrimestres de ocupación media.')
```
