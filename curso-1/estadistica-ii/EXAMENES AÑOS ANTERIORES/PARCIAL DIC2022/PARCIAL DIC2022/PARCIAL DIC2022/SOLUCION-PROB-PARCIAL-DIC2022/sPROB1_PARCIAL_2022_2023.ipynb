### a) ¿Cuál es la probabilidad de que en un día cualquiera no acuda ningún hombre al gimnasio? Justifica la respuesta. (0.5 puntos)

```
# Respuesta

# La información que tenemos es
# Chicas sigue una Poisson con lamda=7; número medio de chicas que acuden diariamente al gym
# Chicos sigue una Poisson con lamda=3; número medio de chicos que acuden diariamente al gym
# Se suponen independientes

lmda_chicas= 7
X_chicas = stats.poisson(lmda_chicas)

lmda_chicos= 3
X_chicos = stats.poisson(lmda_chicos)

#comprueba
print('media', X_chicos.mean())
print('var', X_chicos.var())

print('Se nos pide calcular la Pr(X_chicos=0)')
k=0
print('La probabilidad de que en un dia no acuda ningún chico es:', X_chicos.pmf(k))
```

### b) ¿Cuál es la probabilidad de que acudan menos de tres mujeres? Justifica la respuesta. (0.5 puntos)

```
# Respuesta

print('Se nos pide calcular la Pr(X_chicas<3)=Pr(X_chicas<=2)')
k=2
print('La probabilidad de que en un dia acudan menos de tres mujeres es:', X_chicas.cdf(k))
```

### c) Hallar la probabilidad de que en un día determinado acuda alguna persona al gimnasio. Justifica la respuesta. (1 punto) 

```
# Respuesta

print('Se nos pide Pr(X_chicos>=1 ó X_chicas>=1) = Pr(X_chicos>=1) + Pr(X_chicas>=1) - Pr(X_chicos>=1 y X_chicas>=1)')
print('Como dichas variables son independientes, dicha probabilidad es finalmente')
print('Pr(X_chicos>=1 ó X_chicas>=1) = Pr(X_chicos>=1) + Pr(X_chicas>=1) - Pr(X_chicos>=1) * Pr(X_chicas>=1)')

print('La probabilidad de que en un día determinado acuda alguna persona es:', (1-X_chicos.cdf(0))+(1-X_chicas.cdf(0))- (1-X_chicos.cdf(0))*(1-X_chicas.cdf(0)) )

# Respuesta alternativa

print('Respuesta alternativa usando la reproductivdad de la poisson (S=X_chicos+X_chicas):')
S = stats.poisson(lmda_chicas + lmda_chicos)
prob = 1 - S.cdf(0)
print(prob)
```

### d) Si se sabe que el número de mujeres fue menor que 3, ¿cuál es la probabilidad de que hayan acudido más mujeres que hombres? Justifica la respuesta. (2 puntos)

```
# Respuesta

print('Se pide calcular la siguiente probabilidad:Pr(X_chicas>X_chicos/X_chicas<3)')
print('Pr(X_chicas>X_chicos/X_chicas<3) = Pr((X_chicas>X_chicos) y X_chicas<3)/Pr(X_chicas<3)')
print('La probabilidad del numerador es Pr(X_chicas=1 y X_chicos=0)+Pr(X_chicas=2 y X_chicos=0)+Pr(X_chicas=2 y X_chicos=1)')
print('Como son variables independientes, la probabilidad del numerador se puede escribir como:')
print ('Pr(X_chicas=1)*Pr(X_chicos=0)+Pr(X_chicas=2)*Pr(X_chicos=0)+Pr(X_chicas=2)*Pr(X_chicos=1)')
Pr_numerador=X_chicas.pmf(1)*X_chicos.pmf(0)+X_chicas.pmf(2)*X_chicos.pmf(0)+X_chicas.pmf(2)*X_chicos.pmf(1)
Pr_denominador=X_chicas.cdf(2)
print('La probabilidad solicitada es:', Pr_numerador/Pr_denominador)
```

### g) El gimnasio tiene tres tipos de elípticas diferentes (A, B y C). Hay un 70% del modelo A, un 20% del modelo B y el resto es del modelo C. Las elípticas del modelo A tienen una probabilidad de fallar del 1%, las del modelo B del 0.1% y las del modelo C tan solo del 0.05%. Si hay una elíptica que está fallando, ¿cuál es la probabilidad de que sea una elíptica del modelo B? (2 puntos)

```
# Respuesta

# Tenemos que:
prA=0.7
prB=0.2
prC=1-prA-prB

prFconA=0.01
prFconB=0.001
prFconC=0.0005

print('Se pide calcular P(B|F), y se conoce P(A), P(B) y P(C), así como P(F|A), P(F|B) y P(C|F).')
print('Los modelos A, B y C forman una partición disjunta del espacio muestral a considerar. '
      'Usamos Bayes para calcular la probabilidad a posteriori P(B|F) a partir de las verosimilitudes y las prob. a priori (todas conocidas).')
print('Primero calculamos la probabilidad de que una elíptica falle usando el teorema de la prob. total:')

prAyF = prFconA * prA
prByF = prFconB * prB
prCyF = prFconC * prC
prF = prAyF + prByF + prCyF
print(f'P(F) = P(A y F)  + P(B y F) + P(C y F)  = {prF}')

prob =  prByF/prF
print('Usando Bayes:')
print(f'P(B|F) = P(B y F) / P(F) = P(B y F) / P(F) = {prByF}/{prF} = {prob}')
```
