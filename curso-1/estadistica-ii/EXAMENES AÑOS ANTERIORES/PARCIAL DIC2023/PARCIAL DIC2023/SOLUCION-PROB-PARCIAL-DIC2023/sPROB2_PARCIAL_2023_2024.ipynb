### a) Crea todas las variables aleatorias completando el código siguiente. Para evitar confusiones, utiliza como nombres de las variables (X1z, X2z), donde z puede ser A, B o C. (0.5 puntos)

```
# Tipo Z = A

# X1 se distribuye como una binomial de parámetros n = 10 y p = 0.9
n = 10
p = 0.9
X1A = stats.binom(n=n,p=p)

# X2 se distribuye como una normal de media 410 y desviación típica 7
muX2A = 410
sigmaX2A = 7
X2A = stats.norm(loc = muX2A, scale = sigmaX2A)
```

```
# Tipo Z = B

# X1 se distribuye como una binomial de parámetros n = 30 y p = 0.9
n = 30
p = 0.9
X1B = stats.binom(n=n,p=p)

# X2 se distribuye como una triangular simétrica entre 410 y 430
a = 410 # min
b = 430 # max
c = a + 0.5*(b-a) # moda
print(f'Simetría implica que la moda está en el centro: a + 0.5*(b-a) =  {c}')

loc = a
escalado = b-a
forma = (c-a)/escalado
X2B = stats.triang(loc=loc, scale=escalado, c = forma)
```

```
# Tipo Z = C
# X1 tiene media 50 y varianza 5
# X2 tiene media 435 y varianza 14
# La covarianza entre X1 y X2 vale 4.71

muC = [50, 435]
var_X1C = 5
var_X2C = 14
covC_X1_X2 = 4.71
SIGMA3 = [[var_X1C, covC_X1_X2], [covC_X1_X2, var_X2C]] 
X1X2C = stats.multivariate_normal(mean=muC, cov=SIGMA3)
```

### f) Se ha ejecutado una red neuronal que tenía más de 25 capas y menos de 450 mil parámetros. Calcula la probabilidad de que se tratara de un problema de tipo B. Justifica la respuesta. (2.5 puntos)

```
# Respuesta

print('Se pide calcular la probabilidad condicionada P(B | X1>25 y X2<450).')
print('Hay que utilizar el Teorema de Bayes para calcular la probabilidad pedida.')
print('Los tres tipos de problemas A, B y C forman una partición disjunta del espacio muestral a considerar.')

print('Primero utilizamos el teorema de la probabilidad total para calcular P(X1>25 y X2<450) = P(Y)')
print('P(Y|A) =  P(X1A>25 y X2A<450) =  (1-P(X1A<=25)) * P(X2A<450) ya que X1A y X2A son independientes.')
print('P(Y|B) =  P(X1B>25 y X2B<450) =  (1-P(X1B<=25)) * P(X2B<450) ya que X1B y X2B son independientes.')
print('P(Y|C) =  P(X1C>25 y X2C<450) =  F(inf, 450) - F(25, 450), requiere usar la distribución conjunta.')

# calculamos las probabilidades en cada caso
Pr_Y_condA = (1-X1A.cdf(25)) * X2A.cdf(450)
Pr_Y_condB = (1-X1B.cdf(25)) * X2B.cdf(450)
Pr_Y_condC = X1X2C.cdf(np.array([np.inf,450]))  -  X1X2C.cdf(np.array([25,450]))

print('\nLas verosimilitudes son:')
print(f'P(Y|A) = P(X1A>25) * P(X2A<100) = {Pr_Y_condA}')
print(f'P(Y|B) = P(X1B>25) * P(X2B<100) = {Pr_Y_condB}')
print(f'P(Y|C) = F(inf, 450) - F(25, 450) = {Pr_Y_condC}')

pA = 0.5
pB = 0.3
pC = 0.2

Pr_Y = pA * Pr_Y_condA +  pB * Pr_Y_condB +  pC * Pr_Y_condC
print(f'P(X1>25 y X2<450) = P(Y) = P(A) * P(Y|A) +  P(B) * P(Y|B) +  P(C) * P(Y|C) =  {Pr_Y}')

print('\nUtilizando Bayes se determina la probabilidad pedida:')
prob = (pB * Pr_Y_condB) / Pr_Y 

print('Por tanto, la probabilidad de que se tratara de un problema de tipo B es:')
print(f'P(B | X1>25 y X2<450) = P(B) * P(Y|B) / P(Y) = {pB * Pr_Y_condB} / {Pr_Y}  = {prob}')
```
