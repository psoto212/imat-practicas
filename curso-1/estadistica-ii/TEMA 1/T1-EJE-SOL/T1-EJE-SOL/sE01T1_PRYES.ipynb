## Medidas de tendencia central

np.mean, np.median, stats.mode

```
# import required packages
import numpy as np
from scipy import stats
```

### a) Calcula la media aritmética, mediana y moda de los datos almacenados en la variable x.

```
# datos
x = [1,2,2,2,3,20000]
```

```
# respuesta

me = np.mean(x)
md = np.median(x)
mo = stats.mode(x, keepdims = False)

print('x= ',x)
print('mean(x)= ', me)
print('median(x)= ', md)
print('mode(x)= ', mo) # moda y número de repeticiones
# mo[0]
# print('mode(x)= ', mo[0])
```

### f) Calcula la media aritmética y moda de los datos almacenados en la variable x. Observa los datos de partida en la variable x. Explica por qué se obtienen esa media y esa moda.

```
# datos con "not a number" (NaN) e infinito (inf). Se puede operar con ellos

NaN = np.nan
infinity = np.inf

print(NaN)
print(infinity)

# datos
x = [NaN, 1*NaN, 2+NaN, infinity, infinity, infinity, infinity]
```

```
# respuesta

# Your guess:
#   mean =
#   mode = 

me = np.mean(x)
mo = stats.mode(x, keepdims = False)

print('x= ',x)
print('mean(x)= ', me)
print('mode(x)= ', mo) # moda y número de repeticiones

print('La media es NaN porque basta tener un NaN en x para que el valor esté indeterminado. '
      'Sin embargo, la moda vale inf porque es el valor más frecuente, aunque existan 3 NaNs en el vector (son menos que 4 infs).')
```
