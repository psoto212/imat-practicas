## Gráfico de violín (Violinplot)

```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['figure.dpi'] = 75
```

### b) Representa en la misma figura el histograma, el gráfico de cajas y el gráfico de violín de los datos almacenados en la variable x. Asegúrate de que los dos gráficos compartan el eje x para una correcta interpretación. ¿En qué gráfico es más sencillo ver el mínimo y máximo de los datos?¿Y la media?

```
# datos
x=[1,1,2,5,6,6,8,8,8,8,8,9,9,10,11]
print('x=',x) 
print('Median(x)=', np.median(x), '| Mean(x)=', np.mean(x))
```

```
# respuesta

# dibuja en un mismo gráfico el histograma, el boxplot y el gráfico de violín
fig = plt.figure(1, figsize=(8, 9)) 

# histograma
ax1= plt.subplot(3,1,1)
sns.histplot(data=x)
plt.title('Histograma')

# cajas
ax2=plt.subplot(3,1,2, sharex=ax1)
sns.boxplot(data=x, orient='y')
plt.title('Diagrama de cajas')

# violin
ax3=plt.subplot(3,1,3, sharex=ax1)
sns.violinplot(data=x, orient='y')
sns.rugplot(data=x, color='r')
plt.xlabel('x')
plt.title('Gráfico de violín')

plt.show()

print('El mínimo y máximo de los datos se observa muy bien en el diagrama de cajas, lógicamente también en la versión compacta del mismo que aparece en el diagrama violín.')
print('La media no se observa directamente en ninguno de los tres gráficos.')
```

### c) Se dispone de los datos almacenados en x. Se quiere saber, sin realizar ningún cálculo, si la distribución es simétrica y si es bimodal. Elige los gráficos apropiados que te permitan responder fácilmente a las dos cuestiones. Justifica la elección del gráfico utilizado.

```
# datos
x=[1,1,1,2,2,4,4,4,  180,190,200,200,200]
print('x=', x)
```

```
# respuesta
 
# dibuja el histograma y el boxplot
fig = plt.figure(1, figsize=(8, 7)) 

# pintamos el historgrama para observar la forma de la distribución (posible bimodalidad)
ax1= plt.subplot(2,1,1)
sns.histplot(data=x)
plt.title('Histograma')

# gráfico de cajas (para ver mejor la simetría)
ax2=plt.subplot(2,1,2, sharex=ax1)
sns.boxplot(data=x, orient='y')
plt.title('Diagrama de cajas')
plt.xlabel('x')
plt.show()

print('Según el histograma, se puede deducir que hay cierta asimetría hacia la derecha. '
      'Además, se observa que los datos están concentrados en dos puntos del espacio, sugiriendo una posible distribución bimodal.')
print('En el gráfico de cajas se observa fácilmente la asimetría hacia la derecha, pero este gráfico no sirve para observar si es bimodal, '
      'no permite ver si hay varias modas, la propia representación asume una distribución unimodal.')
```

### d) Se ha representado el histograma de los datos almacenados en la variable x. Únicamente observando dicho gráfico, indica si en un gráfico de cajas habrá o no datos atípicos. ¿Cómo será el gráfico de violín alrededor del cero, suave o apuntado?. Confirma tus respuestas creando dichos gráficos.

```
# datos
x = (np.abs(np.random.randn(5000)))**1.7

# histograma
fig = plt.figure(1,figsize=(8,3))
sns.histplot(data=x)
plt.xlabel('x')
plt.show()
```

```
# respuesta

print('Se observa una distribución muy asimétrica, que dará lugar en un gráfico de cajas a muchos atípicos a la derecha por existir en esa '
      'zona bastantes puntos muy alejados del rango intercuartílico, situado en los valores bajos.')

print('En la zona del 0, un gráfico de violín será apuntado ya que según el histograma hay, comparada con el resto, una alta concentración de puntos.')

# dibuja el diagrama de cajas y violin para confirmar
fig = plt.figure(1, figsize=(8, 7)) 

# pintamos el historgrama para observar la forma de la distribución (posible bimodalidad)
ax1= plt.subplot(2,1,1)
sns.boxplot(data=x, orient='y')
plt.title('Diagrama de cajas')

# gráfico de cajas (para ver mejor la simetría)
ax2=plt.subplot(2,1,2, sharex=ax1)
sns.violinplot(data=x, orient='y')
sns.rugplot(data=x, color='r')
plt.title('Diagrama de violin')
plt.xlabel('x')
plt.show()

print('En el gráfico de cajas se observa una fuerte asimetría a la derecha, con bastantes datos atípicos, siendo llamativo el valor próximo a 12.')
print('Efectivamente, en el violin se obtiene una forma similar a la observada en el histograma, pero bastante suavizada. '
      'En el zona del cero este gráfico indica que hay muchos datos.')
```

### e) Se ha representado el violin de los datos almacenados en la variable x. Indica, de forma aproximada y únicamente observando dicho gráfico, cuál es el intervalo que contiene el 25% de los valores más grandes. Confirma tu respuesta realizando el cálculo exacto.

```
# datos
x=[1,1,2,3,4,4,6,7]

# violin
fig = plt.figure(1,figsize=(12,3))
ax = sns.violinplot(data=x, orient='y')
plt.grid()
plt.show()
```

```
# respuesta

print('El 25% pedido se intuye fácilmente en la versión compacta del gráfico de cajas encapsulado en el gráfico violin. '
      'El intervalo comienza en el tercer cuartil y termina en el valor máximo, que podría decirse es el intervalo aproximado [4.5, 7].')

# cálculo exacto
q=0.75 # entre 0 y 1
qtle = np.quantile(x,q)
max_x = max(x)
print('quantil(x),', q,')=',qtle)
print('máximo valor', max_x)
print(f'El intervalo que contiene el 25% de los valores más grandes es [{qtle}, {max_x}], con ancho {max_x - qtle}.')
print('Justo el que habíamos intuido en el violin.')
```

### f) Se ha representado el violin de los datos almacenados en la variable x. Observando dicho gráfico, indica si la distribución de los datos tiene una forma leptocúrtica o platicúrtica. Corrobora tu respuesta cálculando una media de forma apropiada y añadiendo otro gráfico interesante que también permita observar dicha forma.

```
# datos
x = np.random.rand(1000)

# violin
fig = plt.figure(1,figsize=(12,3))
ax = sns.violinplot(data=x, orient='y')
plt.xlabel('x')
plt.show()
```

```
# respuesta

print('La distribución es claramente platicúrtica, plana. Usamos el exceso de curtosis para confirmarlo.')
      
# calculamos apuntamiento
kurto = stats.kurtosis(x)
print(f'El coeficiente de apuntamiento (exceso de curtosis) es {kurto}, negativo como era de esperar a no estar nada apuntada la distribución.')

# dibujar en un mismo gráfico el histograma, el boxplot y el gráfico de violín
fig = plt.figure(1,figsize=(8,5))
# histograma
ax1=plt.subplot(2,1,1)
sns.histplot(data=x)

# violinplot
plt.subplot(2,1,2, sharex = ax1)
sns.violinplot(data=x, orient='y')
plt.xlabel('x')

plt.show()

print('Con el histograma se observa claramente que la distribución no es nada apuntada, es plana (uniforme).')
```

### g) Representa el histograma y el violin de los datos almacenados en la variable x variando el nº de intervalos (variable nbins) y el ancho de banda (bandwidth), prueba con diferentes valores. ¿Qué ocurre si aumentamos el número de intervalos (bins) en el histograma? ¿Y qué ocurre si reducimos el parámetro del kernel en el violin (bandwidth)? Representa en la misma gráfica tres casos representativos para cada gráfico en los que se observe fácilmente lo indicado.

```
# datos
x = np.random.randn(1000)

nbins = 500
bandwidth = 0.1

# draw histogram and violin plot
fig = plt.figure(1,figsize=(8,5))
# histogram
ax1= plt.subplot(2,1,1)
sns.histplot(data=x, bins = nbins)
# violinplot
ax2=plt.subplot(2,1,2, sharex = ax1)
sns.violinplot(data=x, orient='y', bw_method=bandwidth)
plt.show()
```

```
# respuesta

print('Si aumentamos el número de bins el histograma se hace menos suave, es decir, más abrupto.')
print('Si reducimos el bandwidth el violín se hace menos suave, es decir, más abrupto.')

# figura con los casos pedidos
fig = plt.figure(1,figsize=(18,7))

# en la primera fila los 3 histogramas con diferentes nbins
ax1= plt.subplot(2,3,1)
sns.histplot(data=x, bins = 10)
plt.title('Nbins = 10')

plt.subplot(2,3,2, sharex = ax1)
sns.histplot(data=x, bins = 50)
plt.title('Nbins = 50')

plt.subplot(2,3,3, sharex = ax1)
sns.histplot(data=x, bins = 500)
plt.title('Nbins = 500')

# en la primera fila los 3 violines con diferentes bw
plt.subplot(2,3,4, sharex = ax1)
sns.violinplot(data=x, orient='y', bw_method=1)
plt.title('Bandwidth = 1')

plt.subplot(2,3,5, sharex = ax1)
sns.violinplot(data=x, orient='y', bw_method=0.1)

plt.title('Bandwidth = 0.1')

plt.subplot(2,3,6, sharex = ax1)
sns.violinplot(data=x, orient='y', bw_method=0.01)
plt.title('Bandwidth = 0.01')

plt.show()
```
