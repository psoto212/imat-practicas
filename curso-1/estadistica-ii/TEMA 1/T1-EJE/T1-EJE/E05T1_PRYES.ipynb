## Gráfico de barras y Boxplot

```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
plt.rcParams['figure.dpi'] = 75
```

### a) Representa el gráfico de barras de los datos almacenados en la variable x.

```
# data
value = ['C', 'Java', 'Python']
count = [23, 35, 29]
```

```
# respuesta

fig = plt.figure(1, figsize=(8, 4)) # permite indicar el nº de la figura y las dimensiones (ancho y alto)
plt.bar(value, count)
plt.title('Diagrama de barras')
plt.xlabel('x')
plt.ylabel('Frecuencias')
plt.show()
```

### b) Representa en la misma figura el histograma y el gráfico de cajas de los datos almacenados en la variable x. Asegúrate de que los dos gráficos compartan el eje x para una correcta interpretación. Añade en el gráfico de cajas el valor de la media de x para entender un poco mejor la representación. ¿Cómo es la asimetría de x?

```
x=[1,1,2,5,6,6,8,8,8,8,8,9,9,10,11]
print(x) 
print('Mediana(x)=', np.median(x), '| Media(x)=', np.mean(x))
```

```
# respuesta

# pintamos el historgrama y el boxplot juntos
fig = plt.figure(1,figsize=(8,8))

# histograma
ax1= plt.subplot(2,1,1)
sns.histplot(data=x)
sns.rugplot(data=x, color='r')

# boxplot
plt.subplot(2,1,2, sharex=ax1) # para asegurar que se comparte el eje x de la gráfica anterior <--------
sns.boxplot(data=x, orient='y')
plt.plot(np.mean(x),0,'ks')  # añadimos la media al boxplot <--------
plt.xlabel('x')

plt.show()

print('Según el gráfico de cajas, la variable x es claramente asimétrica a la izquierda. Por eso la media está a la izquierda de la mediana. '
      'En el histograma se observa de una forma menos clara.')
```
