## Medidas estadísticas para la combinación lineal de variables

```
# import required packages
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 75
```

### a) Se dispone de los datos almacenados en la variable x. Crea una nueva variable a partir de x realizando el cambio de origen y escala siguiente: y = 10 + 2 x. Calcula la media y la varianza de los datos en x e y, relacionando los valores a partir de la transformación lineal realizada. Representa en la misma figura el histograma de x e y por un lado, y sus diagramas de cajas por otro lado, asegurando que ambas gráficas comparten el eje horizontal para asegurar su correcta interpretación.

```
# datos
n=5000
rng = np.random.default_rng(999) # fijamos la semilla al valor 999
x = rng.normal(loc=1.0, scale=1.0, size=n)
print('x: ', x[0:5], '...')
```

```
# respuesta

# crear la nueva variable
origin = 10
scale = 2
y = origin + scale * x
print(f'y = {origin} + {scale}*x')

# medidas principales para ambas variables
print('mean(x)=',np.mean(x))
print('mean(y)=',np.mean(y), '| origin + scale * mean(x) = ', origin + scale * np.mean(x))

print('var(x)=',np.var(x))
print('var(y)=',np.var(y), '| scale^2 * var(x) = ', (scale**2)  * np.var(x))

print('Por tanto, se comprueba que los datos en y tienen la media y varianza acorde con la teoría.')

# gráfico de la variable original y de la nueva variable
fig = plt.figure(1, figsize=(10, 4)) # permite indicar el nº de la figura y las dimensiones (ancho y alto)
ax1 = plt.subplot(2,1,1)
sns.histplot(x, bins = 20, color='blue',alpha=0.5, label='x')
sns.histplot(y, bins = 20, color='red',alpha=0.5, label='y')
plt.legend()
plt.subplot(2,1,2, sharex = ax1)
sns.boxplot(data=x, orient='y', color='blue')
sns.boxplot(data=y, orient='y', color='red')
plt.show()
```

### b) Se dispone de los datos almacenados en las variables x e y. Crea una nueva variable a partir de x e y según la siguiente suma ponderada: z = 5 x + 3 y. Calcula la media y la varianza de los datos en las tres variables, relacionando todos los valores a partir de la transformación lineal realizada. Representa en la misma figura el histograma de las tres variables por un lado, y sus diagramas de cajas por otro lado, asegurando que ambas gráficas comparten el eje horizontal para asegurar su correcta interpretación.

```
# datos
n=5000
rng = np.random.default_rng(999) # fijamos la semilla al valor 999
x = rng.normal(loc=1.0, scale=0.5, size=n)
y = rng.uniform(low=4.0, high=7.0, size=n)
```

```
# respuesta

# crear una nueva variable
alpha = 5
beta = 3
z = alpha * x + beta * y
print(f'z= {alpha}*x + {beta}*y')

# principales medidas para ambas variables
print('mean(x)=',np.mean(x))
print('mean(y)=',np.mean(y))
print('mean(z)=',np.mean(z), '| alpha * x + beta * y = ', alpha * np.mean(x) + beta * np.mean(y))

print('var(x)=',np.var(x))
print('var(y)=',np.var(y))
cov_x_y = np.cov(x,y)
print('var(z)=',np.var(z),'| alpha^2 * var(x) + beta^2 * var(y) + 2 * alpha*beta* cov(x,y) = ', (alpha**2)*np.var(x) + (beta**2)*np.var(y) + 2*alpha*beta*cov_x_y[0,1])
print('La cov(x,y) es nula ya que no hay relación entre las variables x e y.')

# gráfico de la variable original y de la nueva variable
fig = plt.figure(1, figsize=(10, 4)) # permite indicar el nº de la figura y las dimensiones (ancho y alto)
ax1= plt.subplot(2,1,1)
ax = sns.histplot(x, bins = 20, color='blue',alpha=0.5, label='x')
ax = sns.histplot(y, bins = 20, color='red',alpha=0.5, label='y')
ax = sns.histplot(z, bins = 20, color='green',alpha=0.5, label='z')
plt.legend()
plt.subplot(2,1,2, sharex = ax1)
sns.boxplot(data=x, orient='y', color='blue')
sns.boxplot(data=y, orient='y', color='red')
sns.boxplot(data=z, orient='y', color='green')
plt.show()
```

### c) Se dispone de los datos almacenados en la variable x. Crea una nueva variable z que sea la variable x estandarizada. Calcula la media y la varianza de los datos en las dos variables, relacionando todos los valores a partir de la transformación realizada. Representa en la misma figura el histograma de las dos variables por un lado, y sus diagramas de cajas por otro lado, asegurando que ambas gráficas comparten el eje horizontal para asegurar su correcta interpretación.

```
# datos
n=500
rng = np.random.default_rng(999) # fijamos la semilla al valor 999
x = rng.normal(loc=10.0, scale=2, size=n)
```

```
#respuesta

# crear una nueva variable, z, la versión estandarizada de la variable original x
mean = np.mean(x)
std = np.std(x)

z = (x - mean)/std
print('z= (x - ',  mean, ')/', std)

# principales medidas para ambas variables
print('La media de z tiene que ser nula y su varianza 1:')
print('mean(x)=',np.mean(x))
print('mean(z)=',np.mean(z))
print('var(x)=',np.var(x))
print('var(z)=',np.var(z))

# gráfico de la variable original y de la nueva variable
fig = plt.figure(1, figsize=(10, 4)) # permite indicar el nº de la figura y las dimensiones (ancho y alto)
ax1 = plt.subplot(2,1,1)
ax = sns.histplot(x, bins = 20, color='blue',alpha=0.5, label='x')
ax = sns.histplot(z, bins = 20, color='green',alpha=0.5, label='z')
plt.legend()
plt.subplot(2,1,2, sharex = ax1)
sns.boxplot(data=x, orient='y', color='blue')
sns.boxplot(data=z, orient='y', color='green')
plt.show()
```

### g) Se tienen las variables x e y de medias y matriz de covarianza conocidas. Si se crease ahora la variable z = 3*x - 2*x, ¿qué media y varianza tendría? Justifica la respuesta indicando el cálculo a partir de las medias y matriz de covarianza de x e y. Finalmente crea la variable z a partir de los datos de las variables x e y, calculando su media y varianza para verificar el resultado.

```
#datos
n=5000
rng = np.random.default_rng(999) # set the seed is 999
x = rng.normal(loc=1.0, scale=np.sqrt(3), size=n)
y = rng.normal(loc=10.0, scale=np.sqrt(2), size=n)

# principales medidas para ambas variables
x_med = np.mean(x)
x_var = np.var(x)

y_med = np.mean(y)
y_var = np.var(y)

xy_covar  = np.cov(x,y)[0][1]

print('mean(x)=', x_med, '| var(x)=', x_var)
print('mean(y)=', y_med, '| var(y)=', y_var)
print('matriz cov(x,y)=\n',np.cov(x,y))
print('cov(x,y)=',xy_covar)
```
