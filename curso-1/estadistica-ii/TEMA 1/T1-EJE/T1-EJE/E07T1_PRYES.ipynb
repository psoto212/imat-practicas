## Nube de puntos (scatterplot) e Histograma-2D

sns.relplot, sns.displot, sns.jointplot

```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
plt.rcParams['figure.dpi'] = 75
```

### a) Representa el gráfico de nube de puntos de los datos almacenados en las variables x e y. Representa el mismo gráfico usando las librerías plt y sns.

```
# datos
x=[1,2,5,6,6,8,8,8,8,8,9,9,10,11,1]
y=[3,4,5,6,6,6,3,3,3,3,2,2,1,1,10]
print('x: ', x)
print('y: ', y)
```

```
# respuesta

# dibuja la nube de puntos de x e y usando la librería plt
fig = plt.figure(1, figsize=(5, 5))
plt.plot(x, y,'.', color='r')
plt.title('Scatterplot (x,y) -plt-')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()

# dibuja la nube de puntos de x e y usando la librería seaborn
sns.relplot(x=x, y=y, color='green')
plt.title('Scatterplot (x,y) -sns-')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()
```

### b) Representa el histograma en dos dimensiones de los datos almacenados en las variables x e y. Ajusta el nº de intervalos de cada variable para una correcta visualización.

```
# datos
x=[1,2,5,6,6,8,8,8,8,8,9,9,10,11,1]
y=[3,4,5,6,6,6,3,3,3,3,2,2,1,1,10]
print('x: ', x)
print('y: ', y)
```

```
# respuesta

nbins_x_y = (5, 10) # número de intervalos histograma en ejes x e y, ajustados para buena visualización

ax=sns.displot(x=x, y=y, bins=nbins_x_y, cbar=True)
plt.title('Histograma de (x,y)')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
