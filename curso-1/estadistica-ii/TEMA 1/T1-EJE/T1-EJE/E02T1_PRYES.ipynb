## Medidas de posición

np.percentile, np.quantile

```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 75
```

### a) Calcula el percentil 75 y el primer cuartil de los datos almacenados en la variable x. Observa que se dispone de un número impar de datos.

```
# datos (número impar de obs. )
x = [1.,2.,5., 30.,40.]
```

```
# respuesta

p = 75 # entre 0 y 100
prctle = np.percentile(x,p)

q=0.25 # entre 0 y 1
qtle = np.quantile(x,q)

print('x = ', x)
print(f'percentil(x, {p}) = {prctle}')
print(f'quantil(x, {q}) = Primer cuartil = {qtle}')

print(f'El primer cuartil también se puede calcular como el percentil 25: { np.percentile(x,25)}')
```

### b) Calcula el percentil 25 y el segundo cuartil de los datos almacenados en la variable x. Observa que ahora se dispone de un número par de datos, por lo que en el cálculo hay implícita algún tipo de interpolación. Representa gráficamente en la misma figura los 4 datos disponibles y los valores calculados para entender mejor el resultado obtenido.

```
# datos (nº par de obs. )
x = [1.,2.,6.,8.]
```

```
# respuesta

p = 25 # entre 0 y 100   #<----------------------
prctle = np.percentile(x,p)
#prctle2 = np.percentile(x,p, interpolation='lower')

q=0.5 # entre 0 y 1
qtle = np.quantile(x,q)

print('x = ', x)
print(f'percentil(x, {p}) = {prctle}') #<----------------------
#print(f'percentil2(x, {p}) = {prctle2}')
print(f'quantil(x, {q}) = segundo cuartil = {qtle}')

# pintar los datos y las medidas
plt.stem(x,np.ones(np.size(x)))
plt.plot(prctle,0,'ks',label='percentil 25')
plt.plot(qtle,0,'kd',label='2ndo cuartil')
plt.legend()
ax=plt.title('Comprender la interpolación de percentiles')
```
