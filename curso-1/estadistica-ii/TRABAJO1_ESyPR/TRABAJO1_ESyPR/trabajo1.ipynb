## Cargamos los datos con los que se trabaja

```python
nom_fi_datos = 'G05A_DATOS_DEM_P1.csv'
```

```python
df_orig = pd.read_csv(nom_fi_datos)

# parseamos la fecha (cadena) para que sea un datetime con formato dd/mm/yyyy 
df_orig.FECHA = pd.to_datetime(df_orig.FECHA, format='%d/%m/%Y') 
s=df_orig.loc[:,'FECHA']
df_orig['DATE'] =  s.dt.date
del df_orig['FECHA']

# añadimos el año-semana para luego poder trabajar cómodamente con el dataframe
ywd=s.dt.isocalendar()
yw = 1000*ywd['year'] + ywd['week'] # por ejemplo 2022032 es la semana 32 del año 2022
df_orig['ANNO-SEMANA'] = yw

print('Tamaño de df_orig con los datos cargados:', df_orig.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables):')
print(df_orig)
```

## DISTRIBUCIÓN:

```python
# DISTRIBUCIÓN HORA 12:
df_hora12 =  pd.DataFrame(list(zip(df_orig.CR12)))

# Tres ejes en vertical para sincronizar la variable x
fig = plt.figure(1, figsize=(9, 7)) # permite indicar el nº de la figura y las dimensiones (ancho y alto)

# histograma
ax1= plt.subplot(3,1,1)
sns.histplot(data=df_hora12)
plt.title('Distribución del coeficiente de reparto a las 12h')

# cajas
ax2=plt.subplot(3,1,2, sharex=ax1)
sns.boxplot(data=df_hora12, orient='horizontal')

# violin
ax3=plt.subplot(3,1,3, sharex=ax1)
sns.violinplot(data=df_hora12, orient='horizontal')
sns.rugplot(data=df_hora12, color='r')
plt.xlabel('Demanda')
plt.show()

# métricas básicas:
print(df_hora12.describe())

# coef. de simetría
ske = stats.skew(df_hora12)
print('Coef. de simetría(hora 12):', ske)

# coef. apuntamiento (incluye corrección con el ajuste de -3)
kur = stats.kurtosis(df_hora12, fisher=True)
print('Coef. de apuntamiento(hora12):', kur)

fig.savefig('distribución hora 12.pdf')
```
