### 4c) Identificación de outliers

```
conditionIrrad06 = df_orig_Irrad['IRRADH06'] <= 0.1
columna_irrad_06 = df_orig_Irrad[conditionIrrad06, 'IRRADH06']
columna_util_06 = df_orig_Util[conditionIrrad06, 'UTILH06']
```

```
outliers = pd.DataFrame()

for hora in ["00", "03", "06", "09", "12", "15", "18", "21"]:
    # Condición 1: Irradiación nula pero utilización no nula
    cond1 = (df_orig_Irrad[f'IRRADH{hora}'] == 0) & (df_orig_Util[f'UTILH{hora}'] > 0)
    
    # Condición 2: Utilización nula pero irradiación no nula
    cond2 = (df_orig_Irrad[f'IRRADH{hora}'] > 0) & (df_orig_Util[f'UTILH{hora}'] == 0)

    outliers_hora = df_orig_Irrad[cond1 | cond2].copy()
    outliers_hora['TramoHorario'] = hora
    outliers = pd.concat([outliers, outliers_hora], ignore_index=True)

# Contar cuántos outliers hay en cada mes de cada año
outliers['Año'] = outliers['FECHA'].dt.year
outliers['Mes'] = outliers['FECHA'].dt.month

# Agrupar por año, mes y tramo horario para contar outliers
outliers_count = outliers.groupby(['Año', 'Mes', 'TramoHorario']).size().reset_index(name='Count')

# Visualización de outliers por mes y tramo horario en un gráfico
plt.figure(figsize=(14, 10))
sns.barplot(x='Mes', y='Count', hue='TramoHorario', data=outliers_count, palette='viridis')
plt.title('Outliers por Mes y Tramo Horario', fontsize=16)
plt.xlabel('Mes', fontsize=14)
plt.ylabel('Cantidad de Outliers', fontsize=14)
plt.legend(title='Tramo Horario', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(range(12), meses, rotation=45)
plt.tight_layout()
plt.show()
```
