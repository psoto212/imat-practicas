## Cargamos los datos con los que se trabaja

```
nom_fi_datos_Irrad = 'G07B_DATOS_IRRAD.csv'
nom_fi_datos_Util = 'G07B_DATOS_UTIL.csv'
```

```
df_orig_Irrad = pd.read_csv(nom_fi_datos_Irrad)

# parseamos la fecha (cadena) para que sea un datetime con formato yyyy/mm/dd 
df_orig_Irrad['FECHA'] = pd.to_datetime(df_orig_Irrad['FECHA'], format='%Y-%m-%d')

s = df_orig_Irrad.loc[:,'FECHA']
df_orig_Irrad['FECHA'] =  s.dt.date

print('Tamaño de df_orig_Irrad con los datos cargados:', df_orig_Irrad.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables):')
print(df_orig_Irrad)
```

```
df_orig_Util = pd.read_csv(nom_fi_datos_Util)

# parseamos la fecha (cadena) para que sea un datetime con formato yyyy/mm/dd 
df_orig_Util['FECHA'] = pd.to_datetime(df_orig_Util['FECHA'], format='%Y-%m-%d')

s = df_orig_Util.loc[:,'FECHA']
df_orig_Util['FECHA'] =  s.dt.date

print('Tamaño de df_orig_Util con los datos cargados:', df_orig_Util.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables):')
print(df_orig_Util)
```

### 1. Visión general de las series

```
# auxiliares útiles para la representación
fechas = df_orig_Irrad.FECHA.tolist()
```

```
fig = plt.figure(1, figsize=(15,6)) # permite indicar el nº de la figura y las dimensiones (ancho y alto)
ax1 = plt.subplot(2,1,1)
plt.plot(fechas, df_orig_Irrad.IRRADH03, label='IRRADH03')
plt.plot(fechas, df_orig_Irrad.IRRADH06, label='IRRADH06')
plt.plot(fechas, df_orig_Irrad.IRRADH09, label='IRRADH09')
plt.title(nom_fi_datos_Irrad)
plt.legend()

plt.subplot(2,1,2,sharex= ax1)
plt.plot(fechas, df_orig_Util.UTILH03, label='UTILH03')
plt.plot(fechas, df_orig_Util.UTILH06, label='UTILH06')
plt.plot(fechas, df_orig_Util.UTILH09, label='UTILH09')
plt.title(nom_fi_datos_Util)
plt.tight_layout()
plt.legend()

plt.show()

fig.savefig('figura1.pdf')
```
