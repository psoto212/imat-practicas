```
nom_fi_datos_Irrad = 'G11A_DATOS_IRRAD.csv'
nom_fi_datos_Util = 'G11A_DATOS_UTIL.csv'

df_orig_Irrad = pd.read_csv(nom_fi_datos_Irrad)

# parseamos la fecha (cadena) para que sea un datetime con formato yyyy/mm/dd 
df_orig_Irrad['FECHA'] = pd.to_datetime(df_orig_Irrad['FECHA'], format='%Y-%m-%d')

s = df_orig_Irrad.loc[:,'FECHA']
df_orig_Irrad['FECHA'] =  s.dt.date

print('Tamaño de df_orig_Irrad con los datos cargados:', df_orig_Irrad.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables):')
print(df_orig_Irrad)
```

```
df_orig_Util = pd.read_csv(nom_fi_datos_Util)

# parseamos la fecha (cadena) para que sea un datetime con formato yyyy/mm/dd 
df_orig_Util['FECHA'] = pd.to_datetime(df_orig_Util['FECHA'], format='%Y-%m-%d')

s = df_orig_Util.loc[:,'FECHA']
df_orig_Util['FECHA'] =  s.dt.date

print('Tamaño de df_orig_Util con los datos cargados:', df_orig_Util.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables):')
print(df_orig_Util)
```

### Estudiar la distribución de los coeficientes de irradiación solar y utilización solar fotovoltaica correspondientes a las horas 12 y 18:

"""
Como podemos observar, la radiacion varia mucho dependiendo de los meses del año. En invierno podemos obversvar que las cifras son muy bajas, sobre todo a las 18h, en las cuales en invierno es practicamente nula. En cuanto a las diferencias entre las 12 y las 18h, las diferencias son significantes, vemos que las irradiación baja muchisimo a lo largo de las horas, cuando más emite es cuando el sol esta justo encima y los rayos llegan los mas perdpendicular posible, una vez el sol va cayendo el angulo va disminuyendo y la irradiación va disminuyendo.
Fundmentalmente cambia dependiendo del mes, en cuanto a la semana no hay grandes diferencias, depende mucho mas de la estación del año, no varia de un dia a otro. 
En lo referido a la utilización solar fotovoltaica en las horas indicadas, vemos que a las 18h es casi nula, no se aprovecha nada de esta radiación salvo en verano que emite algo. Referido a las 12h, se aprovecha mucho más, a lo largo de todo el año, siendo en verano mucho mayor la utilización aunque en invierno también se recoja algo.
Como hemos comentado ya, la radiación no depende del día de la semana sino de la estación del año y la hora, esto ocurre debido a la inclinación de la Tierra en relación con su órbita alrededor del sol, ya que los rayos llegan con un angulo mas vertical y recorren menos atmosfera. Además sabemos que en verano los días son mas largos y por lo tanto hay más horas de luz y por lo tanto más horas de emisión de radiación efectiva
"""

### Comparar las distribuciones de los distintos coeficientes de irradiación y utilización:

```
mat_cov = df_orig_Irrad.cov()                
print('Matriz covarianzas:')
print(np.round(mat_cov, 2)) # Redondeando a solo dos decimales para verlo mejor

# matriz de correlaciones creada directamente con el dataframe (df)
mat_corr = df_orig_Irrad.corr()                
print('Matriz correlaciones:')
print(np.round(mat_corr, 2)) # Redondeando a solo dos decimales para verlo mejor

# Muestra la matriz de correlaciones visualmente
fig = plt.figure(1,figsize=(5, 5))
sns.heatmap(mat_corr, annot=False) # si no se ven los textos con demasiadas variables, poner annot=False
plt.title('Matriz de correlaciones')
plt.show()
```

```

mat_cov = df_orig_Util.cov()                
print('Matriz covarianzas:')
print(np.round(mat_cov, 2)) # Redondeando a solo dos decimales para verlo mejor

# matriz de correlaciones creada directamente con el dataframe (df)
mat_corr = df_orig_Util.corr()                
print('Matriz correlaciones:')
print(np.round(mat_corr, 2)) # Redondeando a solo dos decimales para verlo mejor

# Muestra la matriz de correlaciones visualmente
fig = plt.figure(1,figsize=(5, 5))
sns.heatmap(mat_corr, annot=False) # si no se ven los textos con demasiadas variables, poner annot=False
plt.title('Matriz de correlaciones')
plt.show()
```

Como podemos observar en la matriz de correlaciones, en el caso de irradiación solo hay realción en las horas cercanas al medio día (9,12,15 horas) , en cuanto al resto de horas la correlacion en nula debido a que la irradiación en estas horas es nula.
Si hablamos de la utilidad, podemos ver que la correlación es mucho mayor ya que la utilidad en horas de noche es nula, por lo tanto tienen mucho parecido entre si, al igual que sucede en las horas de luz, fundamentalmente en las del mediodía, ya que es cuando mas irradiación se incide y por lo tanto la utilidad va a tener valores similares.
