# GENERACIÓN DATOS PARTE DESCRIPTIVA PROBLEMA 2 REDES NEURONALES

```
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import seaborn as sns
```

```
pA = 0.5
pB = 0.3
pC = 0.2
```

```


# X1 se distribuye como una binomial de parámetros n = 10 y p = 0.9
n = 10
p = 0.9
X1A = stats.binom(n=n,p=p)
# X2 se distribuye como una normal de media 500 y varianza 5
muX2A = 410
sigmaX2A = 7
X2A = stats.norm(loc = muX2A, scale = sigmaX2A)




# X1 se distribuye como una binomial de parámetros n = 30 y p = 0.9
n = 30
p = 0.9
X1B = stats.binom(n=n,p=p)

# X2 se distribuye como una triangular simétrica entre 410 y 430
a = 410 # min
b = 430 # max
c = a + 0.5*(b-a) # moda
print(f'Simetría implica que la moda está en el centro: a + 0.5*(b-a) =  {c}')

loc = a
escalado = b-a
forma = (c-a)/escalado
X2B = stats.triang(loc=loc, scale=escalado, c = forma)



muC = [50, 435]
var_X1C = 5
var_X2C = 14
covC_X1_X2 = 4.71
SIGMA3 = [[var_X1C, covC_X1_X2], [covC_X1_X2, var_X2C]] 
X1X2C = stats.multivariate_normal(mean=muC, cov=SIGMA3)
```

```
# semilla
np.random.seed(99)
```

```
n = 1000 

#PROBLEMA A
na = int(pA * n)
rvs1 = X1A.rvs(size = na)
rvs2 = np.round(X2A.rvs(size = na), 3)

x1 = np.transpose(rvs1)
x2 = np.transpose(rvs2)

prob = np.tile('A', na)
dic = {'n_capas':x1, 'n_param':x2, 'prob': prob}
d1 = pd.DataFrame(dic)
print(d1)

#PROBLEMA b
nb = int(pB * n)
rvs1 = X1B.rvs(size = nb)
rvs2 = np.round(X2B.rvs(size = nb), 3)

x1 = np.transpose(rvs1)
x2 = np.transpose(rvs2)

prob = np.tile('B', nb)
dic = {'n_capas':x1, 'n_param':x2, 'prob': prob}
d2 = pd.DataFrame(dic)
print(d2)
```

```
# problema C

nc = int(pB * n)
rvs = X1X2C.rvs(size = nc)
x = np.transpose(rvs)
prob = np.tile('C', nc)
dic = {'n_capas':np.round(x[0],0), 'n_param':np.round(x[1],3), 'prob': prob}
d3 = pd.DataFrame(dic)
```

```


df = pd.concat([d1, d2, d3])
df.to_csv('muestraEjecuciones.csv', index=False)
print('listo')
```
