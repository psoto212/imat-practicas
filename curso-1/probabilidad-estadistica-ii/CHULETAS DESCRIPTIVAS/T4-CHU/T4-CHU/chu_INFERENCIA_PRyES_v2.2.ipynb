## 1. Distribución Chi-cuadrado $\chi^2(k)$

## 4. Estimación por máxima verosimilitud

```
# Búsqueda del máximo de una función (aproximación numérica por fuerza bruta)

# puntos en los que evaluamos la función: [ valor_min_x ... valor_max_x], n_puntos equi-espaciados entre un mínimo y un máximo de búsqueda
valor_min_x = -10
valor_max_x = 20 
n_puntos = 10000
x_prueba = np.linspace(valor_min_x, valor_max_x, n_puntos)

# función que se quiere maximizar, evaluada en x_prueba !!
VA = stats.chi2(4)
func_eval_x = VA.pdf(x_prueba)

# barrido básico evaluando la función en todos los puntos x_prueba y cogiendo aquél en el que la función es máxima
# conviene acotar con cuidado la zona de búsqueda
index_max_func = np.argmax(func_eval_x)   # <--- encuentra el argumento que hace máximo la función
x_max_func = x_prueba[index_max_func]
max_func = func_eval_x[index_max_func]

print(f'El máximo está en la posición {index_max_func} del array x_prueba')
print(f'El valor de x que hace que la función sea máxima es aproximadamente: {x_max_func}')
print(f'El valor de la función en el máximo es aproximadamente: {max_func}')

# gráficamente
fig = plt.figure(1, figsize=(15, 5))
plt.plot(x_prueba, func_eval_x, color="orange", label = 'Función a maximizar')
plt.vlines(x_max_func, func_eval_x[0], max_func, color = 'green')
plt.plot(x_max_func, max_func,'o', color = 'green', label = f'Máximo = {(x_max_func, max_func)}')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid()
```

```
# Estimación por MÁXIMA VEROSIMILITUD

# ejemplo de muestra de datos para x
x = [1,2,4,5,3,6,5,8,5,5,6,9,10]

# definimos la función de verosimilitud logarítmica para maximizarla
    
def verosimilitud_log(parameters):
    # sacamos los parámetros
    param1, param2 = parameters
    
    # La función de verosimilitud logarítmica depende del problema y requiere
    # conocer la distribución de la v.a. bajo estudio, que dependerá de los params
    # a estimar por MV
    
    # Ejemplo estimando la media y desv. típica de una normal:
    logLikelihood = np.sum(stats.norm.logpdf(x, loc=param1, scale=param2))  # <--- logpdf es el log de la pdf, ya está preparado ...
    
    # Como la función minimiza y queremos los params que maximizan la verosimilitud
    # simplemente la multimplicamos por -1: max(loglike) = min(-loglike)
    neg_logLikelihood = -1 * logLikelihood
    return neg_logLikelihood 
```

```
# Maximizamos la función de verosimilitud logarítmica (minimizando -1 * funcion)
mle_model = opt.minimize(verosimilitud_log, np.array([1,1]), method='L-BFGS-B')

# cogemos los coefs estimados
param1 = mle_model.x[0]
param2 = mle_model.x[1]

print(mle_model)
print('Parámetros estimados por MV:\n'
      f'  - parámetro 1: {param1}\n'
      f'  - parámetro 2: {param2}')
```

## 4. Intervalos de confianza

```
# ejemplo de muestra de datos para x
x = [1,2,4,5,3,6,5,8,5,5,6,9,10]

# tamaño muestra
n = len(x)

# media muestral
x_media = np.mean(x)

# calculamos la desviación típica muestral sx sin sesgo (dividiendo por n-1)
sx = np.std(x, ddof=1)

# Ahora creamos la v.a. MEDIA MUESTRAL como una t de student
#    De la ayuda de scipy.stats.t :
#    To shift and/or scale the distribution use the loc and scale parameters. 
#    Specifically, t.pdf(x, df, loc, scale) is identically equivalent to t.pdf(y, df) / scale with y = (x - loc) / scale.
Xmedia = stats.t(n-1, loc=np.mean(x), scale = sx/np.sqrt(n))  

#Para calcular un IC de la media de X usamos la distribución teórica de la media muestral
gama = 0.90
alfa = 1 - gama
lim_inf = Xmedia.ppf(alfa/2)
lim_sup = Xmedia.ppf(1-alfa/2)

print(f'El intervalo al {100*gama}% de la media de X es: [{lim_inf}, {lim_sup}]')
```

```
# Gráficamente

# puntos para representar la distribución y el intervalo
xs = np.linspace(Xmedia.ppf(0.00001), Xmedia.ppf(1-0.00001), 1000)
xIC = np.linspace(lim_inf, lim_sup, 1000)

plt.figure(1, figsize=(15, 6))
plt.plot(xs, Xmedia.pdf(xs), linewidth = 4, label=f'Distribución Media muestral (según una t)')
plt.fill_between( xIC, Xmedia.pdf(xIC), color="green", alpha=0.3, label=f'Probabilidad asociada ($\gamma=${gama})')
plt.vlines(x_media, 0, np.max(Xmedia.pdf(xs)), color='black', label=f'Estimación puntual media {x_media}')
plt.errorbar(x_media, -0.01, xerr = np.array([[x_media - lim_inf], [lim_sup - x_media]]), fmt = 'o', label = f'IC [{lim_inf}, {lim_sup}] al {gama*100}%')
plt.title('Distribución del estimador media muestral de X e IC')
plt.tight_layout()
plt.legend()
plt.show()
```

## 5. Bootstrap

```
# ejemplo de muestra de datos para x
x = [1,2,4,5,3,6,5,8,5,5,6,9,10]

# params para llamar a bootstrap (método BCa, el que se suele usar normalmente)
muestra_orig = (x,)  # tiene que ser una secuencia
R = 9999
estadistico = np.mean
nivel_confianza = 0.90

res_BCa = stats.bootstrap(muestra_orig, estadistico, confidence_level = nivel_confianza,  n_resamples = R, method = 'BCA', random_state = 999)

# devuelve como resultado el IC y el SE del estadístico
lim_inf_BCa, lim_sup_BCa = res_BCa.confidence_interval
SE_BCa = res_BCa.standard_error

print(f'IC {100*nivel_confianza}% estimado según método BCa para la media: [{lim_inf_BCa}, {lim_sup_BCa}]')
print(f'Error estándar del estimador media muestral, estimado según método BCa: {SE_BCa}')
```

## 6. Contrastes de hipótesis paramétricos
### Se trabaja con una v.a. X que se distribuye como una Normal.

### 6.1 Contraste de hipótesis para la media de X (varianza conocida)
#### H0: $\mu_X = \mu_0 $
#### H1: $\mu_X \neq \mu_0 $ (bilateral: "two-sided")
#### H1: $\mu_X < \mu_0 $ (unilateral: "less")
#### H1: $\mu_X > \mu_0 $ (unilateral: "greater")

```
# ejemplo de muestra de datos para x
x = [1,2,4,5,3,6,5,8,5,5,6,9,10]

# varianza poblacional conocida
sigma2_pob =  2.1

# media poblacional que se quiere contrastar
mu_0 = 3.4

# calcula el estadístico de contraste con la muestra (var. conocida)
n = len(x)
t0 = (np.mean(x)- mu_0)/(sigma2_pob/np.sqrt(n))

# crea la distribución del estadístico de contraste
T0 = stats.norm(0,1)

# calculamos el pvalor (caso H1 BILATERAL)
pvalor = 2 * (1 - T0.cdf(abs(t0)))

print(f'El valor observado del estadistico de contraste es {t0}, con un p-valor asociado igual a {pvalor}.')


# razonando con valores críticos
# nivel de significación
alfa = 0.05

# determina los valores críticos que definen la región de rechazo
t_crit_iz = T0.ppf(alfa)
t_crit_de = T0.ppf(1-alfa)

print(f'valores criticos (alfa={alfa}): {t_crit_iz, t_crit_de}')
```

### 6.2 Contraste de hipótesis para la media de X (varianza desconocida)
#### H0: $\mu_X = \mu_0 $
#### H1: $\mu_X \neq \mu_0 $ (bilateral: "two-sided")
#### H1: $\mu_X < \mu_0 $ (unilateral: "less")
#### H1: $\mu_X > \mu_0 $ (unilateral: "greater")

```
# ejemplo de muestra de datos para x
x = [1,2,4,5,3,6,5,8,5,5,6,9,10]

# media poblacional que se quiere contrastar
mu_0 = 3.4
```

```
#-------------------------------
# CÁLCULO DETALLADO, PASO A PASO
#-------------------------------

# calcula el estadístico de contraste con la muestra (var. conocida)
n = len(x)
x_media = np.mean(x)
sx = np.std(x, ddof=1) # dividiendo por n-1

t0 = (x_media - mu_0) / (sx/np.sqrt(n))

# crea la distribución del estadístico de contraste
df = n -1 # grados de libertad
T0 = stats.t(df)

# calculamos el pvalor (caso H1 BILATERAL)
pvalor = 2 * (1 - T0.cdf(abs(t0)))

print(f'El valor observado del estadistico de contraste es {t0}, con un p-valor asociado igual a {pvalor}.')


# razonando con valores críticos
# nivel de significación
alfa = 0.05

# determina los valores críticos que definen la región de rechazo (bilateral)
t_crit_iz = T0.ppf(alfa/2)
t_crit_de = T0.ppf(1-alfa/2)

print(f'valores criticos (alfa={alfa}): {t_crit_iz, t_crit_de}')
```

```
# REPRESENTACIÓN GRÁFICA (BILATERAL)

# puntos para representar la distribución y las regiones de aceptación y rechazo
xmin = T0.ppf(0.000001)
xmax = T0.ppf(1-0.000001)
xs = np.linspace(xmin, xmax, 1000)
xRA = np.linspace(t_crit_iz, xmax, 1000)
xRRiz = np.linspace(xmin, t_crit_iz, 1000)
xRRde = np.linspace(t_crit_de, xmax, 1000)

plt.figure(1, figsize=(20, 6))
plt.plot(xs, T0.pdf(xs), linewidth = 4, label=f'Distribución T0 si H0 es cierta')

plt.fill_between( xRA, T0.pdf(xRA), color="green", alpha=0.6, label=f'Probabilidad asociada RA = {1-alfa}')
plt.fill_between( xRRiz, T0.pdf(xRRiz), color="red", alpha=0.6, label=f'P(T0<{t_crit_iz}) = {alfa/2})')
plt.fill_between( xRRde, T0.pdf(xRRde), color="red", alpha=0.6, label=f'P(T0>{t_crit_de}) = {alfa/2})')

plt.vlines(t0, 0, np.max(T0.pdf(xs)), color='black', label=f't0 = {t0}')

plt.hlines(-0.02, t_crit_iz, xmax, color='green', label = f'RA: [{t_crit_iz},{t_crit_de}] al {alfa*100}%')
plt.hlines(-0.02, xmin, t_crit_iz, color='red', linewidth = 6, label = f'RR (izq)')
plt.hlines(-0.02, t_crit_de, xmax, color='red', linewidth = 6, label = f'RR (der)')

plt.title('Distribución del estimador de contraste')
plt.xlabel('T0')
plt.legend()
plt.show()
```

```
#-------------------------------
# CÁLCULO DIRECTO
#-------------------------------

# tioo de laternativa
tipo_alternativa =  'two-sided' # alternativas posibles: ‘two-sided’, ‘less’, ‘greater’

# calculamos el estadístico en la muestra t0 y su pvalor
res = stats.ttest_1samp(x, popmean = mu_0, alternative = tipo_alternativa) 
print('Resultado: ', res)

t0 = res.statistic
pvalor = res.pvalue

print(f'El valor observado del estadistico de contraste es {t0}, con un p-valor asociado igual a {pvalor}.')
```
