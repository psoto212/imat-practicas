### a)  Sabiendo que el Real Madrid marca una media de 2.125 goles por partido. ¿Cuál es la probabilidad de que el Real Madrid marque al menos un gol en un partido? Justifica la respuesta. (0.5 puntos)

```
# Respuesta

print("Definimos la variable aleatoria N_RM como el númeo de goles que el Real Madrid marca por partido.")
print("La variable aleatoria N_RM sigue una distribución de Poisson con parámetro lambda_RM.")
print("Usando la informacion del enunciado sabemos que lambda_RM = E[N_RM] = 2.125 goles.")
lmda_RM = 2.125
N_RM = stats.poisson(lmda_RM)
prob = 1- N_RM.pmf(0)
print("La probabilidad de que el Real Madrid marque al menos un gol es: P(N_RM>=1) = 1-P(N_RM=0)")
print(f"P(N_RM>=1) = 1 - P(N_RM=0) = 1 - {N_RM.pmf(0)} = {prob} ")
```

### b) ¿Cuál es la probabilidad de que el Real Madrid marque entre dos y cuatro goles en un partido? Justifica la respuesta (0.5 puntos)

```
# Respuesta

print("La probabilidad de que el Real Madrid marque entre 2 y 4 goles es: P(2<=N_RM & N_RM<=4)")
print("P(2<=N_RM & N_RM<=4) = P(N_RM<=4) - P(N_RM<=2) + P(N_RM=2)  = "\
      , N_RM.cdf(4) -  N_RM.cdf(2) +  N_RM.pmf(2))
print("También se podría calcular como: P(2<=N_RM>=4) = P(N_RM=2) + P(N_RM=3) + P(N_RM=4) =", N_RM.pmf(2) + N_RM.pmf(3) + N_RM.pmf(4))
```

### c) ¿Cuál es el número de goles más probable que el Real Madrid marque en un partido? Justifica la respuesta (0.5 puntos)

```
# Respuesta

print('Se pide el valor de la variable más probable (la moda). Pintando la función de masa de probabilidad se puede encontrar fácilmente:')

# puntos en los que se evalua la función para representarla
xs = range(-2, 6)

# usando "alfileres"
fig = plt.figure(1, figsize=(6, 3))
plt.stem(xs, N_RM.pmf(xs), basefmt='none')
#plt.ylim(0, 1) para ver mejor las diferencias
plt.xlabel('x')
plt.ylabel('pmf(x)')
plt.grid()
plt.show()
print("Si miramos el gráfico podemos ver que el número de goles más probable es 2 (seguido de cerca por el valor 1).")

print("Se puede calcular automáticamente con la función argmax: valor buscado = argmax(pmf())")
index_moda = np.argmax(N_RM.pmf(xs))
prob_moda = N_RM.pmf(xs)[index_moda]
print(f"El número de goles más probable es {xs[index_moda]} con una probabilidad de {prob_moda}.")
```

### d) El número de goles que el Real Madrid marca por partido depende si su delantero estrella juega dicho partido o no. Se han realizado cálculos y se han obtenido los siguientes resultados: Si el delantero estrella juega, el número de goles que el Madrid marca por partido es de 2.5, sin embargo cuando el delantero estrella no juega en número de goles promedio baja a 1.5.

### El delantero estrella sufrió una caída en el partido de la semana pasada, dañándose la rodilla. Según los médicos la probabilidad de que el delantero juegue el partido de esta semana es del 60%. No hemos podido ver el partido de esta semana, pero si sabemos que el Real Madrid ha marcado 3 goles en el partido, ¿cuál es la probabilidad de que el delantero estrella haya jugado esta semana? Justifica la respuesta (1.5 puntos)

```
# Respuesta

print("Definimos el suceso D: El delantero juega esta semana. Hay dos sucesos (juega o no), que constituyen una partición disjunta del espacio muestral.")
Prob_D = 0.6
Prob_No_D = 0.4

print("La variable N_RM_D representa el número de goles que el Madrid marca si el delantero juega\
y sigue una distribción de Poisson de parámetro 2.5 ")
lmda_RM_D = 2.5
N_RM_D = stats.poisson(lmda_RM_D)

print("La variable N_RM_No_D representa el número de goles que el Madrid marca si el delantero no juega esta semana\
y sigue una distribución de Poisson de parámetro 1.5")
lmda_RM_No_D = 1.5
N_RM_No_D = stats.poisson(lmda_RM_No_D)

Prob_N_RM_3 = N_RM_D.pmf(3)*Prob_D + N_RM_No_D.pmf(3)*Prob_No_D
print("Usando la ley de la probabilidad total obtenemos que la probabilidad de que el Madrid marque tres goles es:")
print(f"P(N_RM=3) = P(N_RM=3|D) P(D) +  P(N_RM=3|No_D) P(No_D) = {N_RM_D.pmf(3)} * {Prob_D} + {N_RM_No_D.pmf(3)} * {Prob_No_D} = {Prob_N_RM_3}")

Prob_D_sabiendo_N_RM_3 = N_RM_D.pmf(3)*Prob_D/Prob_N_RM_3
print("Finalmente usando el Teorema de Bayes obtenemos: P(D|N_RM=3) = P(N_RM=3|D)*P(D)/ P(N_RM=3)")
print(f"P(D|N_RM=3) = P(N_RM=3|D)*P(D)/ P(N_RM=3) = {N_RM_D.pmf(3)} * {Prob_D} / {Prob_N_RM_3}  = {Prob_D_sabiendo_N_RM_3}")
```

### e) Si el Atlético de Madrid marca 2.14 goles de media por partido y el Fútbol Club Barcelona marca un promedio de 1.9 por partido, ¿cuál es la probabilidad de que en un partido que enfrenta al Atlético y al Barcelona se marquen al menos dos goles? Asume que el número de goles marcados por el Atlético y el número de goles marcados por el Barcelona son independientes. Justifica la respuesta (2 puntos)

```
# Respuesta

print("La variable N_AT representa el número de goles que el Atlético de Madrid marca por partido\
 y sigue una distribución de Poisson de parámetro 2.14")
lmda_AT = 2.14
N_AT = stats.poisson(lmda_AT)

print("La variable N_B representa el número de goles que el Barcelona marca por partido\
 y sigue una distribución de Poisson de parámetro 1.9")
lmda_B = 1.9
N_B = stats.poisson(lmda_B)

print("La variable N_T representa el número de goles totales: N_T = N_AT + N_B")
lmda_total = lmda_AT + lmda_B
N_T = stats.poisson(lmda_total)

print(f"Por la ley de reproducibilidad N_T sigue una distribución de Poisson de parámetro {lmda_AT} + {lmda_B} = {lmda_total}")

prob = 1-N_T.cdf(1)
print("Por tanto, la probabilidad de que se marquen al menos dos goles en un partido es: ")
print(f"P(N_T>=2) = P(N_T>1) = 1 - P(N_T<=1) = {prob}")
```

### f) El Atlético de Madrid y el Barcelona han jugado un partido en el que se han metido 3 goles. ¿Cuál es la probabilidad de que el Atlético de Madrid haya ganado? Justifica la respuesta (2.5 puntos)

```
# Respuesta

print("Se pide la probabilidad condicionada P(N_AT > N_B | N_T = 3), siendo N_T = N_AT + N_B")

print("Por la definición de probabilidad condicionada obtenemos que:")
print("P(N_AT > N_B | N_T=3) = P(N_AT - N_B>0 | N_T=3) = P(N_AT - N_B>0 y N_T=3)/P(N_T=3)")

Pr_total_3 = N_T.pmf(3)
print("\nLa probabilidad del denominador se calcula con la ley reproductibilidad de la distribución de Poisson, como en el apartado anterior.")
print(f"P(N_T = 3) = {Pr_total_3}")

print("\nEl numerador implica calcular la probabilidad de que gane el Atlético y se metan 3 goles.")
print("Para calcular el numerador consideramos las 4 posibles opciones que hay, dado que N_T = 3:")
print(" Atleti | Barca")
print("    0       3  ")
print("    1       2  ")
print("    2       1    -> Gana Atleti")
print("    3       0    -> Gana Atleti")
print("La probabilidad del numerador entonces se puede calcular fácilmente así:")
print("P(N_AT - N_B>0 y N_T=3) = P( (N_AT=2 y N_B=1) o (N_AT=3 y N_B=0) )")
print("Considerando que son sucesos independientes (los compuestos y los simples):")
print("P( (N_AT=2 y N_B=1) o (N_AT=3 y N_B=0) ) = P(N_AT=2 y N_B=1) + P(N_AT=3 y N_B=0) = ")
print("P(N_AT=2)*P(N_B=1) + P(N_AT=3)*P(N_B=0)")

print("Estas probabilidades se calculan fácilmente")
Pr_AT_3 = N_AT.pmf(3)
Pr_AT_2 = N_AT.pmf(2)
Pr_B_0 = N_B.pmf(0)
Pr_B_1 = N_B.pmf(1)
Pr_numerador = Pr_AT_2*Pr_B_1  +  Pr_AT_3*Pr_B_0
Pr_final = Pr_numerador/Pr_total_3

print(f"Por tanto P(N_AT - N_B>0 y N_T=3) = {Pr_AT_2*Pr_B_1} + {Pr_AT_3*Pr_B_0} = {Pr_numerador}")
print(f"Finalmente obtenemos que la probabilidad buscada es {Pr_final}")
```

### g) En una temporada el Atlético de Madrid juega 50 partidos. Calcula de manera aproximada, pero realizando una buena aproximación y justificando la respuesta, la probabilidad de que la media de goles por temporada del Atlético de Madrid sea mayor que 2.5. Considera que los partidos son independientes y que el Atlético de Madrid marca 2.14 goles de media por partido. (2.5 puntos)

```
# Respuesta

print("Para calcular la probabilidad pedida necesitamos conocer como se distribuye la v.a.")
print("N_MED_TEMP_AT: nº medio de goles del Atleti en una temporada.")

print("\nN_MED_TEMP_AT es el promedio de 50 variables N_AT, i.i.d. distibuidas según una Poisson de lambda = 2.14")
print(f"Por el Teorema central del Límite sabemos que una buena aproximación de la distribución de la v.a. N_MED_TEMP_AT es:")
print(f"N_MED_TEMP_AT ~ Normal(mu_N_MED_TEMP_AT, var_N_MED_TEMP_AT), en donde")
mu_N_MED_TEMP_AT = lmda_AT
var_N_MED_TEMP_AT = lmda_AT/50
print(f" - la media mu_N_MED_TEMP_AT = lmda_AT = {mu_N_MED_TEMP_AT}")
print(f" - la varianza var_N_MED_TEMP_AT = var(N_AT)/ 50 = {N_AT.var()} / 50 = {var_N_MED_TEMP_AT}")

N_MED_TEMP_AT = stats.norm(loc = mu_N_MED_TEMP_AT, scale = np.sqrt(var_N_MED_TEMP_AT))
prob_pedida_aprox = 1 - N_MED_TEMP_AT.cdf(2.5)
print("\nPor tanto, la probabilidad pedida P(Media Temporada > 2.5), aproximando la media de las 50 v.a. iid, vale:")
print(f"P(N_MED_TEMP_AT>2.5) = 1 - P(N_MED_TEMP_AT<=2.5) = {prob_pedida_aprox}")
```

```
# Calculo exacto (no pedido, pero por curiosidad para comparar y confirmar)

print("\n\n ** CÁLCULO EXACTO usando la reproductividad de la suma de poissones iid **")
print("N_MED_TEMP_AT = N_SUM_TEMP_AT/50, donde N_SUM_TEMP_AT es la suma de 50 v.a. i.i.d. N_AT. Siendo N_AT una Poisson de lambda = 2.14")
print("Como la Poisson es reproductiva, N_SUM_TEMP_AT seguirá también una Poisson de lambda = 50 * 2.14")
lmda_50 = 50 * lmda_AT
N_SUM_TEMP_AT = stats.poisson(lmda_50)

print(f"Aplicando el cambio de variable en la entrada hay que evaluar en: {2.5 * 50}")
prob_pedida_exacta = 1 - N_SUM_TEMP_AT.cdf(2.5 * 50)
print("La probabilidad pedida se puede calcular de forma exacta a partir de la v.a. N_SUM_TEMP_AT, considerando el cambio de escala directamente")
print(f"La Probabilidad exacta es {prob_pedida_exacta}")
```
