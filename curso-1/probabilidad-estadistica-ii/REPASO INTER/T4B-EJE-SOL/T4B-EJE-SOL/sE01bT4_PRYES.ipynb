```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### a) Crea una variable aleatoria X distribuida según una chi-cuadrado con 7 grados de libertad. Representa su función de densidad. Observando únicamente dicha gráfica, intenta imaginar el valor de la media, la mediana y la moda de X. Calcula la media y la mediana y calibra tu percepción inicial.

```
# Respuesta 

X = stats.chi2(7)

# pinta
xs = np.linspace(0, X.ppf(0.9999), 1000)

# pinta sincronizadamente ambas funciones en dos suplots
fig = plt.figure(1, figsize=(15, 6))
plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X.pdf(xs), color="orange", label = 'pdf(x)')
plt.xlabel('x')
plt.legend()
plt.grid()
plt.show()

# cálculo exacto
med_x = X.mean()
mediana_x = X.median()

print('La media de X vale:', med_x)
print('La mediana de X vale:', mediana_x)

print('Tiene sentido, ya que la media de una chi2 es igual a sus grados de libertad.')
print('La moda es 5, dos menos que los grados de libertad en este caso.')
print('La mediana es un poco mayor que la moda, y la media un poco mayor que la mediana, debido a la asimetría positiva')
```

### b) Calcula numéricamente la moda de la v.a. X haciendo un barrido por fuerza bruta.

```
# Respuesta 

# hacemos un barrido en la función
valor_min_x = X.ppf(0.2)
valor_max_x = X.ppf(0.5)

n_puntos = 1000000
x_prueba = np.linspace(valor_min_x, valor_max_x, n_puntos)

print(f'Hacemos un barrido con {n_puntos} en el intervalo acotado [{valor_min_x},{valor_max_x}], '
      'sabemos que la moda seguro que está dentro viendo la gráfica anterior.')

# función que se quiere maximizar, evaluada en x_prueba !!
func_eval_x = X.pdf(x_prueba)

# barrido básico evaluando la función
index_max_func = np.argmax(func_eval_x)
x_max_func = x_prueba[index_max_func]
max_func = func_eval_x[index_max_func]

# la moda es el valor de x que hace que la pdf sea máxima
moda_x = x_max_func

print(f'La moda calculada numéricamente vale {moda_x}, es el punto dónde se alcanza la máxima densidad ({max_func})')
print('El valor obtenido de forma aproximada se parece bastante al valor teórico (5)')
```

### c) Representa, en la misma gráfica, la función de densidad de X y el valor de la moda, la media y la mediana. Estos tres últimos valores indícalos con una recta vertical para que se observen claramente.

```
# Respuesta 

xs = np.linspace(0, X.ppf(0.999), 1000)

# pinta sincronizadamente ambas funciones en dos suplots
fig = plt.figure(1, figsize=(15, 6))
plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X.pdf(xs), color="orange", label = 'pdf(x)')

# añadimos los tres puntos interesantes
plt.vlines(med_x, 0, max_func, color="black", linewidth = 2, label=f'media(X)={med_x}')
plt.vlines(mediana_x, 0, max_func, color="red", linewidth = 2, label=f'mediana(X)={mediana_x}')
plt.vlines(moda_x, 0, max_func, color="blue", linewidth = 2, label=f'moda(X)={moda_x}')

plt.xlabel('x')
plt.legend()
plt.grid()
plt.show()
```

### d) Calcula los percentiles 10, 20, 30, ..., 90 de la v.a. X (se denominan también deciles 1, 2, ..., 9).

```
# Respuesta 

percentiles_calcular =  np.linspace(0.1, 0.9, 9)
print('Utilizando la inversa de la cdf(X), evaluada en los puntos:')
print(percentiles_calcular)

percentiles = X.ppf(percentiles_calcular)

print('Los percentiles pedidos son:')
print(percentiles)
```

### e) Representa en una misma figura dos gráficas, una con la función acumulada de probabilidad y otra, debajo, con la función de densidad de la v.a. X. Sobre cada gráfica incluye los deciles calculados en el apartado anterior. Utiliza rectas verticales y rectas horizontales para indicar claramente cómo se han calculado los deciles. Además, indica en la función de densidad el área que representa la probabilidad de que la v.a. X sea menor o igual que el decil 8. ¿Están los percentiles obtenidos equiespaciados? ¿Cómo tendría que ser la función acumulada de probabilidad para que eso ocurriese?

```
# Respuesta

xs = np.linspace(0, X.ppf(0.99), 1000)
xs_decil8 = np.linspace(0, X.ppf(0.8), 1000)

# pinta sincronizadamente ambas funciones en dos suplots
fig = plt.figure(1, figsize=(15, 6))

ax1 = plt.subplot(2,1,1)
plt.plot(xs, X.cdf(xs), color="red", label = 'cdf(x)')
plt.plot(percentiles, percentiles_calcular, 's', color="black")
plt.hlines(percentiles_calcular, 0, percentiles, color="darkgray", linewidth = 2, label=f'{percentiles_calcular}')
plt.vlines(percentiles, 0, X.cdf(percentiles), color="black", linewidth = 2, label=f'Deciles')
plt.legend()
plt.grid()

plt.subplot(2,1,2, sharex = ax1)
plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot(xs, X.pdf(xs), color="orange", label = 'pdf(x)')

# añadimos los percentiles
max_func = 0.13
plt.vlines(percentiles, 0, max_func, color="black", linewidth = 2, label=f'Deciles')

# el área pedido para el decil 8
plt.fill_between(xs_decil8, X.pdf(xs_decil8), color="green", alpha=0.5, label=f'P(X<{X.ppf(0.8)}) = 0.8')


plt.xlabel('x')
plt.legend()
plt.grid()
plt.show()

print('Por ejemplo, el percentil 20 (decil 2) se obtiene entrando en la cdf por la ordenada con el valor 0.2 (función inversa de la cdf).')
      
print('Se puede observar claramente que los percentiles calculados no están equiespaciados.')
print('La cdf tendría que ser una recta para garantizar que estén equiespaciados, es decir, la v.a. X tendría que distribuirse como una uniforme.')
```
