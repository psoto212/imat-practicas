```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### a) Representa, en la misma gráfica, la función de densidad de dos variables aleatorias X1 e X2. La variable X1 sigue una distribución F de Snedecor con grados de libertad (4,10) y la v.a. X2 se distribuye según una F de (100, 400) grados de libertad. Asegúrate de que se observan bien las colas de la distribución. Observando el resultado, ¿qué variable tiene una distribución más asimétrica? ¿Qué variable aleatoria tiene mayor desviación típica? Calcula su desviaciones y verifica que lo observado en la gráfica es coherente con el resultado.

```
# Respuesta 

# las dos variables
X1 = stats.f(4,10)
X2 = stats.f(100, 400)

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(0, X1.ppf(0.99), 1000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(10, 6))
plt.plot( xs, X1.pdf(xs), color="orange", label = 'pdf(X1)')
plt.plot( xs, X2.pdf(xs), color="blue", label = 'pdf(X2)')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('La variable X1 es mucho más asimétrica que X2 y tiene mayor desviación típica, es evidente según la gráfica.')

std_X1 = X1.std()
std_X2 = X2.std()

print(f'std(X1)={std_X1} y std(X2)={std_X2}. Mucho mayor la desviación típica de X1, lógico.')
```

### c) Calcula la probabilidad de que X1 sea mayor que 2. Representa gráficamente el área asociada a dicha probabilidad en la función de densidad, así como la función de probabilidad acumulada que permite calcularla.

```
# Respuesta

prob = 1 - X1.cdf(2)

print(f'La P(X1>2) = 1 - cdf(2) =  {prob}')

# gráficamente

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(0, X1.ppf(0.99), 1000)
xs_mayor2 = np.linspace(2, X1.ppf(0.99), 1000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(10, 6))

ax1 = plt.subplot(2,1,1)
plt.plot( xs, X1.cdf(xs), color="blue", label = 'cdf(X1)')
plt.plot(2, X1.cdf(2), 's', color="red")
plt.vlines(2, 0, X1.cdf(2), color="red", label = f'cdf(2)={X1.cdf(2)}')
plt.ylabel('p')
plt.legend()
plt.grid()

plt.subplot(2,1,2, sharex = ax1)
plt.plot( xs, X1.pdf(xs), color="blue", label = 'pdf(X1)')
plt.fill_between( xs_mayor2, X1.pdf(xs_mayor2), color="red", alpha=0.3, label = f'P(X1>2)=1-cdf(2)={prob}')
plt.ylabel('p')
plt.legend()
plt.grid()

plt.show()
```

### d) Crea una v.a. X3 que se distribuya según una F con (10000, 10000) grados de libertad. Calcula su media y desviación típica. Representa en una misma gráfica la distribución de densidad de X3 y de una nueva variable aleatoria N3 que se distribuya como una normal de media la de X3 y varianza la de X3. Observa el resultado. ¿Se parecen?

```
# Respuesta

X3 = stats.f(10000, 10000)
media_x3 = X3.mean()
sigma_x3 = X3.std()

print(f'La media(X3)={media_x3} y la std(X3)={sigma_x3}')

N3 = stats.norm(loc = media_x3, scale = sigma_x3)

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace( X3.ppf(0.0001), X3.ppf(0.9999), 1000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(15, 6))
plt.plot( xs, X3.pdf(xs), color="blue", label = 'X3~F(10000,10000)')
plt.plot( xs, N3.pdf(xs), color="red", label = f'N3~N({media_x3},{sigma_x3})')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('Se parecen bastante, parece que hay un pequeño "deslizamiento". Para verlo mejor representamos las diferencias:')
fig = plt.figure(1, figsize=(15, 6))
plt.plot( xs, X3.pdf(xs) - N3.pdf(xs), color="orange", label = 'Diferencia pdf(X3) - pdf(N3)')
plt.hlines(0, xs[0], xs[-1], color="black", label = f'Diferencia nula')
plt.legend()
plt.grid()
plt.show()
```

### e) Representa las funciones acumuladas de probabilidad de las variables X3 y N3 del apartado anterior. Representa también la diferencia de ambas funciones para ver mejor lo que ocurre. Observa el resultado. ¿Se parecen?

```
# Respuesta

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace( X3.ppf(0.001), X3.ppf(0.999), 1000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(15, 6))
ax1 = plt.subplot(2,1,1)
plt.plot( xs, N3.cdf(xs), color="blue", label = 'X3~F(10000,10000)')
plt.plot( xs, N3.cdf(xs), '--', color="red", label = f'N3~N({media_x3},{sigma_x3})')
plt.ylabel('p')
plt.legend()
plt.grid()

plt.subplot(2,1,2, sharex = ax1)
fig = plt.figure(1, figsize=(15, 6))
plt.plot( xs, X3.cdf(xs) - N3.cdf(xs), color="orange", label = 'Diferencia cdf(X3) - cdf(N3)')
plt.hlines(0, xs[0], xs[-1], color="black", label = f'Diferencia nula')
plt.legend()
plt.grid()
plt.show()

print('Se parecen mucho mas las cdfs que las pdfs. Además, en la gráfica con las diferencias, '
      'se observa una función con una simetría par. Sin embargo en la gráfica con las diferencias '
      'de las pdfs la asimetría era impar.')
```
