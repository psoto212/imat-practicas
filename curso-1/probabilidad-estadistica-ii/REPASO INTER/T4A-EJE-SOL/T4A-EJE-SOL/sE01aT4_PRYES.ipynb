```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### a) Representa, en la misma gráfica, la función de densidad de dos variables aleatorias X e Y. La variable X sigue una distribución chi-cuadrado con 10 grados de libertad y la v.a. Y se distribuye según una chi-cuadrado de 30 grados de libertad. Asegúrate de que se observa bien las colas de ambas distribuciones. Observando el resultado, ¿qué distribución es más asimétrica? ¿Eres capaz de localizar la media de ambas variables en la gráfica? ¿La moda de X es menor o mayor que su media?

```
# Respuesta 

# las dos variables
X = stats.chi2(10)
Y = stats.chi2(30)

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(0, Y.ppf(0.9999), 1000)
print('Hemos usado la v.a. Y para determinar hasta dónde pintar por la izquierda para que se vea bien las colas, ya que la v.a. Y,  '
      'al tener más grados de libertad, está más desplazada a la derecha')

# pinta sincronizadamente ambas funciones en dos suplots
fig = plt.figure(1, figsize=(6, 6))
plt.fill_between(xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot(xs, X.pdf(xs), color="orange", label = 'pdf(X)')
plt.fill_between(xs, Y.pdf(xs), color="blue", alpha=0.3)
plt.plot(xs, Y.pdf(xs), color="blue", label = 'pdf(Y)')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('La v.a. X es claramente más asimétrica (a la derecha) que la v.a. Y')
print('La medida de una distribución chi-cuadrado es el nº de grados de libertad. Por tanto, la media de X es 10 y la de Y 30, coherente con lo observado.')
print('La moda de X es el valor con mayor densidad, y es claramente inferior a 10. Lógico al ser una distribución asimétrica a la derecha.')
```

### b) Calcula la probabilidad de que la variable X sea menor que la media de la variable Y. Comenta el resultado comparando con la gráfica del apartado anterior.

```
# Respuesta 

# usamos la cdf de X para realizar el cálculo
prob = X.cdf(Y.mean())
print(f'La probabilidad de que la variable X sea menor que la media de la variable Y es P(X<mean(Y)) = cdf({Y.mean()}) = {prob}')
print('Es lógico que sea una probabilidad muy alta, ya que la densidad de X está concentrada por debajo de 30')
```

### c) Calcula la probabilidad de que la v.a. Y sea mayor que el percentil 95 de la v.a. X

```
# Respuesta

# calculamos el percentil 95 de X usando la función inversa
percentil_95_X = X.ppf(0.95)

prob = 1 - Y.cdf(percentil_95_X)

print(f'La probabilidad de que la v.a. Y sea mayor que el percentil 95 de la v.a. X es P(Y>{percentil_95_X})=1-cdf({percentil_95_X})={prob}')
```

### d) Calcula la probabilidad de que la v.a. Y sea, al mismo tiempo, mayor que 20 y menor que 40.

```
# Respuesta

prob = Y.cdf(40) - Y.cdf(20)

print(f'La probabilidad pedida es P(20<Y<40)=cdf(40)-cdf(20)={prob}')
```

### e) Calcula la mediana de la v.a. Y. A continuación, sin realizar ningún cálculo, razona si la probabilidad de que la v.a. Y sea mayor que su mediana + 10, es decir P(Y>mediana+10),  tiene que ser mayor o menor que la probabilidad de que Y sea menor que su mediana - 10, es decir P(Y<mediana-10). Una vez razonado, realiza el cálculo y confirma tu razonamiento. Representa la función de densidad de Y y las áreas de las probabilidades que se comparan.

```
# Respuesta

mediana_Y = Y.ppf(0.5)

print(f'La mediana de Y es {mediana_Y}')
print('Dado que la mediana deja el 50% de la distribución a cada lado, si nos alejamos 10 desde ese punto, la P(Y>mediana+10) será mayor '
      'que P(Y<mediana-10) ya que la distribución es un poco asimétrica a la derecha.')

prob_Y_menor20 = Y.cdf(mediana_Y - 10)
prob_Y_mayor40 = 1 - Y.cdf(mediana_Y + 10)

print(f'La P(Y<{mediana_Y - 10}) = {prob_Y_menor20}, que es llamativamente menor que P(Y>{mediana_Y + 10}) = {prob_Y_mayor40}')


# visualmente
print('Mostramos gráficamente con la pdf las áreas asociadas a cada probabilidad:')

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(0, Y.ppf(0.9999), 1000)
xs_menor20 = np.linspace(0, mediana_Y - 10, 1000)
xs_mayor40 = np.linspace(mediana_Y + 10,  Y.ppf(0.9999), 1000)

fig = plt.figure(1, figsize=(14, 6))
plt.fill_between( xs, Y.pdf(xs), color="blue", alpha=0.3)
plt.plot( xs, Y.pdf(xs), color="blue", label = 'pdf(Y)')
plt.vlines(mediana_Y, 0, Y.pdf(mediana_Y), color = 'black', label = f'Mediana={mediana_Y}')
plt.fill_between(xs_menor20, Y.pdf(xs_menor20), color="green", alpha=0.3, label = f'P(Y<{mediana_Y - 10})')
plt.fill_between(xs_mayor40, Y.pdf(xs_mayor40), color="red", alpha=0.3, label = f'P(Y>{mediana_Y + 10})')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('La gráfica produce la falsa sensación que el área roja de la derecha es solo un poco mayor que el área verde de la izquierda, '
      f'cuando en realidad es {prob_Y_mayor40/prob_Y_menor20} veces mayor. Esto ocurre porque la cola de la izquierda empieza en 0 '
      'al ser una distribución chi-cuadrado, pero la de la derecha es extiende teóricamente hasta el infinito.')
```
