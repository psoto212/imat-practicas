```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### a) Representa, en la misma gráfica, la función de densidad de dos variables aleatorias X1 e X2. La variable X1 sigue una distribución t con 4 grados de libertad y la v.a. X2 se distribuye según una t de Student de 100 grados de libertad. Observando el resultado, ¿qué variable aleatoria tiene mayor varianza? Calcula su varianzas y verifica que lo observado en la gráfica es coherente con el resultado.

```
# Respuesta 

# las dos variables
X1 = stats.t(4)
X2 = stats.t(100)

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(X1.ppf(0.001), X1.ppf(0.999), 1000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(10, 6))
plt.plot( xs, X1.pdf(xs), color="orange", label = 'pdf(X1)')
plt.plot( xs, X2.pdf(xs), color="blue", label = 'pdf(X2)')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('La variable X1 tiene mayor varianza, es evidente según la gráfica.')

var_X1 = X1.var()
var_X2 = X2.var()

print(f'var(X1)={var_X1} y var(X2)={var_X2}. Casi el doble, no parecía que fuesen a ser tan diferentes.')
```

### b) Deduce, si realizar ningún cálculo, la probabilidad de que la variable X1 sea menor que la media de la variable X2. Una vez determinado, realiza el cálculo para confirmar que estabas en lo cierto.

```
# Respuesta 

print('Dado que X1 y X2 se distribuyen como una t de Student, su media es cero (se observa fácilmente en la gráfica anterior). Además son simétricas, por lo que '
      'la P(X1<0) es exactamente 0.5')

# verificamos

prob_X1_menor_media_X2 = X1.cdf(X2.mean())

print(f'Realizando el cálculo con la cdf se obtiene: {prob_X1_menor_media_X2}')
```

### c) Sean los sucesos A: X2<-1 y B: X2>+1. Calcula las siguientes probabilidades: P(A y B), P(A o B) y P(noA o noB). Representa la función de densidad de X2 y las áreas de las probabilidades involucradas en los cálculos, comentando los resultados.

```
# Respuesta

print('Las probabilidades de los sucesos A y B son:')
Pr_A = X2.cdf(-1)
Pr_B = 1 - X2.cdf(1)

print(f'  - P(A) = P(X2<-1) = {Pr_A}')
print(f'  - P(B) = P(X2>-1) = {Pr_B}')
print('Lógicamente, dada la simetría de la distribución t, estas probabilidades tienen que ser exactamente iguales.')

# Probabilidades pedidas
Pr_AyB = 0

print('\nLa P(A y B) es nula ya que los sucesos A y B son incompatibles, no pueden ocurrir simultáneamente.')
print(f'La P(A o B) se puede calcular de varias formas:')
prob1 = 1 - (X2.cdf(1) - X2.cdf(-1))
prob2 = 2 * X2.cdf(-1)
Pr_AoB = Pr_A + Pr_B - Pr_AyB

print(f'- Usando el complementario: P(A o B) = 1 - P( no (A o B)) = 1 - P(X2>-1 y X2<1) = 1 - (cdf(1)-cdf(-1)) = {prob1}')
print(f'- Considerando que la distribución es simétrica y centrada en 0: P(A o B) = 2 * P(X2<-1) = 2 * cdf(-1) =  {prob2}')
print(f'- Usando la fórmula de la unión: P(A o B)= P(X2<-1 o X2>-1) = P(A) + P(B) - P(AyB) = {Pr_A} + {Pr_B} - {Pr_AyB} = {Pr_AoB}')

print('\nPor último, para calcular P(noA o noB) no es necesario realizar ningún cálculo, vale 1. La unión de noA y noB '
      'cubre todo el espacio muestral. Se puede confirmar haciendo los cálculos correctos para quedarnos tranquilos '
      'con el razonamiento:')
Pr_noA = 1 - Pr_A
Pr_noB = 1 - Pr_B
Pr_noA_y_noB = X2.cdf(1) - X2.cdf(-1)
Pr_noA_o_noB = Pr_noA + Pr_noB - Pr_noA_y_noB

print(f'- Usando la fórmula de la unión: P(noA o noB)= P(noA) + P(noB) - P(noA y noB) = {Pr_noA} + {Pr_noB} - {Pr_noA_y_noB} = {Pr_noA_o_noB}')
```

```
# visualmente
print('Mostramos gráficamente con la pdf las tres áreas asociadas a las probabilidades:')

# puntos en los que se evalua las funciones para representarlas
xmin = X2.ppf(0.000001)
xmax = X2.ppf(1-0.000001)
xs = np.linspace(xmin, xmax, 1000)
xs_menor_menos1 = np.linspace(xmin, -1, 1000)
xs_mayor_mas1 = np.linspace(1,  xmax, 1000)

fig = plt.figure(1, figsize=(14, 4))
plt.fill_between( xs, X2.pdf(xs), color="blue", alpha=0.3)
plt.plot( xs, X2.pdf(xs), color="blue", label = 'pdf(Y)')
plt.fill_between(xs_menor_menos1, X2.pdf(xs_menor_menos1), color="green", alpha=0.3, label = f'P(X2<-1)')
plt.fill_between(xs_mayor_mas1, X2.pdf(xs_mayor_mas1), color="red", alpha=0.3, label = f'P(X2>+1)')

# mostramos visualmente los sucesos A y B
plt.hlines(-0.02, xmin, -1, color='green', linewidth = 6, label = f'Suceso A (X2<-1)')
plt.hlines(-0.02, 1, xmax, color='red', linewidth = 6, label = f'Suceso B (X2>+1)')

plt.hlines(-0.04, -1, xmax, color='blue', linewidth = 6, label = f'Suceso no A (X2>-1)')
plt.hlines(-0.06, xmin, 1, color='orange', linewidth = 6, label = f'Suceso no B (X2<+1)')

plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```

### e) Representa en una misma gráfica la distribución de densidad de X2 y de una nueva variable aleatoria que se distribuya como una normal de media la de X2 y varianza la de X2. Observa el resultado, ¿Se parecen? Repite la misma comparación pero con X1 ¿Son tan parecidas como en el caso anterior?

```
# Respuesta

# la normal a la que se debería parecer X2 es
N2 = stats.norm(loc = X2.mean(), scale = X2.std())

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(X2.ppf(0.001), X2.ppf(0.999), 1000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(10, 6))
plt.plot( xs, X2.pdf(xs), color="blue", label = 'X2~t(100)')
plt.plot( xs, N2.pdf(xs), color="red", label = f'N2~N({X2.mean()}, {X2.std()})')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('Sí se parecen, era lo esperado ya que X2 se distribuye como una t de Student con muchos grados de libertad.')

# la normal a la que se debería parecer X1
N1 = stats.norm(loc = X1.mean(), scale = X1.std())

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(10, 6))
plt.plot( xs, X1.pdf(xs), color="blue", label = 'X1~t(4)')
plt.plot( xs, N1.pdf(xs), color="red", label = f'N1~N({X1.mean()}, {X1.std()})')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('No se parecen, era lo esperado ya que X1 se distribuye como una t de Student con pocos grados de libertad.')
```

### f) Aproximamos la v.a. X1, que se distribuye según una t de Student, por una normal N1 que tenga la misma media y varianza que X1. Decidimos utilizar esta aproximación para calcular la probabilidad de que X1 sea menor que un valor "a", es decir, P(X1<a). ¿Para qué valor "a" el cálculo realizado con la normal N1 será igual que el valor real de dicha probabilidad?  Representa en una misma gráfica las funciones acumuladas de probabilidad de X1 y de la normal N1 y observa si ocurre lo que pensabas.

```
# Respuesta

# la normal N1 con la que se aproxima X1
N1 = stats.norm(loc = X1.mean(), scale = X1.std())

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(X2.ppf(0.00001), X2.ppf(0.99999), 10000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(15, 6))
plt.plot( xs, X1.cdf(xs), color="blue", label = 'X1~t(4)')
plt.plot( xs, N1.cdf(xs), color="red", label = f'N1~N({X1.mean()}, {X1.std()})')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('Efectivamente, la probabilidad cuando a = 0 que se obtendrá con la aproximación es exactamente 0.5, el valor real de que P(X1<a). '
      'Según no alejamos del cero la aproximación empeora, pero cuando estamos suficientemente lejor del centro entonces vuelve a mejorar al sumarse todo o nada en ambas distribuciones.')
print('Observando detenidamente los extremos, parece que la aproximación de la normal se queda corta para valores negativos (cerca del -4) y larga para positivos (cerca de 4). '
      'Esto tiene sentido ya que la t de Student tiene las colas más pesadas, con mayor probabilidad que la normal.')
```
