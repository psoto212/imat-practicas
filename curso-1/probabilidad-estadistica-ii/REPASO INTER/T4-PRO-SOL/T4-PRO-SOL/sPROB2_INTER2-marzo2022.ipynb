### a) Para estimar la media de X con precisión suficiente, a RAZAMAN le vale si el intervalo de confianza al 95% de confianza para la media de X tiene un ancho de intervalo de 60 segundos (límite superior - límite inferior). Determinar el tamaño mínimo que tiene que tener la muestra para que ocurra, asumiendo para ello que la variable X se distribuye como una normal. Para estimar dicho tamaño se cuenta con la muestra de 50 datos (variable x50). (4 puntos)

```
# Respuesta

print('El IC_95% de la media de X se obtiene a partir de la distribución teórica de la media muestral (una t de Student con n-1 grados de libertad), '
      'ya que suponemos que X se distribuye según una normal y no se conoce la varianza de X.')

print('Este intervalo es simétrico y su ancho está dado por 2 x radio, siendo el radio = valor_critico * sx/raiz_n, '
      'donde el valor_critico se obtiene de una t con n-1 grados de libertad y la desviación estándar sx se puede estimar a partir '
      'de la muestra de n=50.')

# tamaño muestral original
n = len(x50)

# valor crítico de una t con n-1 grados
gama = .95
alfa = 1-gama
vat = stats.t(n-1)
valor_critico = vat.ppf(1-alfa/2) # valor positivo
print(f'El valor crítico, según una t con {n -1} grados de libertad, vale {valor_critico}.')

# Estimamos la desviación típica a partir de la muestra de 50 datos
sx = np.std(x50,ddof=1) # dividiendo por n-1
print(f'La desviación típica de X estimada sin sesgo con la muestra de 50 vale {sx} segundos')

print(f'Obtenemos el valor de n a partir de: 2*valor_critico* sx/raiz_n = 60, es decir: \n'
      f'2*{valor_critico}*{sx}/raiz_n = 60, despejando: \n'
      f'raiz_n = 2*{valor_critico}*{sx}/60')

raiz_n = 2*valor_critico*sx / 60
n_min = np.ceil(raiz_n**2) # el valor entero por encima

print(f'El tamaño mínimo estimado de la muestra es n_min={n_min}')
```

```
# Comentario adiccional
# Podemos repetir el cálculo trabajando con una normal en lugar de una T de Student

# valor crítico de una normal
gama = .95
alfa = 1-gama
z = stats.norm(0,1)
valor_critico = z.ppf(1-alfa/2) # valor positivo
print(f'El valor crítico, según una Z vale {valor_critico}.')

print(f'Obtenemos el valor de n a partir de: 2 * valor_critico * sx/raiz_n = 60, es decir: \n'
      f'2*{valor_critico}*{sx}/raiz_n = 60, despejando: \n'
      f'raiz_n = 2*{valor_critico}*{sx}/60')

raiz_n = 2*valor_critico*sx / 60
n_min = np.ceil(raiz_n**2) # el valor entero por encima

print(f'El tamaño mínimo estimado de la muestra trabajando con una normal es n_min = {n_min}')
```

### Con independencia del valor de n calculado en el apartado anterior, el director de la unidad de analítica de RAZAMAN, viendo el presupuesto disponible, decide comprar la muestra de tamaño n=1000.

```
# Respuesta
# cargamos los datos de la muestra de 1000 datos comprada

df_orig = pd.read_csv('muestra1000.csv')
print('Tamaño de df_orig con los datos cargados:', df_orig.shape)

x1k = df_orig.x.tolist()

plt.figure
plt.hist(x1k,30);plt.xlabel('x (tiempo en segundos)');plt.title('Muestra 1k')
plt.show()
```

### b) Calcular, utilizando la muestra de 1000 datos (variable x1k) y un enfoque clásico, un intervalo de confianza de la media de X con un nivel de confianza del 95%, asumiendo que X se distribuye como una normal. (4 puntos)

```
# Respuesta

n = len(x1k)

print('Para determinar dicho IC, necesitamos estimar:\n'
      '  - la media de X a partir de la muestra de 1k\n'
      '  - el error estándar del estimador media muestral a partir de la muestra de 1k')

mux_1k = np.mean(x1k)
sx_1k = np.std(x1k,ddof=1) # dividiendo por n-1
se_Xmedia = sx_1k/np.sqrt(n)

# creamos la v.a. que indica cómo se distribuye la media muestral
Xmedia_t = stats.t(n-1, loc=mux_1k, scale = se_Xmedia)
print(f'\nv.a. Xmedia_t (t-student): media = {Xmedia_t.mean()} y varianza = {Xmedia_t.var()}')

gama = 0.95
alfa = 1 - gama
lim_inf = Xmedia_t.ppf(alfa/2)
lim_sup = Xmedia_t.ppf(1-alfa/2)

print(f'El intervalo al {100*gama}% de la media de X es: [{lim_inf}, {lim_sup}]')
print(f'Efectivamente, el ancho del IC es {lim_sup - lim_inf} , menor de 60, lo que encaja con el cálculo inicial al ser n bastante más grande.')

# si se utiliza una normal en lugar de la t de Student
Xmedia_n = stats.norm(loc=mux_1k, scale = se_Xmedia)
lim_inf = Xmedia_n.ppf(alfa/2)
lim_sup = Xmedia_n.ppf(1-alfa/2)
print(f'\nv.a. Xmedia_n (Normal): media = {Xmedia_n.mean()} y varianza = {Xmedia_n.var()}')
print(f'El intervalo al {100*gama}% de la media de X es: [{lim_inf}, {lim_sup}]')
print('Como era de esperar, el resultado es muy similar ya que n es grande')
```

### c) El director pide que se estime ahora el intervalo de confianza de la media de X con un nivel de confianza del 95%, pero sin realizar ninguna suposición sobre la distribución de X, para compararlo con el intervalo de confianza estimado en el apartado anterior. (2 puntos)

```
# Respuesta

# Usamos bootstrap con el método BCa
muestra_orig = (x1k,)  # tiene que ser una secuencia

# params para llamar a bootstrap
R = 9999
estadistico = np.mean
nivel_confianza = 0.95

res_BCa = stats.bootstrap(muestra_orig, estadistico, confidence_level=nivel_confianza,  n_resamples=R, method = 'BCA', random_state= 999)
lim_inf_BCa, lim_sup_BCa = res_BCa.confidence_interval

print(f'IC {100*nivel_confianza}% estimado utilizando bootstrap con el método BCa: [{lim_inf_BCa}, {lim_sup_BCa}]')
print('Los resultados son muy similares a los obtenidos con el enfoque clásico, el nuevo intervalo es un poco más ancho.')
print('La suposición realizada sobre la normalidad de la distribución de la variable bajo estudio parece bastante acertada.')
```
