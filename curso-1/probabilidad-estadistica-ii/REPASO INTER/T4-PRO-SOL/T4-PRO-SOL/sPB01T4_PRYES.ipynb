```
# Importar librerías
import pandas as pd
import numpy as np
from scipy import stats
from scipy import optimize as opt  # para estimación por máxima verosimilitud
import matplotlib.pyplot as plt
import seaborn as sns
```

El salario medio en España es de $2128.4$ euros al mes. Se ha realizado una encuesta a  $50$ antiguos alumnos de ICAI que terminaron la universidad hace 2 años, y se han obtenido los siguientes resultados: Los alumnos de ICAI ganan de media $2300$ euros al mes a los dos años de haber terminado la universidad con una (quasi) desviación típica de $100$ euros.

### a) ¿ Podemos afirmar con un nivel de significación del $5\%$ que el salario medio mensual de los alumnos de ICAI es diferente al salario medio mensual de España? Realiza el contraste paso a paso y verifica el resultado haciendo un cálculo directo.

Recordatorio del contraste de hipótesis bilateral para la media poblacional de una v.a. X que sigue una Normal de varianza desconocida:
 $$\begin{array}
H H_0 : & \mu = 2128.4\\
 H_1 : & \mu \neq 2128.4
 \end{array}$$ En este caso rechazamos $H_0$ si:
$$|t_0| > t_{n-1,1-\alpha/2}$$ donde
$$t_0 = \frac{\bar{x} - \mu_0}{s_x/\sqrt{n}} .$$

```
# Respuesta

# los datos que nos da el problema
mu_0 = 2128.4
n = 50

# media muestral
x_media = 2300 
# la desviación típica muestral sx sin sesgo (dividiendo por n-1)
sx = 100

# nivel de significación para el contraste
alpha = 0.05

print('Se trata de un contraste bilateral para la media con varianza desconocida en el que:')
print(f'   H0:   mu =   {mu_0}')
print(f'   H1:   mu !=  {mu_0}')

print(f'Sabemos que la desviación típica muestral vale {sx} y que n = {n}')
```

```
# Calculamos estadistico de contraste en la muestra
t0 = (x_media - mu_0) / (sx/np.sqrt(n))

# crea la distribución del estadístico de contraste
gl = n-1 # grados de libertad
T0 = stats.t(gl)

# calculamos el pvalor (caso H1 BILATERAL)
pvalor = 2 * (1 - T0.cdf(abs(t0)))

print(f'El valor observado del estadistico de contraste es {t0}, con un p-valor asociado igual a {pvalor}.')

# razonando con valores críticos
# nivel de significación
alfa = 0.05

# determina los valores críticos que definen la región de rechazo (bilateral)
t_crit_iz_bilat = T0.ppf(alfa/2)
t_crit_de_bilat = T0.ppf(1-alfa/2)
print(f'Los valores criticos son (alfa={alpha}): {t_crit_iz_bilat, t_crit_de_bilat}')
```

```
# Usemos la regla de rechazo para ver si rechazamos o no H0

if(np.abs(t0) > t_crit_de_bilat):
    print(f"Rechazamos H_0: mu = {mu_0} con un nivel de significación {alfa} ")
    print(f"Ya que |t0| = {np.abs(t0)} es mayor que t_crit_de = {t_crit_de_bilat}.")
    print(f"Se concluye lo mismo viendo que el p-valor = {pvalor} es menor que alfa = {alfa}")
else:
    print(f"No rechazamos H_0: mu = {mu_0} con un nivel de significación {alfa} ")
    print(f"Ya que |t0| = {np.abs(t0)} es menor que t_crit_de = {t_crit_de_bilat}.")
    print(f"Se concluye lo mismo viendo que el p-valor = {pvalor} es mayor que alfa = {alfa}")


print('Como podemos observar hemos rechazado H0 y podemos decir que el salario medio '
      'mensual de los antiguos alumnos de ICAI es diferente al salario medio en España '
      'con un nivel de significación del 5%.')
```

### b) ¿ Podemos afirmar con un nivel de significación del $5\%$ que el salario medio mensual de los alumnos de ICAI es mayor que el salario medio mensual de España? Realiza el contraste paso a paso y verifica el resultado haciendo un cálculo directo.

Para contestar esta pregunta planteamos el siguiente contraste de hipótesis unilateral por la derecha
 $$\begin{array}
H H_0 : & \mu = 2128.4\\
 H_1 : & \mu > 2128.4.
 \end{array}$$ Recuerda que en este caso rechazamos $H_0$ si:
$$t_0 > t_{n-1,1-\alpha}.$$

```
# Respuesta

# nivel de significación para el contraste
alpha = 0.05

print('Se trata de un contraste unilateral por la derecha para la media con varianza desconocida en el que:')
print(f'   H0:   mu =  {mu_0}')
print(f'   H1:   mu >  {mu_0}')

print(f'Sabemos que la desviación típica muestral vale {sx} y que n = {n}')
```

```
# Calculamos estadistico de contraste en la muestra
t0 = (x_media - mu_0) / (sx/np.sqrt(n))

# crea la distribución del estadístico de contraste
gl = n-1 # grados de libertad
T0 = stats.t(gl)

# calculamos el pvalor (caso UNILATERAL DERECHA)
pvalor = 1 - T0.cdf(t0)

print(f'El valor observado del estadistico de contraste es {t0}, con un p-valor asociado igual a {pvalor}.')

# razonando con valores críticos
# nivel de significación
alfa = 0.05

# determina el valor crítico que define la región de rechazo (unilateral derecha)
t_crit_de_unilat = T0.ppf(1-alfa)
print(f'El valor crítico es (alfa={alfa}): {t_crit_de_unilat}')
```

```
# Usemos la regla de rechazo para ver si rechazamos o no H0

if(t0 > t_crit_de_unilat):
    print(f"Rechazamos H_0: mu = {mu_0} con un nivel de significación {alfa} ")
    print(f"Ya que t0 = {t0} es mayor que t_crit_de = {t_crit_de_unilat}.")
    print(f"Se concluye lo mismo viendo que el p-valor = {pvalor} es menor que alfa = {alfa}")
else:
    print(f"No rechazamos H_0: mu = {mu_0} con un nivel de significación {alfa} ")
    print(f"Ya que t0 = {t0} es menor que t_crit_de = {t_crit_de_unilat}.")
    print(f"Se concluye lo mismo viendo que el p-valor = {pvalor} es mayor que alfa = {alfa}")

print('Como podemos observar hemos rechazado H0 y podemos decir que el salario medio '
      'mensual de los antiguos alumnos de ICAI es mayor al salario medio en España '
      'con un nivel de significación del 5%.')
```

### c) ¿ Podemos afirmar con un nivel de significación del $5\%$ que el salario medio mensual de los alumnos de ICAI es menor que el salario medio mensual de España? Realiza el contraste paso a paso y verifica el resultado haciendo un cálculo directo.

Para contestar esta pregunta planteamos el siguiente contraste de hipótesis unilateral por la izquierda
 $$\begin{array}
H H_0 : & \mu = 2128.4\\
 H_1 : & \mu < 2128.4
 \end{array}$$ En este caso rechazamos $H_0$ si:
$$t_0 < t_{n-1,\alpha}.$$

```
# Respuesta

# nivel de significación para el contraste
alfa = 0.05

print('Se trata de un contraste unilateral por la izquierda para la media con varianza desconocida en el que:')
print(f'   H0:   mu =  {mu_0}')
print(f'   H1:   mu <  {mu_0}')

print(f'Sabemos que la desviación típica muestral vale {sx} y que n = {n}')
```

```
# Calculamos estadistico de contraste en la muestra
t0 = (x_media - mu_0) / (sx/np.sqrt(n))

# crea la distribución del estadístico de contraste
gl = n-1 # grados de libertad
T0 = stats.t(gl)

# calculamos el pvalor (caso UNILATERAL IZQUIERDA)
pvalor = T0.cdf(t0)

print(f'El valor observado del estadistico de contraste es {t0}, con un p-valor asociado igual a {pvalor}.')

# razonando con valores críticos
# nivel de significación
alfa = 0.05

# determina el valor crítico que define la región de rechazo (unilateral derecha)
t_crit_iz_unilat = T0.ppf(alfa)
print(f'El valor crítico es (alfa={alfa}): {t_crit_iz_unilat}')
```

```
# Usemos la regla de rechazo para ver si rechazamos o no H0

if(t0 < t_crit_iz_unilat):
    print(f"Rechazamos H_0: mu = {mu_0} con un nivel de significación {alfa} ")
    print(f"Ya que t0 = {t0} es menor que t_crit_iz = {t_crit_iz_unilat}.")
    print(f"Se concluye lo mismo viendo que el p-valor = {pvalor} es menor que alfa = {alfa}")
else:
    print(f"No rechazamos H_0: mu = {mu_0} con un nivel de significación {alfa} ")
    print(f"Ya que t0 = {t0} es mayor que t_crit_iz = {t_crit_iz_unilat}.")
    print(f"Se concluye lo mismo viendo que el p-valor = {pvalor} es mayor que alfa = {alfa}")

print('Como podemos observar no podemos rechazar H0 y no podemos afirmar que el salario medio '
      'mensual de los antiguos alumnos de ICAI es menor al salario medio en España '
      'con un nivel de significación del 5%.')
```

### d) Representa, en tres gráficas diferentes, para cada contraste anterior, la función de densidad de la distribución del estadistico de contraste utilizado si H0 es cierta, las regiones de aceptación y rechazo para un nivel de significación del 5% y valor del estadistico de contraste en la muestra.

```
# Representa contraste bilateral

# puntos para representar la distribución y las regiones de aceptación y rechazo
xmin = T0.ppf(0.000001)
xmax = max(T0.ppf(1-0.000001), t0)
xs = np.linspace(xmin, xmax, 1000)
xRA = np.linspace(t_crit_iz_bilat, t_crit_de_bilat, 1000)
xRRiz = np.linspace(xmin, t_crit_iz_bilat, 1000)
xRRde = np.linspace(t_crit_de_bilat, xmax, 1000)

plt.figure(1, figsize=(20, 6))
plt.plot(xs, T0.pdf(xs), linewidth = 4, label=f'Distribución T0 si H0 es cierta')

plt.fill_between( xRA, T0.pdf(xRA), color="green", alpha=0.6, label=f'Probabilidad asociada RA = {1-alfa}')
plt.fill_between( xRRiz, T0.pdf(xRRiz), color="red", alpha=0.6, label=f'P(T0<{t_crit_iz_bilat}) = {alfa/2})')
plt.fill_between( xRRde, T0.pdf(xRRde), color="red", alpha=0.6, label=f'P(T0>{t_crit_de_bilat}) = {alfa/2}')

plt.vlines(t0, 0, np.max(T0.pdf(xs)), color='black', label=f't0 = {t0}')

plt.hlines(-0.02, xmin, t_crit_iz_bilat, color='red', linewidth = 6, label = f'RR (izq)')
plt.hlines(-0.02, t_crit_iz_bilat, xmax, color='green', label = f'RA: [{t_crit_iz_bilat}, +INF] al {alfa*100}%')
plt.hlines(-0.02, t_crit_de_bilat, xmax, color='red', linewidth = 6, label = f'RR (der)')

plt.title('Constraste BILATERAL')
plt.xlabel('T0')
plt.legend()
plt.show()
```

```
# Representa contraste unilateral derecha

# puntos para representar la distribución y las regiones de aceptación y rechazo
xmin = T0.ppf(0.000001)
xmax = max(T0.ppf(1-0.000001), t0)
xs = np.linspace(xmin, xmax, 1000)
xRA = np.linspace(xmin, t_crit_de_unilat, 1000)
xRRde = np.linspace(t_crit_de_unilat, xmax, 1000)

plt.figure(1, figsize=(20, 6))
plt.plot(xs, T0.pdf(xs), linewidth = 4, label=f'Distribución T0 si H0 es cierta')

plt.fill_between( xRA, T0.pdf(xRA), color="green", alpha=0.6, label=f'Probabilidad asociada RA = {1-alfa}')
plt.fill_between( xRRde, T0.pdf(xRRde), color="red", alpha=0.6, label=f'P(T0>{t_crit_de_unilat}) = {alfa}')

plt.vlines(t0, 0, np.max(T0.pdf(xs)), color='black', label=f't0 = {t0}')

plt.hlines(-0.02, xmin, xmax, color='green', label = f'RA: [-INF, {t_crit_de_unilat}] al {alfa*100}%')
plt.hlines(-0.02, t_crit_de_unilat, xmax, color='red', linewidth = 6, label = f'RR')

plt.title('Constraste UNILATERAL DERECHA')
plt.xlabel('T0')
plt.legend()
plt.show()
```

```
# Representa contraste unilateral izquierda

# puntos para representar la distribución y las regiones de aceptación y rechazo
xmin = T0.ppf(0.000001)
xmax = max(T0.ppf(1-0.000001), t0)
xs = np.linspace(xmin, xmax, 1000)
xRA = np.linspace(t_crit_iz_unilat, xmax, 1000)
xRRiz = np.linspace(xmin, t_crit_iz_unilat, 1000)

plt.figure(1, figsize=(20, 6))
plt.plot(xs, T0.pdf(xs), linewidth = 4, label=f'Distribución T0 si H0 es cierta')

plt.fill_between( xRRiz, T0.pdf(xRRiz), color="red", alpha=0.6, label=f'P(T0<{t_crit_iz_unilat}) = {alfa}')
plt.fill_between( xRA, T0.pdf(xRA), color="green", alpha=0.6, label=f'Probabilidad asociada RA = {1-alfa}')

plt.vlines(t0, 0, np.max(T0.pdf(xs)), color='black', label=f't0 = {t0}')

plt.hlines(-0.02, xmin, t_crit_iz_unilat, color='red', linewidth = 6, label = f'RR')
plt.hlines(-0.02, t_crit_iz_unilat, xmax, color='green', label = f'RA: [{t_crit_iz_unilat}, +INF] al {alfa*100}%')

plt.title('Constraste UNILATERAL IZQUIERDA')
plt.xlabel('T0')
plt.legend()
plt.show()
```
