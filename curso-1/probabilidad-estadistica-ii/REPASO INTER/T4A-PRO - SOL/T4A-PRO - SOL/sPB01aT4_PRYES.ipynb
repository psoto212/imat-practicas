```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
```

### a) Crea tres variables aleatorias con las siguientes características:
- X1: Una v.a. que represente la media muestral de tamaño 20 de una variable poblacional X que sigue una normal de media 10 y varianza 4.
- X2: Una v.a. que represente la media muestral de tamaño 100 de una variable poblacional X que sigue una normal de media 10 y varianza 4.
- X3: Una v.a. que represente la media muestral de tamaño 100 de una variable poblacional Y que sigue una normal de media 10 y varianza 9.

### Una vez creadas las vv.aa., calcula la media y varianza de cada una de las tres variables. ¿Tienen la misma media? ¿Qué variable tiene mayor varianza? Las varianzas de X1 y X2 tienen que ser menores que 4, ¿por qué? ¿Por qué el error estándar de X3 es mayor que el de X2?

```
# Respuesta 

# creamos la v.a. que representan esas medias muestrales
mu_X1 = 10
sigma_X1 = np.sqrt(4/20)
X1 = stats.norm(loc = mu_X1, scale = sigma_X1)

mu_X2 = 10
sigma_X2 = np.sqrt(4/100)
X2 = stats.norm(loc = mu_X2, scale = sigma_X2)

mu_X3 = 10
sigma_X3 = np.sqrt(9/100)
X3 = stats.norm(loc = mu_X3, scale = sigma_X3)

print(f'E[X1]={X1.mean()}, E[X2]={X2.mean()}, E[X3]={X3.mean()}')
print(f'Var(X1)={X1.var()}, Var(X2)={X2.var()}, Var(X3)={X3.var()}')

print('X1 tiene mayor varianza.')
print('Las varianzas de X1 y X2 corresponden a las medias muestrales de la variable X, con tamaños 20 y 100, respectivamente. '
      'Lógicamente, seguro que tienen una varianza menor que la varianza poblacional (4) ya que su varianza es el resultado '
      'de dividir la varianza poblacional por el tamaño muestral.')
print('El error estándar de X3 es mayor que el de X2 porque, aunque en ambas variables se usa el mismo tamaño muestral, '
      'la varianza de la variable poblacional Y es mayor que la de X.')
```

### b) Representa, en una misma gráfica, la función de densidad de las variables X1, X2 y X3 creadas en el apartado anterior. Si X1 y X2 se utilizan para estimar la media de la variable poblacional X, ¿qué estimación será mejor y por qué? ¿Qué media se estimará mejor, la de X usando X2 o la de Y usando X3?

```
# Respuesta 

# Representamos las pdfs
# puntos en los que se evalua las funciones para representarlas ( dados por la de mayor varianza)
xs = np.linspace(X1.ppf(0.001), X1.ppf(0.999), 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(10, 4))

plt.fill_between( xs, X1.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X1.pdf(xs), color="orange", label = 'pdf(X1)')

plt.fill_between( xs, X2.pdf(xs), color="blue", alpha=0.3)
plt.plot( xs, X2.pdf(xs), color="blue", label = 'pdf(X2)')

plt.fill_between( xs, X3.pdf(xs), color="green", alpha=0.3)
plt.plot( xs, X3.pdf(xs), color="green", label = 'pdf(X3)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('X1 y X2 son estimadores insesgados de la media de la variable X.')
print('La estimación proporcionada por X2 será mejor que la de X1 ya que el estimador X2 tiene menor varianza, '
      'lógico al estar construido con una muestra de mayor tamaño.')
print('\nSe estimará mejor la media de X usando X2 que la de Y con X3, ya que el error estándar de X2 es menor que el de X3')
```

### c) Sin hacer ningún cálculo, indica cuál es la probabilidad de que la estimación proporcionada por X1 de la media de X sea menor que 10. ¿Cuál es exactamente la probabilidad de que la estimación de X2 sea superior a 10.5?

```
# Respuesta

print('Dado que X1 se distribuye como una normal de media 10, la P(X1<10) es 0.5 al ser una distribución simétrica centrada en 10')

# calculamos P(X2>10.5)
prob = 1 - X2.cdf(10.5)
print(f'La probabilidad de que la estimación de X2 sea superior a 10.5 es P(X2>10.5) = 1-cdf(10.5) = {prob}')
```

### d) Asumiendo que la media poblacional es desconocida (lo habitual), calcula la probabilidad de que las estimaciones de la media poblacional de X, realizadas con X1 y con X2, estén dentro del intervalo [9.9, 10.1]. Para ello utiliza las distribuciones de las variables X1 y X2. ¿Qué probabilidad es mayor? ¿Qué significa?

```
# Respuesta

# usando las cdfs es directo
probX1 = X1.cdf(10.1) - X1.cdf(9.9)
probX2 = X2.cdf(10.1) - X2.cdf(9.9)

print(f'La probabilidad de que la media poblacional de X esté dentro del intervalo [9.9, 10.1] es:')
print(f'  - Según X1: P(9.9<X1<10.1) = cdf(10.1) - cdf(9.9) = {probX1}')
print(f'  - Según X2: P(9.9<X2<10.1) = cdf(10.1) - cdf(9.9) = {probX2}')
print(f'Es mayor la probabilidad según el estimador X2, es {probX2/probX1} veces mayor.')
print('Quiere decir que si disponemos de un estimador basado en una muestra de datos mayor, tenemos más confianza.')
```
