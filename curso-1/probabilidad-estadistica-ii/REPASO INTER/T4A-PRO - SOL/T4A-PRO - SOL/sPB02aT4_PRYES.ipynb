```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
```

### a) Se dispone de tres v.a. independientes X, Y y Z. Las tres variables se distribuyen como una Normal de media nula y varianza 1. La variable aleatoria V = X + Y + Z, ¿cómo se distribuye? Representa las funciones de densidad de X, Y, Z y V en la misma figura.

```
# Respuesta 

# creamos las v.a. de partida (son independientes)
X = stats.norm(loc = 0, scale = 1)
Y = stats.norm(loc = 0, scale = 1)
Z = stats.norm(loc = 0, scale = 1)

# V = X + Y + Z
mu_V = 0+0+0
sigma2_V = 1 + 1 + 1
sigma_V = np.sqrt(sigma2_V)

V = stats.norm(loc = mu_V, scale =sigma_V)


print('La nueva variable se distribuye como una normal de parámetros:\n'
      f'  - Media: la suma de las medias de X,Y y Z (que son nulas) = {mu_V}\n'
      f'  - Varianza: La suma de las varianzas de X, Y y Z ya que son independientes = {sigma2_V}')
```

```
#Representamos las pdfs
# puntos en los que se evalua las funciones para representarlas (dados por la de mayor varianza)
xs = np.linspace(V.ppf(0.001),V.ppf(0.999), 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(10, 4))

plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X.pdf(xs), color="orange", label = 'pdf(X)')

plt.fill_between( xs, Y.pdf(xs), color="blue", alpha=0.3)
plt.plot( xs, Z.pdf(xs), color="blue", label = 'pdf(Y)')

plt.fill_between( xs, Z.pdf(xs), color="green", alpha=0.3)
plt.plot( xs, Z.pdf(xs), color="green", label = 'pdf(Z)')

plt.fill_between( xs, V.pdf(xs), color="red", alpha=0.3)
plt.plot( xs, V.pdf(xs), color="red", label = 'pdf(V)')

plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('Lógicamente las distribuciones de X, Y y Z son exactamente iguales (verde), pero la pdf de la V tiene mayor varianza.')
```

### e) Calcula la probabilidad de que V sea menor que Z incrementado en 2, es decir P(V<Z+2)

```
# Respuesta

print('V y Z son distribuciones normales. Si V y Z fuesen independientes, '
      'entonces su suma o resta es una distribución normal con parámetros fáciles '
      'de determinar. Sin embargo, como son dependientes el resultado ya no es directo.'
      'Por ejemplo, si ocurriese que V = - Z, entonces V + Z valdría siempre cero.')

print('La forma más sencilla de calcular P(V<Z+2) es tener en cuenta que V=X+Y+Z.')
print('P(V<Z+2) = P(X+Y+Z<Z+2) = P(X+Y<2). '
      'X e Y se distribuyen normalmente, por lo que es sencillo determinar la distribución '
      'de la suma X+Y.')

print('La P(V<Z+2) = P(X+Y<2) = P(U<2), siendo la v.a. U = X+Y, que se distribuye como una normal.')

# creamos la v.a. suma U = X+Y
mu_U = X.mean() + Y.mean()
sigma2_U = X.var() + Y.var()
sigma_U= np.sqrt(sigma2_U)
U = stats.norm(loc = mu_U, scale = sigma_U)

# calculamos 
prob = U.cdf(2)
print(f'La probabilidad P(V<Z) = P(U<2) ={prob}')
```

```
#Representamos
# puntos en los que se evalua las funciones para representarlas (dados por la de mayor varianza)
xs = np.linspace(U.ppf(0.001), U.ppf(0.999), 1000)
x_prob = np.linspace(U.ppf(0.001), 2, 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(15, 6))

plt.fill_between( xs, Z.pdf(xs), color="green", alpha=0.3)
plt.plot( xs, Z.pdf(xs), color="green", label = 'pdf(Z)')

plt.fill_between( xs, V.pdf(xs), color="red", alpha=0.3)
plt.plot( xs, V.pdf(xs), color="red", label = 'pdf(V)')

plt.fill_between( xs, U.pdf(xs), color="black", alpha=0.4)
plt.plot( xs, U.pdf(xs), color="black", label = 'pdf(U)')

plt.fill_between( x_prob, U.pdf(x_prob), color="magenta", alpha=0.3, label = 'P(U<2)')
plt.vlines(2, 0, 0.4, color = 'magenta', label = 'Valor 2')

plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```

```
# Respuesta INCORRECTA

print('Ignorando que V y Z no son independientes se puede plantear que '
      'P(V<Z+2) = P(V-Z<2) = P(U2<2), siendo la v.a. U2 = V-Z, que se distribuye como una normal')

# creamos la v.a. diferencia U = V-Z
mu_U2 = V.mean() - Z.mean()
sigma2_U2 = V.var() + Z.var()
sigma_U2= np.sqrt(sigma2_U2)
U2 = stats.norm(loc = mu_U2, scale = sigma_U2)

# calculamos 
prob2 = U2.cdf(2)
print(f'La probabilidad P(V<Z) = P(U2<0) = {prob2}')
print('Este resultado es incorrecto, el error está en considerar que las variables son independientes')
```
