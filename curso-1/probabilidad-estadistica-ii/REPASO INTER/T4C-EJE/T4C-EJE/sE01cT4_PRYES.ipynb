```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### Sea una v.a. X que se distribuye como una Normal, de la que se dispone de la muestra almacenada en el fichero muestra_n100v01y2022.csv de tamaño n que se guarda en la variable x con el siguiente código: 

```
# cargamos los datos de la muestra disponible

df_orig = pd.read_csv('muestra_n100v01y2022.csv')
print('Tamaño de df_orig con los datos cargados:', df_orig.shape)

# datos de la muestra para trabajar
x = df_orig.x.tolist()    #   <-------------------------------- x (valores en la muestra)
n = len(x)                #   <-------------------------------- n (tamaño de la muestra)

# observamos su distribución
plt.figure
plt.hist(x);plt.xlabel('x');plt.title(f'Muestra (n={n})')
plt.show()
```

### a) Calcula un intervalo de confianza para la media poblacional con un nivel de confianza del 90%. Explica claramente los pasos seguidos para crearlo.

```
# Respuesta

# media muestral
x_media = np.mean(x)

# calculamos la desviación típica muestral sx sin sesgo (dividiendo por n-1)
sx = np.std(x, ddof=1)

# Ahora creamos la v.a. MEDIA MUESTRAL como una t de student
#    De la ayuda de scipy.stats.t :
#    To shift and/or scale the distribution use the loc and scale parameters. 
#    Specifically, t.pdf(x, df, loc, scale) is identically equivalent to t.pdf(y, df) / scale with y = (x - loc) / scale.
Xmedia = stats.t(n-1, loc=np.mean(x), scale = sx/np.sqrt(n))  

#Para calcular un IC de la media de X usamos la distribución teórica de la media muestral
gamma = 0.90
alfa = 1 - gamma
lim_inf = Xmedia.ppf(alfa/2)
lim_sup = Xmedia.ppf(1-alfa/2)

print(f'El intervalo al {100*gamma}% de la media de X es: [{lim_inf}, {lim_sup}]')
```

### b) Representa en la misma gráfica, la función de densidad de la distribución del estadistico utilizado para crear el intervalo de confianza de la media poblacional, y el intervalo de confianza construido. Indica claramente el área asociada al nivel de confianza utilizado. ¿Es simétrico?

```
# Respuesta

# Gráficamente

# puntos para representar la distribución y el intervalo
xs = np.linspace(Xmedia.ppf(0.00001), Xmedia.ppf(1-0.00001), 1000)
xIC = np.linspace(lim_inf, lim_sup, 1000)

plt.figure(1, figsize=(15, 6))
plt.plot(xs, Xmedia.pdf(xs), linewidth = 4, label=f'Distribución Media muestral (según una t)')
plt.fill_between( xIC, Xmedia.pdf(xIC), color="green", alpha=0.3, label=f'Probabilidad asociada ($\gamma=${gamma})')
plt.vlines(x_media, 0, np.max(Xmedia.pdf(xs)), color='black', label=f'Estimación puntual media {x_media}')
plt.errorbar(x_media, -0.01, xerr = np.array([[x_media - lim_inf], [lim_sup - x_media]]), fmt = 'o', label = f'IC [{lim_inf}, {lim_sup}] al {gamma*100}%')
plt.title('Distribución del estimador media muestral de X e IC')
plt.tight_layout()
plt.legend()
plt.show()
```

### c) Representa en la misma gráfica, junto a la función de densidad de la distribución muestral del estadistico pivote para la media poblacional, tres intervalos de confianza para la media poblacional, construidos con niveles de confianza del 85%, 90% y 95%. ¿El incremento del ancho del intervalo de confianza al cambiar del 90% al 95% es igual que el del 85% al 90%? ¿Por qué?

```
# Respuesta

# Para calcular los tres IC de la media de X usamos la distribución teórica de la media muestral
# cambiando simplemente el nivel de confianza

# IC al 85%
gamma85 = 0.85
alfa = 1 - gamma85
lim_inf_85 = Xmedia.ppf(alfa/2)
lim_sup_85 = Xmedia.ppf(1-alfa/2)


# IC al 85%
gamma90 = 0.90
alfa = 1 - gamma90
lim_inf_90 = Xmedia.ppf(alfa/2)
lim_sup_90 = Xmedia.ppf(1-alfa/2)


# IC al 85%
gamma95 = 0.95
alfa = 1 - gamma95
lim_inf_95 = Xmedia.ppf(alfa/2)
lim_sup_95 = Xmedia.ppf(1-alfa/2)

# representamos los tres intervalos juntos

# puntos para representar la distribución y el intervalo
xs = np.linspace(Xmedia.ppf(0.00001), Xmedia.ppf(1-0.00001), 1000)
xIC = np.linspace(lim_inf, lim_sup, 1000)

plt.figure(1, figsize=(15, 4))
plt.plot(xs, Xmedia.pdf(xs), linewidth = 4, label=f'Distribución Media muestral (según una t)')

plt.vlines(x_media, 0, np.max(Xmedia.pdf(xs)), color='black', label=f'Estimación puntual media {x_media}')
plt.errorbar(x_media, -0.1, xerr = np.array([[x_media - lim_inf_85], [lim_sup_85 - x_media]]), fmt = 'o', label = f'IC [{lim_inf_85}, {lim_sup_85}] al {gamma85*100}%')
plt.errorbar(x_media, -0.2, xerr = np.array([[x_media - lim_inf_90], [lim_sup_90 - x_media]]), fmt = 'o', label = f'IC [{lim_inf_90}, {lim_sup_90}] al {gamma90*100}%')
plt.errorbar(x_media, -0.3, xerr = np.array([[x_media - lim_inf_95], [lim_sup_95 - x_media]]), fmt = 'o', label = f'IC [{lim_inf_95}, {lim_sup_95}] al {gamma95*100}%')

plt.title('Distribución del estimador media muestral de X e IC')
plt.tight_layout()
plt.legend()
plt.show()


print('No es necesario calcular el ancho de los tres intervalos para observar que el ancho no aumenta linealmente con el nivel de confianza. '
      'La diferencia de los anchos de los IC 95% - 90% es mayor que la diferencia entre los anchos de los IC 90% - 85%. '
      'Es lógico que sea así ya que cuando se va aumentando el nivel de confianza es necesario alejarse más del centro para conseguir aumentar '
      'el área porque la pdf cada vez toma valores más pequeños.')
```
