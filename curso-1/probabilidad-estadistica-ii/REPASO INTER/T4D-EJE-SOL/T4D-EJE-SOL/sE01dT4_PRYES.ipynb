```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### Sea una v.a. X que se distribuye como una Normal. Se quiere realizar el siguiente contraste de hipótesis para la media de X:
### H0: $\mu_X$ = 130.32
### H1: $\mu_X \neq$ 130.32
### Para ello se dispone de la muestra almacenada en el fichero muestra_n100v01y2022.csv de tamaño n que se guarda en la variable x con el siguiente código: 

```
# cargamos los datos de la muestra disponible

df_orig = pd.read_csv('muestra_n100v01y2022.csv')
print('Tamaño de df_orig con los datos cargados:', df_orig.shape)

# datos de la muestra para trabajar
x = df_orig.x.tolist()    #   <-------------------------------- x (valores en la muestra)
n = len(x)                #   <-------------------------------- n (tamaño de la muestra)

# observamos su distribución
plt.figure
plt.hist(x);plt.xlabel('x');plt.title(f'Muestra (n={n})')
plt.show()
```

### a) Representa en la misma gráfica, la función de densidad de la distribución del estadistico de contraste utilizado si H0 es cierta, las regiones de aceptación y rechazo para un nivel de significación del 5% y valor del estadistico de contraste en la muestra.

```
# Primero cálculos de todo lo necesario

print('Es un contraste para la media de una distribución normal con varianza desconocida.')
print('El tamaño de n no es muy pequeño, pero tampoco muy grande. Lo correcto es hacer un contraste usando una t (prueba t para una muestra).')
print('La hipótesis alternativa es bilateral, por lo que hay una región de rechazo a la derecha y otra a la izquierda.')

# nivel de significación
alfa = 0.05

# creamos la distribución del estadístico de contraste
df = n -1 # grados de libertad
T0 = stats.t(df)

# determinamos los valores críticos que definen las regiones de rechazo
t_crit_iz = T0.ppf(alfa/2)
t_crit_de = T0.ppf(1 - alfa/2)
print(f'\nvalor critico izq = {t_crit_iz}   valor critico dcha = {t_crit_de} (simétricos respecto al cero, claro)')
print(f'(-INF ___RECHAZO___ )[{t_crit_iz} ___ACEPTACIÓN___  {t_crit_de}](  ___RECHAZO___ +INF)')

# calculamos el valor del estadístico de contraste en la muestra
mu0 = 130.32
x_media = np.mean(x)
sx = np.std(x, ddof=1) # dividiendo por n-1

t0 = (x_media - mu0) / (sx/np.sqrt(n))
print(f'\nEstadístico de contraste en la muestra t0 = {t0}')

# comprobamos si t0 está en la zona de rechazo
if t0 < t_crit_iz or t0 > t_crit_de: 
    rechazar = 'SÍ'
else:
    rechazar = 'NO'

print(f't0 en la zona de rechazo: {rechazar}')
```

```
# Ahora representamos

# puntos para representar la distribución y las regiones de aceptación y rechazo
xmin = T0.ppf(0.000001)
xmax = T0.ppf(1-0.000001)
xs = np.linspace(xmin, xmax, 1000)
xRA = np.linspace(t_crit_iz, t_crit_de, 1000)
xRRiz = np.linspace(xmin, t_crit_iz, 1000)
xRRde = np.linspace(t_crit_de, xmax, 1000)

plt.figure(1, figsize=(20, 6))
plt.plot(xs, T0.pdf(xs), linewidth = 4, label=f'Distribución T0 si H0 es cierta')

plt.fill_between( xRA, T0.pdf(xRA), color="green", alpha=0.6, label=f'Probabilidad asociada RA = {1-alfa}')
plt.fill_between( xRRiz, T0.pdf(xRRiz), color="red", alpha=0.6, label=f'P(T0<{t_crit_iz}) = {alfa/2})')
plt.fill_between( xRRde, T0.pdf(xRRde), color="brown", alpha=0.6, label=f'P(T0>{t_crit_de}) = {alfa/2})')

plt.vlines(t0, 0, np.max(T0.pdf(xs)), color='black', label=f't0 = {t0}')

plt.hlines(-0.02, xmin, t_crit_iz, color='red', linewidth = 6, label = f'RR izq')
plt.hlines(-0.02, t_crit_iz, t_crit_de, color='green', label = f'RA: [{t_crit_iz}, {t_crit_de}] al {alfa*100}%')
plt.hlines(-0.02, t_crit_de, xmax, color='red', linewidth = 6, label = f'RR de')
plt.title('Distribución del estimador de contraste')
plt.xlabel('T0')
plt.legend()
plt.show()
```

### b) A partir de la representación anterior, determina si el test es significativo o no, indicando las conclusiones del contraste.

```
# Respuesta

print(f'Dado que t0 está en la región de no rechazo (verde), no se puede rechazar la hipótesis nula al nivel de significación {alfa*100}%.')
print(f'Por tanto, la prueba no es significativa al nivel de significación {alfa*100}%, concluyéndose que la muestra no aporta suficiente '
      f'evidencia para pensar que la media de X sea diferente de {mu0}')
```

### c) Calcula el pvalor a partir del valor t0 y la distribución de T0 bajo el supuesto de H0. Razona, considerando el valor obtenido si hay que rechazar H0 con un nivel de significación del 5%. ¿Hay que rechazar H0 al 10%?

```
# Respuesta

# calculamos el pvalor
pvalor = 2* min([T0.cdf(t0), 1-T0.cdf(t0)])
print(f'El pvalor = {pvalor}')

# concluimos a partir del pvalor y el nivel de significación que se quiera
print(f'Como el pvalor >= {alfa}, no se puede rechazar H0 con un nivel de significación del {alfa*100}%. ' 
      'Lógicamente, esto tiene que ser coherente con los resultados del apartado anterior.')
print('Sin embargo, dado que el pvalor es inferor al 10%, con este nivel de significación sí que '
      'se consideraría el contraste significativo ya que la diferencia de la media observada y '
      'la planteada en H0 son significativamente diferentes, por lo que se puede rechazar H0 al 10% de significación.')
```

### d) Representa en la misma gráfica, la función de densidad de la distribución del estadistico de contraste utilizado si H0 es cierta, las regiones de aceptación y rechazo para un nivel de significación del 5%, el valor del estadistico de contraste en la muestra y el pvalor asociado.

```
# Respuesta

# puntos para representar la distribución y las regiones de aceptación y rechazo
xmin = T0.ppf(0.000001)
xmax = T0.ppf(1-0.000001)
xs = np.linspace(xmin, xmax, 1000)
xRA = np.linspace(t_crit_iz, t_crit_de, 1000)
xRRiz = np.linspace(xmin, t_crit_iz, 1000)
xRRde = np.linspace(t_crit_de, xmax, 1000)

# puntos para pintar el pvalor. Como el valor t0 es menor que cero, representamos el area de la izquierda
xPvalor = np.linspace(xmin, t0, 1000)

plt.figure(1, figsize=(20, 6))
plt.plot(xs, T0.pdf(xs), linewidth = 4, label=f'Distribución T0 si H0 es cierta')

plt.fill_between( xRA, T0.pdf(xRA), color="green", alpha=0.6, label=f'Probabilidad asociada RA = {1-alfa}')

# como el pvalor es mayor que alfa, representamos primero el area del pvalor para que se vea bien
plt.fill_between( xPvalor, T0.pdf(xPvalor), color="black", alpha=1, label=f'P-valor = {pvalor})')
plt.hlines(-0.01, xmin, t0, color='black', linewidth = 6, label=f'Región T0 < {t0}')

plt.fill_between( xRRiz, T0.pdf(xRRiz), color="red", alpha=0.6, label=f'P(T0<{t_crit_iz}) = {alfa/2})')
plt.fill_between( xRRde, T0.pdf(xRRde), color="brown", alpha=0.6, label=f'P(T0>{t_crit_de}) = {alfa/2})')
plt.vlines(t0, 0, np.max(T0.pdf(xs)), color='black', label=f't0 = {t0}')
plt.hlines(-0.02, xmin, t_crit_iz, color='red', linewidth = 6, label = f'RR izq')
plt.hlines(-0.02, t_crit_iz, t_crit_de, color='green', label = f'RA: [{t_crit_iz}, {t_crit_de}] al {alfa*100}%')
plt.hlines(-0.02, t_crit_de, xmax, color='red', linewidth = 6, label = f'RR de')
plt.title('Distribución del estimador de contraste')
plt.xlabel('T0')
plt.legend()
plt.show()
```
