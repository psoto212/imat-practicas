```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### Sea una v.a. X que se distribuye como una Normal. Se quiere realizar el siguiente contraste de hipótesis para la media de X:
### H0: $\mu_X$ = 130.32
### H1: $\mu_X \neq$ 130.32
### Para ello se dispone de la muestra almacenada en el fichero muestra_n100v01y2022.csv de tamaño n que se guarda en la variable x con el siguiente código: 

```
# cargamos los datos de la muestra disponible

df_orig = pd.read_csv('muestra_n100v01y2022.csv')
print('Tamaño de df_orig con los datos cargados:', df_orig.shape)

# datos de la muestra para trabajar
x = df_orig.x.tolist()    #   <-------------------------------- x (valores en la muestra)
n = len(x)                #   <-------------------------------- n (tamaño de la muestra)

# observamos su distribución
plt.figure
plt.hist(x);plt.xlabel('x');plt.title(f'Muestra (n={n})')
plt.show()
```

### a) Indica qué tipo de contraste de hipótesis hay que realizar y calcula rápidamente, de forma directa, el valor del estadístico de contraste con la muestra observada y el p-valor asociado.

```
# Respuesta

print('Es un contraste para la media de una distribución normal con varianza desconocida y una única muestra.')
print('Dado que la varianza poblacional es desconocida, hay que hacer un contraste t para una muestra (ttest_1samp).')
print('La hipótesis alternativa es bilateral (two-sided)')

# media poblacional que se quiere contrastar (H0)
popmean = 130.32
# tioo de laternativa
tipo_alternativa =  'two-sided'

# calculamos el estadístico en la muestra t0 y su pvalor
res = stats.ttest_1samp (x, popmean, alternative = tipo_alternativa) # alternative: ‘two-sided’, ‘less’, ‘greater’
print('Resultado: ', res)

t0 = res.statistic
pvalor = res.pvalue

print(f'El valor observado del estadistico de contraste es {t0}, con un p-valor asociado igual a {pvalor}.')
```

### b) A partir del p-valor calculado, saca conclusiones con un nivel de significación del 5%.

```
# Respuesta

# usando el nivel de significación indicado
alfa = 0.05

# comparamos el pvalor con el nivel de significación
if res.pvalue < alfa: 
    conclusion = f'RECHAZAR H0 ya que p-valor < {alfa}'
else:
    conclusion = f'No se puede rechazar H0 ya que p-valor >= {alfa}'
    
print(f'P-valor = {pvalor}')
print(f'{conclusion}')

# concluimos a partir del pvalor y el nivel de significación indicado
print(f'Conclusión: El contraste no es significativo al {alfa*100}% ya que '
      'la diferencia entre la media observada y la planteada en H0 no son significativamente diferentes, '
      f'por lo que no se puede rechazar H0 al {alfa*100}% de significación. Es decir, no tenemos evidencias significativas al {alfa*100}% de que la media de X sea diferente de {popmean}')
```
