### Se tiene la siguiente función que permite calcular cómodamente el error estándar SE del estimador media muestral para una sigma y n dadas

```
# Función que permite calcular SE para una sigma y n dadas

def se_teo(sigma, n):
    """
    Calcula el error estándar del estimador media muestral.
    - sigma: desviación típica
    - n: tamaño de la muestra
    """
    se = sigma / np.sqrt(n)
    return se
```

```
# Prueba de la función

se = se_teo(1, 100)
print(f'Cuando la sigma={1} y n={100}, se = {se}')
```
