### 0: Carga de librerías necesarias

```
# Importar librerías
import pandas as pd
import numpy as np
from scipy import stats
from scipy import optimize as opt  # para estimación por máxima verosimilitud
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1: Carga de datos

```
# leemos los datos del fichero con las temperaturas usando pandas
# Los datos están en ºC

# df = pd.read_csv('TemperaturesSpain.csv', index_col=['YEAR','MONTH','DATE'])
df_orig = pd.read_csv('TemperaturesSpain.csv')

# parseamos la fecha (cadena) para que sea un datetime con formato dd/mm/yyyy
df_orig["DATE"] = pd.to_datetime(df_orig["DATE"], format='%d/%m/%Y') 
print('Tamaño de df_orig con los datos cargados:', df_orig.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables temperaturas):')
df_orig
```

### 2. Filtrado y selección de variables a estudiar

```
# subconjunto de filas (una lista de meses y un rango de años)
anno_ini = 2000
anno_fin = 2016

mes = 2

# subconjunto de columnas con las que nos quedamos 
variables = ['TMAX3946', 'TMAX3839']
columnas = ['MONTH','YEAR'] + variables

# creamos nuevo df con los datos de los años y vars indicadas
ind_filas = (df_orig.YEAR >= anno_ini) & (df_orig.YEAR <= anno_fin) & (df_orig.MONTH == mes)
df = df_orig.loc[ind_filas, columnas]
print('Tamaño nuevo df_selec:', df.shape)
print(df)

x = df.TMAX3946
y = df.TMAX3839

# tamaño muestra
n = len(x)
```

```
# Distribución empírica de las variables bajo estudio

plt.figure(1, figsize=(15, 6))
plt.hist(x, 40, alpha = 0.5, label=f'Histograma muestra X')
plt.hist(y, 40, alpha = 0.5, label=f'Histograma muestra Y')
plt.legend()
plt.show()

plt.figure(2, figsize=(15, 6))
plt.plot(np.random.permutation(x), '.', label=f'Muestra desordenada tamaño {n}')
plt.plot(np.random.permutation(y), '.', label=f'Muestra desordenada tamaño {n}')
plt.legend()
plt.show()
```
