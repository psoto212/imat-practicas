## 1: Carga de datos

```
# leemos los datos del fichero con las temperaturas usando pandas
# Los datos están en ºC

import pandas as pd

# df = pd.read_csv('TemperaturesSpain.csv', index_col=['YEAR','MONTH','DATE'])
df_orig = pd.read_csv('TemperaturesSpain.csv')

# parseamos la fecha (cadena) para que sea un datetime con formato dd/mm/yyyy
df_orig["DATE"] = pd.to_datetime(df_orig["DATE"], format='%d/%m/%Y') 
print('Tamaño de df_orig con los datos cargados:', df_orig.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables temperaturas):')
df_orig
```

## 2: Análisis exploratorio preliminar de alguna variable
Sirve para entender rápidamente los datos que se quieren estudiar

```
# importamos los paquetes necesarios
import matplotlib.pyplot as plt
plt.rcParams['figure.dpi'] = 75

# mostramos toda la serie de la min y max de 229
fig = plt.figure(1,figsize=(15,4))
plt.plot(df_orig.DATE, df_orig.TMIN229, label= 'TMIN229')
plt.plot(df_orig.DATE, df_orig.TMAX229, label= 'TMAX229')
ax=plt.title('Evolución temporal de las temperaturas min y max en una misma estación')
plt.legend()

# mostramos un año de las series de Tmin y Tmax de 229

# nuevo df con los datos de solo 2010
df_2010 = df_orig.loc[df_orig['YEAR'] == 2010]

fig = plt.figure(2,figsize=(15,4))
plt.plot(df_2010.DATE, df_2010.TMIN229, label= 'TMIN229')
plt.plot(df_2010.DATE, df_2010.TMAX229, label= 'TMAX229')
plt.grid()
ax=plt.title('Detalle año 2010')
plt.legend()

# guardamos la figura en un fichero gráfico para poder utilizarla
plt.savefig('seriesTminTmax229.jpg')
```

```
# miremos la distribución de tmin y tmax de la estación 229
import seaborn as sns
import numpy as np

fig = plt.figure(1,figsize=(10,4))
# histograma
plt.subplot(2,1,1)
ax = sns.histplot(data=df_orig.TMAX229, color = 'orange', bins=40)
ax=sns.rugplot(data=df_orig.TMAX229, color='b')
# ponemos la media y la std en el título
print('media: ', np.mean(df_orig.TMAX229), ' std: ', np.std(df_orig.TMAX229))

# boxplot
plt.subplot(2,1,2)
sns.violinplot(data=df_orig.TMAX229, color= 'orange', orient='y')
sns.rugplot(data=df_orig.TMAX229, color='b')
ax=plt.xlabel('TMAX229')
```

```
# relación entre la tmin y tmax de la estación 229

fig = plt.figure(1,figsize=(6,6))
plt.plot(df_orig.TMIN229, df_orig.TMAX229, 'k.')
plt.xlabel('TMIN229')
plt.ylabel('TMAX229')
plt.grid()
ax=plt.title('Relación entre Tmin y Tmax en una misma estación')

fig = plt.figure(2,figsize=(6,6))
ax=sns.displot(data=df_orig, x='TMIN229', y='TMAX229', bins=(40,40), color = 'orange', cbar=True,  rug=True) # with rug!!
plt.grid()
plt.title('Histograma 2D')
plt.xlabel('TMIN229')
ax=plt.ylabel('TMAX229')
```

## *4: Selección de datos (MUY IMPORTANTE)
Trabajamos con un subconjunto de meses y años (filas) y variables (columnas)

```
## MUY IMPORTANTE !!
# -----------------------------------------------------------
# subconjunto de filas (una lista de meses y un rango de años)
anno_ini = 2010
anno_fin = 2015

# subconjunto de columnas con las que nos quedamos 
variables = ['TMIN3922', 'TMIN1392', 'TMIN1396', 'TMIN230', 'TMIN237', 'TMIN3959',
            'TMAX3922', 'TMAX1392', 'TMAX1396', 'TMAX230', 'TMAX237', 'TMAX3959']

columnas = ['MONTH','YEAR'] + variables

# también se puede hacer así:
#nomb_estac = ['3922', '1392', '1396',  '230', '237', '3959']
#nomb_tmin  = [f"TMIN{s}" for s in nomb_estac]
#nomb_tmax  = [f"TMAX{s}" for s in nomb_estac]
#columnas = ['MONTH','YEAR'] + nomb_tmin + nomb_tmax
#print(columnas)
# -----------------------------------------------------------

# creamos nuevo df con los datos de los años y vars indicadas
ind_filas = (df_orig.YEAR >= anno_ini) & (df_orig.YEAR <= anno_fin)
df_selec = df_orig.loc[ind_filas, columnas]
print('Tamaño nuevo df_selec:', df_selec.shape)
#print(df_selec)

# pintamos los datos seleccionados para chequeo rápido, ignorando el mes y el año que no aportan aquí
df_selec.drop(columns=['MONTH','YEAR']).plot(figsize=(15,4))
```

## 5: Métricas univariantes principales de las variables

```
# vista general de todas las variables 
print('Métricas')
df_selec.describe(percentiles=[0.05,0.25,0.50,0.75,0.95])
```

### Métricas concretas de las variables

```
# vista general de todas las variables 
print('Mediana:')
print(df_selec.median())
print('Coeficiente de variación:')
print(df_selec.std() / df_selec.mean())
print('Coeficiente de asimetría:')
print(df_selec.skew())
print('Coeficiente de apuntamiento:')
print(df_selec.kurtosis())
```

## 6: Distribuciones de las variables

```
# histrogramsa de las variables, compartiendo eje x para poder comparar
df_selec.drop(columns=['YEAR','MONTH']).hist(figsize=(15,8), bins = 20, sharex = True)
plt.draw()
```

```
# diagramas de cajas de cada temperatura
df_selec.drop(columns=['YEAR','MONTH']).boxplot(figsize=(8,4), rot=45)
plt.show()
```

## 7: Análisis de correlaciones

```
# matriz de correlaciones creada directamente con el df
corrMatrix = df_selec.drop(columns=['YEAR','MONTH']).corr()
print(np.round(corrMatrix,2))

fig = plt.figure(2,figsize=(8,8))
sns.heatmap(corrMatrix, annot=True)
ax = plt.title('Matriz de correlaciones')
```

```
# Matriz de nubes de dispersión
fig = plt.figure(1,figsize=(10,10))
ax=sns.pairplot(df_selec, markers='.', height=2.)
```

## 8: Distribución según el mes (uso de groupby)

```
# creamos nuevo df con los datos del mes y las temperatuas a considerar
df_mes_temp = df_selec.loc[:, ['MONTH', 'TMIN230', 'TMAX230']]

# mostramos un boxplot para cada mes de esta temperatura
df_mes_temp.boxplot(figsize=(10,6),by='MONTH')
plt.show()

# obtenemos los valores numerícos para poder cuantificarlo
print('Medianas por meses:')
print(df_mes_temp.groupby('MONTH').median())

print('Mínimos por meses:')
print(df_mes_temp.groupby('MONTH').min())

# o toda la descripción
print('Principales métricas')

df_mes_temp.groupby('MONTH').describe()
```

```
from scipy import stats

# cogemos los datos de un mes en concreto para analizarlo en profundidad
mes = 3
temp_un_mes = df_mes_temp.loc[df_mes_temp.MONTH==mes, 'TMIN230']

print('El rango intercuartílico de TMIN230 en marzo es:', stats.iqr(temp_un_mes), 'grados')
```

## 9: Creamos una temperatura como una media ponderada

```
temp_promedio = 0.5 * df_selec.TMIN230 + 0.5 * df_selec.TMAX230

# mostramos las originales y la nueva serie
plt.figure(1,figsize=(15,5))
plt.subplot(1,2,1)
plt.plot(df_selec.TMIN230,'-')
plt.plot(df_selec.TMAX230,'-')
plt.plot(temp_promedio,'.-')

# los boxplots
plt.subplot(1,2,2)
vars = [df_selec.TMIN230, df_selec.TMAX230, temp_promedio]
sns.boxplot(data=[d for d in vars])

# comparamos varianzas
print('Varianza de TMIN230:   ', np.var(df_selec.TMIN230))
print('Varianza de TMAX230:   ', np.var(df_selec.TMAX230))
print('Varianza de Tpromedio: ', np.var(temp_promedio))
var_promedio = 0.25*np.var(df_selec.TMIN230)+0.25*np.var(df_selec.TMAX230)+2*0.25*np.cov(df_selec.TMIN230,df_selec.TMAX230)[0,1]
print('Var(Tpromedio) según (0.25*var(tmin)+0.25*var(tmax)+2*0.5*0.5*cov(tmin,tmax): ', var_promedio)
```
