## Cargamos los datos con los que se trabaja

```
nom_fi_datos_Velmed = 'G07A_REG1_VELMED.csv'
nom_fi_datos_Velvar = 'G07A_REG1_VELVAR.csv'
```

```
df_orig_Velmed = pd.read_csv(nom_fi_datos_Velmed)

# parseamos la fecha (cadena) para que sea un datetime con formato yyyy/mm/dd 
df_orig_Velmed['FECHA'] = pd.to_datetime(df_orig_Velmed['FECHA'], format='%d/%m/%Y')

s = df_orig_Velmed.loc[:,'FECHA']
df_orig_Velmed['FECHA'] =  s.dt.date

print('Tamaño de df_orig_Velmed con los datos cargados:', df_orig_Velmed.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables):')
print(df_orig_Velmed)
```

```
df_orig_Velvar = pd.read_csv(nom_fi_datos_Velvar)

# parseamos la fecha (cadena) para que sea un datetime con formato yyyy/mm/dd 
df_orig_Velvar['FECHA'] = pd.to_datetime(df_orig_Velvar['FECHA'], format='%d/%m/%Y')

s = df_orig_Velvar.loc[:,'FECHA']
df_orig_Velvar['FECHA'] =  s.dt.date

print('Tamaño de df_orig_Velvar con los datos cargados:', df_orig_Velvar.shape)
print('Dataframe con todos los datos leidos (filas: días, columnas: variables):')
print(df_orig_Velvar)
```

### 1. Visión general de las series

```
# auxiliares útiles para la representación
fechas = df_orig_Velmed.FECHA.tolist()
```

```
fig = plt.figure(1, figsize=(15,6)) # permite indicar el nº de la figura y las dimensiones (ancho y alto)
ax1 = plt.subplot(2,1,1)
plt.plot(fechas, df_orig_Velmed.VELMED01, label='VELMED01')
plt.plot(fechas, df_orig_Velmed.VELMED13, label='VELMED13')
plt.title(nom_fi_datos_Velmed)
plt.legend()

plt.subplot(2,1,2,sharex= ax1)
plt.plot(fechas, df_orig_Velvar.VELVAR01, label='VELVAR01')
plt.plot(fechas, df_orig_Velvar.VELVAR13, label='VELVAR13')
plt.title(nom_fi_datos_Velvar)
plt.tight_layout()
plt.legend()

plt.show()

fig.savefig('figura1.pdf')
```
