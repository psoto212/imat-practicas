# Plus one

You are given a large integer represented as an integer array digits, where each digits[i] is the ith digit of the integer. The digits are ordered from most significant to least significant in left-to-right order. The large integer does not contain any leading 0's.

Increment the large integer by one and return the resulting array of digits.

```
def plusOne(digits: list[int]) -> list[int]:
    num = int(''.join([str(d) for d in digits])) + 1
    return [int(d) for d in str(num)]
```

```
def plusOne(digits: list[int]) -> list[int]:
    for i in range(len(digits) - 1, -1, -1):
        if digits[i] == 9:
            digits[i] = 0
        else:
            digits[i] += 1
            return digits
    # [0,0,0]
    # digits.insert(0, 1)
    digits = [1] + digits
    # digits.append(1)[::-1]
    # [0,0,0,1]
    return digits
```

```
def plusOne(digits: list[int]) -> list[int]:
    new_list = ([0] + digits)[::-1]
    nine = True
    for i in range(len(new_list)):
        if not nine:
            break
        if (new_list[i] + 1) < 10:
            new_list[i] += 1
            nine = False
        else:
            new_list[i] = 0
    return new_list[::-1][1:] if new_list[-1] == 0 else new_list[::-1]
```

### Test Cases

```
assert plusOne([1, 2, 3]) == [1, 2, 4]
assert plusOne([4, 3, 2, 1]) == [4, 3, 2, 2]
assert plusOne([0]) == [1]
assert plusOne([9]) == [1, 0]
assert plusOne([9, 9]) == [1, 0, 0]
assert plusOne([9, 9, 9]) == [1, 0, 0, 0]
print("All tests passed.")
```
