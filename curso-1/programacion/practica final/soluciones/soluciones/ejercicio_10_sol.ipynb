# Best Time to Buy and Sell Stock

You are given an array prices where prices[i] is the price of a given stock on the ith day.

You want to maximize your profit by choosing a single day to buy one stock and choosing a different day in the future to sell that stock.

Return the maximum profit you can achieve from this transaction. If you cannot achieve any profit, return 0.

 

Example 1:<br>

Input: prices = [7,1,5,3,6,4]<br>
Output: 5<br>
Explanation: Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6-1 = 5.<br>
Note that buying on day 2 and selling on day 1 is not allowed because you must buy before you sell.<br>

```
def maxProfit(prices: list[int]) -> int:
    if prices == []:
        return 0
    min_tuple = min(enumerate(prices), key=lambda x: x[1])
    first_value = max(prices[min_tuple[0]:]) - min_tuple[1]
    max_tuple = max(enumerate(prices), key=lambda x: x[1])
    second_value = max_tuple[1] - min(prices[:max_tuple[0]+1]) 

    return max([first_value, second_value, maxProfit(prices[max_tuple[0]+1:min_tuple[0]])])

```

```
def maxProfit(prices: list[int]) -> int:    
    min_price = float('inf')
    max_profit = 0
    
    for price in prices:
        if price < min_price:
            min_price = price

        current_profit = price - min_price
        
        if current_profit > max_profit:
            max_profit = current_profit
            
    return max_profit
```

```
# Test cases
assert maxProfit([7,1,5,3,6,4]) == 5
assert maxProfit([7,6,4,3,1]) == 0
assert maxProfit([1,2]) == 1
assert maxProfit([2,4,1]) == 2
assert maxProfit([3,2,6,5,0,3]) == 4
assert maxProfit([3,3,5,0,0,3,1,4]) == 4
assert maxProfit([1,2,4,2,5,7,2,4,9,0]) == 8
assert maxProfit([1,2,4,2,5,7,2,4,9,0,9]) == 9
assert maxProfit([7,2,4,6,1]) == 4
print("All test cases passed!")
```
