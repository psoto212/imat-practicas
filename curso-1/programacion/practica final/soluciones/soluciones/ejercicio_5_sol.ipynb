# Longest Palindromic Substring

Given a string s, return the longest 
palindromic substring in s.


Example 1:

Input: s = "babad"
Output: "bab"
Explanation: "aba" is also a valid answer.
Example 2:

Input: s = "cbbd"
Output: "bb"
 

Constraints:

1 <= s.length <= 1000
s consist of only digits and English letters.

```
def longestPalindrome(s: str) -> str:
    if len(s) == 0:
        return ""
    
    def expand_around_center(left: int, right: int) -> str:
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return s[left + 1:right]
    
    longest_palindrome = ""
    
    for i in range(len(s)):
        # Odd length palindromes (single character center)
        odd_palindrome = expand_around_center(i, i)
        if len(odd_palindrome) > len(longest_palindrome):
            longest_palindrome = odd_palindrome
        
        # Even length palindromes (consecutive characters center)
        even_palindrome = expand_around_center(i, i + 1)
        if len(even_palindrome) > len(longest_palindrome):
            longest_palindrome = even_palindrome
    
    return longest_palindrome
```

```
def longestPalindrome(s: str) -> str:
    n = len(s)
    if n == 0:
        return ""
    
    # Create a 2D table to store palindrome information
    dp = [[False] * n for _ in range(n)]
    start = 0
    max_length = 1
    
    # All substrings of length 1 are palindromes
    for i in range(n):
        dp[i][i] = True
    
    # Check for substrings of length 2
    for i in range(n - 1):
        if s[i] == s[i + 1]:
            dp[i][i + 1] = True
            start = i
            max_length = 2
    
    # Check for substrings longer than 2
    for length in range(3, n + 1): # length is the length of the substring
        for i in range(n - length + 1): # i is the starting index of the substring
            j = i + length - 1 # j is the ending index of the substring
            if s[i] == s[j] and dp[i + 1][j - 1]:
                dp[i][j] = True
                start = i
                max_length = length
    
    return s[start:start + max_length]
```

```
# Test cases
assert longestPalindrome("babad") == "aba" or longestPalindrome("babad") == "bab"
assert longestPalindrome("cbbd") == "bb"
assert longestPalindrome("a") == "a"
assert longestPalindrome("ac") == "a"
assert longestPalindrome("bb") == "bb"
assert longestPalindrome("ccc") == "ccc"
assert longestPalindrome("aaaa") == "aaaa"
assert longestPalindrome("") == ""
# Complicated test cases
assert longestPalindrome("abacdfgdcaba") == "aba"
assert longestPalindrome("abacdfgdcabba") == "abba"
assert longestPalindrome("abacdfgdcabbaaaa") == "abba" or longestPalindrome("abacdfgdcabbaaaa") == "aaaa"
assert longestPalindrome("abacdfgdcabbaaaaa") == "aaaaa"
assert longestPalindrome("abdbabdbbadbbbadbbadbabbdba") == "bdbabdb" 
print("All test cases passed!")
```
