# Counting Word Frequencies in a Text File

Write a Python program that reads a text file, counts the frequency of each word in the file, and writes the word frequencies to another text file in descending order of frequency.

The program should:
-	Read the content of `input_text.txt`.
-	Count the frequency of each word (case insensitive).
-	Write the word frequencies to a new file named `word_frequencies.txt` in descending order of frequency.
-	Words should be treated as case insensitive, and punctuation should be ignored.


An example of the output file `word_frequencies.txt` is:
```
this: 10
is: 8
file: 6
sample: 2
text: 2
frequencies: 2
word: 2
a: 1
for: 1
counting: 1
```

````
def counting_words(input_file: str, output_file: str) -> None:
    """
    Counting words in a file.
    The function reads the file and counts the number of words in it.
    Then it writes the result to the output file.

    Args:
        input_file: path to the input file
        output_file: path to the output file

    Returns:
        None

    """

    # Read the input file
    with open(input_file, 'r') as file:
        text = file.read()

    # Remove punctuation and convert to lowercase
    text = text.lower()

    # Split the text into words
    words = text.split()

    # Count the frequency of each word
    word_counts = {}
    for word in words:
        word = word.strip('.,!?')
        if word in word_counts:
            word_counts[word] += 1
        else:
            word_counts[word] = 1

    # Sort the word frequencies in descending order
    sorted_word_counts = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)

    # Write the word frequencies to the output file
    with open(output_file, 'w') as file:
        for word, count in sorted_word_counts:
            file.write(f"{word}: {count}\n")

counting_words("../files/input_text.txt", "output.txt")
````
