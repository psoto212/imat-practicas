# Classroom Management System

You are tasked with creating a simple classroom management system using Python. This system should be able to handle the basic functionalities of managing a class of students, including adding students, removing students, and saving the class list to a file. Additionally, the system should be able to read from a file to load the student list.

Requirements:

-	Create a class Student with the following attributes:
	-	name (str): The name of the student.
	-	id (str): The unique ID of the student.

-	Create a class Classroom with the following attributes and methods:
	-	students (list): A list to store Student objects.
	-	add_student(student) (method): Adds a Student object to the students list.
	-	remove_student(student_id) (method): Removes a Student object from the students list by their id.
	-	save_to_file(filename) (method): Saves the list of students to a file. Each student’s information should be written on a new line in the format id: name.
	-	load_from_file(filename) (method): Loads students from a file and populates the students list. Each line in the file is in the format id: name.
	-	add_exam (method): Adds an exam to the class. The exam will be read from a text file and the results will be stored in an Exam object.
	-	get_exam_results(student_id) (method): Returns the exam results for a student.
	-	get_best_student (method): Returns the student with the highest average score in all exams.

-	Create a class Exam with the following attributes:
	-	name (str): The name of the exam.
	-	path (str): The path to the file containing the exam results.
	-	get_results (method): Returns the exam scores for all students.
	-	get_average_score (method): Returns the average score of the exam.
	-	get_highest_score (method): Returns the highest score of the exam.
	-	get_lowest_score (method): Returns the lowest score of the exam.

-	Ensure that the file handling methods (save_to_file and load_from_file) properly handle file operations, including error handling for cases like file not found or read/write errors.

```
class Student:
    def __init__(self, name, student_id):
        self.name = name
        self.id = student_id
```

```
class Exam:
    def __init__(self, name, path):
        self.name = name
        self.path = path
        self.results = self.get_results()

    def get_results(self):
        results = {}
        try:
            with open(self.path, 'r') as file:
                for line in file:
                    student_id, score = line.strip().split(':')
                    results[student_id] = float(score)
        except FileNotFoundError:
            print(f"Error: The file {self.path} was not found.")
        return results

    def get_average_score(self):
        if not self.results:
            return 0
        return sum(self.results.values()) / len(self.results)

    def get_highest_score(self):
        if not self.results:
            return 0
        return max(self.results.values())

    def get_lowest_score(self):
        if not self.results:
            return 0
        return min(self.results.values())
```

```
class Classroom:
    def __init__(self, name):
        self.name = name
        self.students: list[Student] = []
        self.exams: list[Exam] = []

    def add_student(self, student):
        self.students.append(student)

    def remove_student(self, student_id: str):
        self.students = [student for student in self.students if student.id != student_id]

    def save_to_file(self):
        with open(self.name + '.txt', 'w') as file:
            for student in self.students:
                file.write(f'{student.id}: {student.name}\n')

    def load_from_file(self, filename):
        try:
            with open(filename, 'r') as file:
                self.students = []
                for line in file:
                    student_id, name = line.strip().split(': ')
                    self.students.append(Student(name, student_id))
        except FileNotFoundError:
            print(f"Error: The file {filename} was not found.")

    def add_exam(self, exam: Exam):
        self.exams.append(exam)

    def get_exam_results(self, student_id: str):
        results = {}
        for exam in self.exams:
            if student_id in exam.results:
                results[exam.name] = exam.results[student_id]
        return results

    def get_best_student(self):
        best_student = None
        highest_average = -1
        for student in self.students:
            results = self.get_exam_results(student.id)
            if results:
                average_score = sum(results.values()) / len(results)
                if average_score > highest_average:
                    highest_average = average_score
                    best_student = student
        return best_student
```

```

# Initialize classroom
classroom = Classroom('Lions')

# Load students from file
classroom.load_from_file('../files/students.txt')

# Add an exam
exam1 = Exam('Math', '../files/exam1.txt')
classroom.add_exam(exam1)

# Add another exam
exam2 = Exam('Science', '../files/exam2.txt')
classroom.add_exam(exam2)

# Add a student
student = Student('Dave', '4')
classroom.add_student(student)

# Get exam results for a student
student_id = '1'
results = classroom.get_exam_results(student_id)
print(f"Exam results for student {student_id}: {results}")

# Get the best student
best_student = classroom.get_best_student()
print(f"Best student: {best_student.name} (ID: {best_student.id})")

# Show the best student's exam results
results = classroom.get_exam_results(best_student.id)
print(f"Exam results for the best student: {results}")


# Save the updated students list to file
classroom.save_to_file()
```
