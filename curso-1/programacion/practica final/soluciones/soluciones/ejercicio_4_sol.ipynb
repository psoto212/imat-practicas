# 3 Sum

Given an integer array nums, return all the triplets [nums[i], nums[j], nums[k]] such that i != j, i != k, and j != k, and nums[i] + nums[j] + nums[k] == 0.

Notice that the solution set must not contain duplicate triplets.

```
def check_duplicates(terns: list[list[int]]) -> list[list[int]]:
    final_terns = []
    if len(terns) == 0 or len(terns) == 1:
        return terns
    for i in range(len(terns)):
        set_i = set(terns[i])
        isin = False
        for j in final_terns:
            set_j = set(j)
            if set_i == set_j:
                isin = True
                break
        if not isin:
            final_terns.append(terns[i])
    return final_terns


    

def threeSum(nums: list[int]) -> list[list[int]]:
    terns = []
    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            for k in range(j+1, len(nums)):
                if nums[i] + nums[j] + nums[k] == 0:
                    terns.append([nums[i], nums[j], nums[k]])
    return check_duplicates(terns)
    
```
