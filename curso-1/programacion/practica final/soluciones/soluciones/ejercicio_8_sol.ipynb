# Permutation Sequence

The set [1, 2, 3, ..., n] contains a total of n! unique permutations.

By listing and labeling all of the permutations in order, we get the following sequence for n = 3:

"123"
"132"
"213"
"231"
"312"
"321"
Given n and k, return the kth permutation sequence.

```
def getPermutation(n: int, k: int) -> str:
    # Compute the factorials up to (n-1)!
    factorials = [1] * n
    for i in range(1, n):
        factorials[i] = factorials[i - 1] * i
    
    # Create a list of numbers to get indices
    numbers = list(range(1, n + 1))
    
    # Adjust k to be zero-indexed
    k -= 1
    
    # Build the k-th permutation
    permutation = []
    for i in range(n, 0, -1):
        # Determine the current factorial level
        factorial_level = factorials[i - 1]
        
        # Determine the index of the current digit
        index = k // factorial_level
        k %= factorial_level
        
        # Append the number at the index
        permutation.append(numbers.pop(index))
    
    # Convert the permutation list to a string and return
    return ''.join(map(str, permutation))
```

```
# Test cases
assert getPermutation(3, 3) == '213'
assert getPermutation(4, 9) == '2314'
assert getPermutation(3, 1) == '123'
assert getPermutation(4, 1) == '1234'
assert getPermutation(9, 10) == '123457896'
assert getPermutation(6, 700) == '651342'
assert getPermutation(9, 362880) == '987654321'
print('All test cases pass')
```
