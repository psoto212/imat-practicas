The transaction file should have the following format:
```
Date: 2023-05-26 12:00:00
Transaction: Deposit
Amount: 100
Balance: 100

Date: 2023-05-26 12:14:36
Transaction: Withdraw
Amount: 50
Balance: 50

Date: 2024-05-26 23:42:49.588299
Transaction: Transfer Out
Amount: 25
Balance: 25

...
```

The file should be named `transactions.txt` and should be saved in the same directory as the script. Each transaction should be appended to the file. Also, transfers should be recorded in both accounts as Transfer In and Transfer Out.

To get the date you can use:
```python

```

````
import datetime

class BankAccount:
    account_counter = 0  # Class variable to keep track of the number of accounts

    def __init__(self):
        BankAccount.account_counter += 1  # Increment the counter each time a new account is created
        self.account_id = BankAccount.account_counter 
        self.balance = 0
        self.transaction_file = f'transactions{self.account_id}.txt'

    def deposit(self, amount: int):
        if amount > 0:
            self.balance += amount
            self.log_transaction('Deposit', amount)
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount: int):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                self.log_transaction('Withdraw', amount)
            else:
                print("Insufficient funds.")
        else:
            print("Withdraw amount must be positive.")

    def get_balance(self):
        return self.balance

    def transfer(self, amount: int, other_account: 'BankAccount'):
        if amount > 0:
            if amount <= self.balance:
                self.balance -= amount
                other_account.deposit(amount)
                self.log_transaction('Transfer Out', amount)
                other_account.log_transaction('Transfer In', amount)
            else:
                print("Insufficient funds.")
        else:
            print("Transfer amount must be positive.")

    def log_transaction(self, transaction_type: str, amount: int):
        with open(self.transaction_file, 'a') as file:
            file.write(f"Date: {datetime.datetime.now()}\n")
            file.write(f"Transaction: {transaction_type}\n")
            file.write(f"Amount: {amount}\n")
            file.write(f"Balance: {self.balance}\n")
            file.write("\n")
````

````
account1 = BankAccount()
account2 = BankAccount()

account1.deposit(100)
account1.withdraw(50)
account1.transfer(25, account2)

print(f"Account 1 Balance: {account1.get_balance()}")
print(f"Account 2 Balance: {account2.get_balance()}")
````
