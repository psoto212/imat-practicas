# Exact Pattern Matching

Given an input string (s) and a pattern (p), implement wildcard pattern matching with support for '?' and '*' where:

'?' Matches any single character.<br>
'*' Matches any sequence of characters (including the empty sequence) and there can only be <strong>1</strong> * in the pattern.<br>
The matching should cover the entire input string (not partial).<br>



```
def isMatch(s: str, p: str) -> bool:
    if (len(s) == len(p)) or ('*' in p):
        if not '*' in p:
            for char1, char2 in zip(s, p):
                if char2 != '?' and char1 != char2:
                    return False
        else:
            p_splitted = p.split('*')
            for char1, char2 in zip(s[:len(p_splitted[0])], p_splitted[0]):
                if char2 != '?' and char1 != char2:
                    return False
            for char1, char2 in zip(s[-len(p_splitted[-1]):][::-1], p_splitted[-1][::-1]):
                if char2 != '?' and char1 != char2:
                    return False
        return True
    return False
```

```
# Test Cases
assert isMatch("aa", "a") == False
assert isMatch("aa", "*") == True
assert isMatch("cb", "?a") == False
assert isMatch("adceb", "a*b") == True
assert isMatch("acdcb", "a*c?b") == False
assert isMatch("acdcb", "a*d?b") == True
assert isMatch("a", "a") == True
print("All test cases passed!")
```
