```
def capitalize(string):
    string = string[0].upper() + string[1:].lower()
    return string
```

```
def nombres_limpios(nombres):
    return capitalize(nombres)
```

```
def edad_limpia(edad):
    clean = ''
    for char in edad:
        if char.isdigit():
            clean += char
    return int(clean)

        
```
