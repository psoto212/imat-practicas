```
s1 = "Esto es una prueba."
s2 = "Esta. asune purebo"
cadena1_ord = []
cadena2_ord = []
for i in s1:
    cadena1_ord.append(ord(i))
    cadena1_ord.sort()
    for j in s2:
        cadena2_ord.append(j)
        cadena2_ord.sort()
if cadena1_ord == cadena2_ord:
    print("si tienen los mismos caracteres")
else:
    print("no tienen los mismos caracteres")
```

```
def generar_combinaciones(caracteres, longitud):
    # Genera todas las combinaciones de 'longitud' caracteres
    if longitud == 0:
        return ['']
    combinaciones = []
    for i in range(len(caracteres)):
        char = caracteres[i]
        resto_caracteres = caracteres[:i] + caracteres[i+1:]
        for combinacion in generar_combinaciones(resto_caracteres, longitud - 1):
            combinaciones.append(char + combinacion)
    return combinaciones

def generar_todas_palabras(caracteres):
    resultados = set()
    # Genera combinaciones de diferentes tamaños
    for longitud in range(1, len(caracteres) + 1):
        combinaciones = generar_combinaciones(caracteres, longitud)
        for combinacion in combinaciones:
            resultados.add(combinacion)
    return resultados

# Ejemplo de uso
caracteres = 'ab'
resultado = generar_todas_palabras(caracteres)
print(resultado)
```

```
def contar_caracteres(palabra):
    contador = {}
    for char in palabra:
        if char in contador:
            contador[char] += 1
        else:
            contador[char] = 1
    return contador

def es_subconjunto(palabra1, palabra2):
    # Contar la frecuencia de cada carácter en ambas palabras
    contador1 = contar_caracteres(palabra1)
    contador2 = contar_caracteres(palabra2)
    
    # Verificar si cada carácter en palabra2 está en palabra1 con al menos la misma frecuencia
    for char in contador2:
        if char not in contador1 or contador2[char] > contador1[char]:
            return False
    return True

```
