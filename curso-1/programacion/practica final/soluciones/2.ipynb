```
def palindromo(palabra=None):
    if palabra is None:
        return False  # Retorna False si no se proporciona ninguna palabra

    # Eliminamos los acentos de la palabra
    acentos = {
        'á': 'a', 'é': 'e', 'í': 'i', 'ó': 'o', 'ú': 'u', 'Á': 'A', 'É': 'E', 'Í': 'I', 'Ó': 'O', 'Ú': 'U',
    }
    palabra = ''.join(acentos.get(caracter, caracter) for caracter in palabra)

    # Convertimos la palabra a minúsculas y eliminamos los espacios y signos de puntuación
    palabra = palabra.lower().replace(' ', '').replace(',', '').replace('.', '').replace(';', '').replace(':', '').replace('!', '').replace('¡', '').replace('?', '').replace('¿', '').replace('-', '').replace('_', '').replace('(', '').replace(')', '').replace('[', '').replace(']', '').replace('{', '').replace('}', '').replace('"', '').replace("'", '').replace('@', '').replace('#', '').replace('$', '').replace('%', '').replace('&', '').replace('*', '').replace('+', '').replace('=', '').replace('/', '').replace('\\', '').replace('|', '').replace('<', '').replace('>', '')

    # Comprobamos si la palabra es un palíndromo
    if palabra == palabra[::-1]:
        return True  # Es un palíndromo
    else:
        return False  # No es un palíndromo

# Probamos las frases proporcionadas
frases = [
    "Ají traga la lagartija",
    "Allí ves a Sevilla",
    "Allí ves a Seevilla",
    "Allí por la tropa portado, traído a ese paraje de maniobras, una tipa como capitán usar boina me dejara, pese a odiar toda tropa por tal ropilla",
    "Allí si María avisa y así va a ir a mi silla",
    "Allí si María avisa y así va a no ir a mi silla",
    "Allí va Ramón y no maravilla",
    "Allí ves a Sevilla",
]

for frase in frases:
    resultado = palindromo(frase)
    print(f'La frase "{frase}" es un palíndromo: {resultado}')
```

```
def cantidad_unos_binario(numero):
    """Función para calcular la cantidad de unos en la representación binaria de un número."""
    contador = 0
    while numero:
        contador += numero & 1
        numero >>= 1
    return contador

def calcular_avellanas(N):
    """Función para calcular la máxima cantidad de avellanas que Adrián podría obtener."""
    max_avellanas = 0
    for x in range(1, N):
        y = N - x
        avellanas = cantidad_unos_binario(x) + cantidad_unos_binario(y)
        max_avellanas = max(max_avellanas, avellanas)
    return max_avellanas

def main():
    with open('valores_N.txt', 'r') as file:
        num_valores_N = int(file.readline().strip())
        valores_N = [int(file.readline().strip()) for _ in range(num_valores_N)]

    resultados = []
    for N in valores_N:
        max_avellanas = calcular_avellanas(N)
        resultados.append(max_avellanas)

    for resultado in resultados:
        print(resultado)

if __name__ == "__main__":
    main()
```

```
def es_capicua(numero):
    """Verifica si un número es capicúa."""
    return str(numero) == str(numero)[::-1]

def es_primo(numero):
    """Verifica si un número es primo."""
    if numero <= 1:
        return False
    for i in range(2, int(numero**0.5) + 1):
        if numero % i == 0:
            return False
    return True

def encontrar_menor_capicua_primo(N):
    """Encuentra el menor número capicúa y primo mayor o igual a N."""
    while True:
        if es_capicua(N) and es_primo(N):
            return N
        N += 1

encontrar_menor_capicua_primo(100000)
```
