## Pregunta 5. Programación (45 min -  5,5 puntos)

Responde en **OTRA HOJA**.

Desarrollar un programa en Python que calcule la frecuencia (número de repeticiones) de los caracteres que aparecen en una cadena dada y las muestre de forma ordenada descendentemente, de mayor a menor grado de aparición.

Se partirá de una cadena ya inicializada, como por ejemplo:
```
cadena = "cadaad"
```
Y el resultado a mostrar será las veces que aparece el caracter "c", "a" y "d", pero de forma ordenada por las veces que aparece en la cadena, su frecuencia:
```
a-> 3
d-> 2
c-> 1
```

**OJO**: el número máximo de caracteres repetidos en una cadena será de 9.

### Input
El examen se realizará con la siguiente cadena de muestra:
```
Cadena de entrada: adcdabbdbcdbdc
```
### Output
La **salida del programa** deberá ser exactamente la que se muestra a continuación:
```
Tabla de frecuencia: d:5,b:4,c:3,a:2,
```

### Solución

Rúbrica de calificación:
- Cuenta de un caracter en una cadena. 1
- Cálculo de la frecuencia. 1
- Almacenamiento y recuperación de las frecuencias.  2
- Ordenación. 1,5

````
cadena = "adcdabbdbcdbdc"

caracteres_unicos = ""
frecuencia = ""

print("Cadena de entrada:", cadena)

i = 0
while i < len(cadena):
    # Detectamos caracteres únicos en la cadena para no crear duplicados de frecuencias
    letra = cadena[i]
    j = 0
    encontrado = False
    while j < len(caracteres_unicos) and not encontrado:
        if caracteres_unicos[j] == letra:
            encontrado = True
        j += 1

    if not encontrado:
        caracteres_unicos += letra  # Agregamos el caracter único
        
        # Cálculo de la frecuencia
        cuenta = 0
        k = 0
        while k < len(cadena):
            if cadena[k] == letra:
                cuenta += 1
            k += 1

        # Ordenación de las frecuencias
        k = 0
        pos_final = 0
        while k < len(frecuencia):
            num1 = int(frecuencia[k+2])
            if cuenta > num1:
                pos_final = k
                k = len(frecuencia)
            else:
                k += 4
                pos_final = k

        frecuencia = frecuencia[:pos_final] + f"{letra}:{cuenta}," + frecuencia[pos_final:]

    i += 1

print("Tabla de frecuencia:", frecuencia)
````
