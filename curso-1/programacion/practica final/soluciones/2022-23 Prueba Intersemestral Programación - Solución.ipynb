````
cadena = "AabbBCcddD"
contador = 0

while contador < len(cadena)-1:
    if cadena[contador].lower() == cadena[contador+1].lower() and contador % 2 == 0:
        cont_bin = bin(contador)[2:4]  # la función bin(2) devuelve 
        if cont_bin == "10":
            print(contador, ">", cadena[contador:contador+2].lower())
    contador += 1
````

## Pregunta 5. Programación (45 min -  5 puntos)

Desarrollar un programa en Python que realice la funcionalidad que se describe a continuación. Responde en **otra hoja**.

Se parte de dos frases que se suminstran a modo de ejemplo, pero el programa a realizar debe funcionar para cualquier tipo de palabras y de cualquier tamaño.

El objetivo del programa es mezclar, fusionar o intercalar las dos frases para que cambien su significado: 

El proceso de intercalado de las palabras será recorriendo las frases, provocando saltos a la palabra contraria cuando nos encontremos con una palabra que tenga más de tres caracteres (de forma muy burda, no será una "preposición", "artículo", "adverbio", ...). Los acrónimos, como por ejemplo IA (todas sus letras en mayúsculas), serán tratados como palabras de salto.

Ejemplo de la **IZQUIERDA**:

Como todas las palabras tienen más de 3 caracteres, se producirán saltos en todas las palabras. Las palabras formadas serán una secuencia de los colores. Matemáticas, ya sufre el primer salto.

Ejemplo de la **DERECHA**:

Como las palabras "es" y "un" no tienen más de 3 caracteres siguen en la frase de la palabra anterior que produjo el cambio, es decir, siguen con "Perro" en la frase_final2. 

Ejemplo de frases largas:

```
frase1 = "Python, un lenguaje de programación muy utilizado en desarrollar los algortimos de IA"
frase2 = "Seat, un fabricante de vehículos muy prestigioso en España de modelos de automóviles"

frase1_final = "Seat, un lenguaje de vehículos muy utilizado en España de algortimos de automóviles"
frase2_final = "Python, un fabricante de programación muy prestigioso en desarrollar los modelos de IA"
```

### Solución

Rúbrica de calificación:
- Identifica palabras en la cadena (busca espacios) de una palabra. 1,5
- Lleva la cuenta paralela de las dos cadenas identificando palabras en las dos cadenas. 2,25
- Condición de salto de palabras: longitud.  0,5
- Acrónimo. 0,5
- Concatenación de la nueva palabra. 0,25

````
frase1 = "Python, un lenguaje de programación muy utilizado en desarrollar los algortimos de IA"
frase2 = "Seat, un fabricante de vehículos muy prestigioso en España de modelos de automóviles"

#frase1_final = "Seat, un lenguaje de vehículos muy utilizado en España de algortimos de automóviles"
#frase2_final = "Python, un fabricante de programación muy prestigioso en desarrollar los modelos de IA"

#frase1 = "Perro es un animal"
#frase2 = "Juan es muy listo"

#frase1 = "Matemáticas están on-fire"
#frase2 = "Inteligencia Artificial iMAT"

frase1_final = ""
frase2_final = ""

sw = True

longitud = len(frase2)
if len(frase1) > len(frase2):
    longitud = len(frase1)

contador1 = 0
contador2 = 0
while contador1 < longitud:
    palabra1 = ""
    while contador1 < len(frase1) and frase1[contador1] != " ":
        palabra1 += frase1[contador1]
        contador1 += 1
    
    palabra2 = ""    
    while contador2 < len(frase2) and frase2[contador2] != " ":
        palabra2 += frase2[contador2]
        contador2 += 1
    
    if len(palabra1) > 3 and len(palabra2) > 3 or (palabra1 == palabra1.upper() or palabra2 == palabra2.upper()):
        sw = not sw
    
    if sw:
        frase1_final += palabra1
        frase2_final += palabra2
    else:
        frase1_final += palabra2
        frase2_final += palabra1
        
    frase1_final += " "
    frase2_final += " "

    contador1 += 1
    contador2 += 1

print(frase1_final)
print(frase2_final)
````
