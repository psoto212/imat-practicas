```
def EsPrimo(n):
    if n <= 1:
        return 0
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return 0
    return 1

def separar_primos(arr):
    primos = [x for x in arr if EsPrimo(x)]
    no_primos = [x for x in arr if not EsPrimo(x)]
    return primos + no_primos

# Ejemplo de uso:
array = [4, 25, 14, 1, 13, 62, 18, 2, 13, 52, 11, 22, 14, 16, 7, 9, 10]
resultado = separar_primos(array)
print(resultado)  # Salida: [7, 11, 13, 13, 2, 1, 4, 25, 14, 62, 18, 52, 22, 14, 16, 9, 10]
```

```
class Nodo:
    def __init__(self, value):
        self.clave = value
        self.color = 1  # 1 for red, 0 for black
        self.padre = None
        self.izq = None
        self.dcha = None

class Arbol:
    def __init__(self):
        self.raiz = None

    def altura(self, nodo):
        if nodo is None:
            return -1
        else:
            izquierda = self.altura(nodo.izq)
            derecha = self.altura(nodo.dcha)
            return 1 + max(izquierda, derecha)
    
    def profundidad(self, nodo):
        profundidad = 0
        while nodo.padre is not None:
            nodo = nodo.padre
            profundidad += 1
        return profundidad

# Example usage
nodo1 = Nodo(1)
nodo2 = Nodo(2)
nodo3 = Nodo(3)
arbol = Arbol()
arbol.raiz = nodo1
nodo1.izq = nodo2
nodo2.padre = nodo1
nodo1.dcha = nodo3
nodo3.padre = nodo1

print("Altura del árbol:", arbol.altura(arbol.raiz))
print("Profundidad del nodo2:", arbol.profundidad(nodo2))
```
