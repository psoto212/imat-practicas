### Programa #1. Bucles nivel básico

Calcular el factorial y el sumatorio de un número dado.

```
numero = 5

El sumatorio de 5 es 15
El factorial de 5 es 120
```

**Solución en la celda siguiente**

`````
numero = 5
contador = 1
sumatorio = 0
factorial = 1

while contador <= numero:
    sumatorio = sumatorio + contador
    factorial = factorial * contador
    contador = contador + 1

print ("el sumatorio de", numero, "es", sumatorio)
print("el factorial de", numero, "es", factorial)


`````

### Generar una lista aleatoria de 20 elementos en el rango [1, 10]. Obtener un número objetivo de forma aleatoria en el rango [1, 10].

Indicar qué parejas de valores consecutivos de la lista suman el número aleatorio objetivo.

La ejecución del programa deberá generar una salida similar a la siguiente.

Salida

`````
import random
lista = []
i = 0
for i in range(20):
    rnd = random.randint(1, 10)
    lista.append(rnd)
numero_objetivo = random.choice(lista)
print(lista)
print(f"el numero objetivo es: {numero_objetivo}")

for i in range(len(lista) - 1):
    for j in range(i+1,len(lista)-1,1):
        if lista[i] + lista[j] == numero_objetivo:
            print(f" ({lista[i]} + {lista[j]}) = {numero_objetivo}")
`````

### Desarrollar un programa que permita representar gráficamente dos funciones matemáticas.



Los distintos puntos a obtener de las funciones se almacenarán en listas y se obtendrán en el rango del eje X [-5, +5] cada 0.25, calculando en total 41 puntos.

Será obligatorio trabajar con la función range() y todos los bucles se generarán con estructuras for in. Estará prohibido trabajar con while.

Para representar las funciones se trabajará con la librería matplotlib de la siguiente forma:

import matplotlib.pyplot as plt

plt.plot(eje_x, f_x)  # eje_x y f_x son listas.
plt.plot(eje_x, g_x)  # eje_x y g_x son listas.
plt.show()

`````
import matplotlib.pyplot as plt

X_MIN = -5
X_MAX = 5

eje_x = []
f_x = []
g_x = []

for x in range(X_MIN, X_MAX):
    eje_x.append(x)
    for ticks in range(1, 4):
        eje_x.append(x + ticks/4)

eje_x.append(X_MAX)

for x in eje_x:
    f_x.append(x**2 - 3*x + 5)
         
for x in eje_x:
    g_x.append(-5*x**2 + 4*x + 25)

plt.plot(eje_x, f_x)
plt.plot(eje_x, g_x)
plt.show()

print("Tabla de puntos de la función f(x): (x, y)")

cont = 0
for x in eje_x:
    print(f"({x}, {f_x[cont]})")
    cont += 1
`````
