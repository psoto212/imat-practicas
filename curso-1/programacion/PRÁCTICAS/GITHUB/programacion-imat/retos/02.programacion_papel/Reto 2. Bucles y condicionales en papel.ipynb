# Reto 2. Bucles y condicionales en papel (Tema 4)

Programar **en papel** un código que responda a la lógica descrita en el enunciado y comprobar mentalmente si funciona. 
Una vez depurado en papel, compartir con tu compañero para que detecte errores. A continuación, escribir el código en un Notebook y comprobar su funcionalidad. Por último, compartirlo con ChatGPT para refinar el estilo de programación.

## Objetivos

Dominar el flujo de un programa mediante estrucuturas de control repetitivas y condicionales.

## Funcionalidad

1. Pedir dos números al usuario (a y b), de forma repetitiva, hasta que b sea mayor que a. Mostrar un mensaje de error cuando no se cumpla esta condición. Ver Salida.
2. El programa de forma automática forzará que los dos números tengan valor par. Indicar al usuario que se han corregido los valores a un valor par. Ver Salida. 
3. Incrementar el valor de b en 2 unidades hasta que b sea múltiplo de a.
4. Mostrar el valor final de los dos números.

## Salida

```
Número a: 5
Número b (b > a): 3
b debe ser mayor que a
Número a: 3
Número b (b > a): 5
Numeros corregidos a par con los valores a: 4  b: 6
Programa finalizado con los valores a: 4  b: 8
```
