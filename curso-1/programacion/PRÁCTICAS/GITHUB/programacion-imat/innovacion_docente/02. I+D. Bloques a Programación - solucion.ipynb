## Pseudocódigo

```
si rebajas = "si":
    si manzanas = "si":
        comprar manzanas
    sino 
        si naranjas = "si":
            comprar naranjas
    comprar zumo
    mientras bolsa < 1kg:
        agregar magdalena

```

## Python

````
# Definición del contexto del supermercado
rebajas = "si"
manzanas = "no"
naranjas = "si"
peso_magdalena = 100

if rebajas == "si":
    if manzanas == "si":
        print("Comprar manzanas")
    else:
        if naranjas == "si":
            print("Comprar naranjas")
            
    print("Comprar zumo")
    
    print("Pesando Magdalenas")
    bolsa = 0
    while bolsa < 1000:
        print("  + Metiendo una magdalena en la bolsa...")
        bolsa = bolsa + peso_magdalena
    print("Comprar bolsa de magdalenas de", bolsa, "gramos")    
    print("Pagar")
print("Vuelvo a casa")
````
