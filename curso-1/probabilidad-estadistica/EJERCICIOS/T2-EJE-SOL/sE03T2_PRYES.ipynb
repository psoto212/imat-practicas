### Dos amigos sólo utilizan el correo electrónico o el chat para enviar mensajes. Utilizan el 60% del tiempo el correo electrónico para comunicarse. La probabilidad de que un mensaje enviado por correo electrónico sea finalmente leído es de 0.90, mientras que la probabilidad de que un mensaje enviado por chat sea finalmente leído es de 0.95.

### a) Determina la probabilidad de que un mensaje enviado sea finalmente leído. Justifica la respuesta.

```
# respuesta

print('Definimos los sucesos:')
print('  mail: el mensaje se manda por correo electrónico')
print('  chat: el mensaje se manda por chat')
print('  leer: lee el mensaje enviado')

# datos de partida
Pr_mail = 0.6
Pr_chat = 1 - Pr_mail

Pr_leer_cond_mail = 0.9
Pr_leer_cond_chat = 0.95

print(f'Se pide P(leer), la probabilidad de leer un mensaje enviado, sabiendo que P(leer|mail) = {Pr_leer_cond_mail} y que P(leer|chat) = {Pr_leer_cond_chat}, ',
      f'además de conocer las probabilidades de que el mensaje se haya enviado por mail p(mail)={Pr_mail}, o por chat p(chat)={Pr_chat}.')

# Calculamos la probabilidad pedida usando ley probabilidad total
# p(read)=p(read|mail)P(mail)+p(read|chat)p(chat)

Pr_leer = (Pr_leer_cond_mail * Pr_mail) + (Pr_leer_cond_chat * Pr_chat)

print('Usando la ley de probabilidad se puede calcular fácilmente, ya que existe una partición disjunta del espacio '
      'muestral (mail o chat):')
print('   P(leer) = P(leer|mail) P(mail) + P(leer|chat) P(chat) = ')
print(f'   P(leer) = {Pr_leer_cond_mail} * {Pr_mail} + {Pr_leer_cond_chat} * {Pr_chat} = {Pr_leer_cond_mail * Pr_mail} + {Pr_leer_cond_chat * Pr_chat} = {Pr_leer}')
      
```
