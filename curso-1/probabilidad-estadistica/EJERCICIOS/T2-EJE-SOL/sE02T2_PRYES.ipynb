### El 30% de las personas que viajan en avión entre dos ciudades concretas lo hacen con cierta aerolínea que pierde el equipaje del 0.1% de sus pasajeros. 

### a) Calcula la probabilidad de que una persona elija dicha aerolínea y desaparezca su equipaje. Justifica la respuesta.

```
# respuesta

print('Se puede viajar o no con esa aerolínea. Dicha aerolínea puede o no perder el equipaje.')

print('Definimos los sucesos:')
print('  A: viaja con esa compañía')
print('  B: desaparece el equipaje')

# datos de partida
Pr_A = 0.3
Pr_B_cond_A = 0.001

print('\nSe tienen las probabilidades:')
print(f'  Prob. de elegir dicha aerolínea: P(A)= {Pr_A}')
print(f'  Prob. de perder el equipaje sujeto a que se a elegido dicha aerolínea: P(B|A)= {Pr_B_cond_A}')


print('\nSe pide P(A y B), la probabilidad de la intersección de A y B, que se puede calcular a partir de las condicionadas:')
print('P(A y B) = P(A|B) P(B) = P(B|A) P(A)')


Pr_AyB = Pr_B_cond_A * Pr_A

print(f'Como tenemos P(B|A), entonces usamos P(A y B) = P(B|A) P(A) = {Pr_B_cond_A} * {Pr_A} = {Pr_AyB}')

print(f'Por tanto, la probabilidad de que una persona elija dicha aerolínea y desaparezca su equipaje vale {Pr_AyB}')
```
