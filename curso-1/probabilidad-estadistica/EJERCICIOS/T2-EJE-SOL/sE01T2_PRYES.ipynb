### En una clase en la que todos practican algún deporte, el 60% de los alumnos juega al fútbol o al baloncesto y el 10% practica ambos deportes. Además hay un 60% que no juega al fútbol.

### a) ¿Cuál es la probabilidad de que, escogido al azar un alumno de la clase, juegue sólo al fútbol? Justifica la respuesta.

```
# respuesta

print('Definimos los sucesos:')
print('  F: juega al fúltol')
print('  B: juega al baloncesto')

# datos de partida
Pr_FoB = 0.6
Pr_FyB = 0.1
Pr_noF = 0.6

print('\nSe tienen las probabilidades:')
print(f'  P(F o B)= {Pr_FoB}')
print(f'  P(F y B)= {Pr_FyB}')
print(f'  P(no F)= {Pr_noF}')

print('\nSe pide la probabilidad de jugar únicamente al fútbol, es decir, jugar al fútbol y no al baloncesto: P(F y noB).')
print('Se puede calcular así: P(F y noB):')
print('|----------------- OMEGA ------------------------|')
print('      |----------- F --------------|')
print('                      |---------- B ------|')
print('                      |-- F y B  --|')
print('      |--  F y no B --|')

print('\nEs decir: P(F y noB) = P(F) - P(F y B)')

# la probabilidad de que juegue al futbol
Pr_F = 1 - Pr_noF

# si ahora queremos que solo juegue al futbol, le quitamos la probabilidad de jugar al futbol y al baloncesto a la vez
Pr_solo_F = Pr_F  - Pr_FyB

print('\nLa probabilidad de jugar al futbol es P(F) = 1 - P(no F) = ', Pr_F)
print(f'Por tanto, la probabilidad de jugar únicamente al futbol es P(F y noB) = P(F) - P(F y B) = {Pr_F} - {Pr_FyB} = {Pr_solo_F}')
```

### b) ¿Cuál es la probabilidad de que, escogido al azar un alumno de la clase, juegue sólo al baloncesto? Justifica la respuesta.

```
# respuesta

print('Ahora se pide P(B y noF):')
print('|----------------- OMEGA -----------------------------------|')
print('      |----------- F --------------|')
print('                      |---------- B -----------------|')
print('                      |-- F y B  --|')
print('                                   |--- B y no F ----|')

print('\nEs decir, hay que calcular P(B y noF) = P(B) - P(F y B)')
print('Para calcular P(B) se puede obtener de la probabilidad de la unión: P(F o B) = P(F) + P(B) - P(F y B)')

# probabilidad de que juegue al baloncesto
Pr_B = Pr_FoB - Pr_F + Pr_FyB

print(f'Despejando: P(B) = P(F o B) - P(F) + P(F y B) = {Pr_FoB} - {Pr_F} + {Pr_FyB} = {Pr_B}')

# por tanto, ahora le quitamos la probabilidad de la intersección como antes y tenemos la probabilidad solicitada
Pr_solo_B = Pr_B - Pr_FyB

print(f'\nFinalmente, la probabilidad pedida de jugar solo a baloncesto es P(B y noF) = P(B) - P(F y B) = {Pr_B} - {Pr_FyB} = {Pr_solo_B}')
```

### d) ¿Cuál es la probabilidad de que, escogido al azar un alumno de la clase, no juegue ni al fútbol ni al baloncesto? Justifica la respuesta.

```
# respuesta

print('Se pide calcular P(noF y noB). Esta probabilidad se calcula fácilmente usando las leyes de De Morgan:')

prob = 1- Pr_FoB
print(f'P(noF y noB) = P(no(F o B)) = 1 - P(F o B) = 1 - {Pr_FoB} = {prob}')
```
