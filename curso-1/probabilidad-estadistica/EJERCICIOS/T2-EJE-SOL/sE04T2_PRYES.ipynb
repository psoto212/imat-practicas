### Se quiere estudiar la situación laboral de los trabajadores en cuatro sectores de la economía que denotaremos por B1, B2, B3 y B4. La probabilidad de que una persona esté sin trabajo, que denotaremos por el suceso S, en cada uno de los sectores es 0.05, 0.01, 0.02 y 0.1 respectivamente. Adicionalmente se sabe que la mitad de las personas pertenecen al primer sector, y el resto se reparten a partes iguales entre el resto de sectores. 

### a) Calcular la probabilidad de que una persona elegida al azar esté en paro. Justifica la respuesta.

```
# respuesta

# Sucesos:
print('Definimos los sucesos:')
print('  S: estar en paro')
print('  B1: pertenecer al sector 1 de la economía')
print('  B2: pertenecer al sector 2 de la economía')
print('  B3: pertenecer al sector 3 de la economía')
print('  B4: pertenecer al sector 4 de la economía')

# datos de partida
Pr_S_cond_B1 = 0.05
Pr_S_cond_B2 = 0.01
Pr_S_cond_B3 = 0.02
Pr_S_cond_B4 = 0.1

Pr_B1 = 0.5
Pr_B2 = 0.5/3
Pr_B3 = 0.5/3
Pr_B4 = 0.5/3

print('\nSe tienen las probabilidades:')
print(f'  P(S|B1) = {Pr_S_cond_B1}')
print(f'  P(S|B2) = {Pr_S_cond_B2}')
print(f'  P(S|B3) = {Pr_S_cond_B3}')
print(f'  P(S|B4) = {Pr_S_cond_B4}')

print(f'  P(B1) = {Pr_B1}')
print(f'  P(B2) = {Pr_B2}')
print(f'  P(B3) = {Pr_B3}')
print(f'  P(B4) = {Pr_B4}')

print('\nSe pide P(S). Para ello se puede utilizar el teorema de la probabilidad total ya que los sucesos B1 a B4 forman '
      'una partición disjunta del espacio muestral considerado, y se conocen las probabilidades a priori y las verosimilitudes.')

Pr_S = Pr_S_cond_B1*Pr_B1 + Pr_S_cond_B2*Pr_B2 + Pr_S_cond_B3*Pr_B3 + Pr_S_cond_B4*Pr_B4 

print('La probabilidad de estar en paro entonces se puede calcular como:')
print('P(S) = P(S|B1)*P(B1) + P(S|B2)*P(B2) + P(S|B3)*P(B3) + P(S|B4)*P(B4)')
print(f'P(S) = {Pr_S_cond_B1*Pr_B1} + {Pr_S_cond_B2*Pr_B2} + {Pr_S_cond_B3*Pr_B3} + {Pr_S_cond_B4*Pr_B4} = {Pr_S}')
```
