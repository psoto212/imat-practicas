```

# respuestas
df_mes_temp = df_selec.loc[:,  ['TMIN3971', 'TMAX3912', 'TMIN3947', 'TMIN3913', 'TMIN232']]

# mostramos un boxplot para cada mes de esta temperatura
df_mes_temp.boxplot(figsize=(10,6),by='MONTH')
plt.show()

# obtenemos los valores numerícos para poder cuantificarlo
print('Medianas por meses:')
print(df_mes_temp.groupby('MONTH').median())

print('Mínimos por meses:')
print(df_mes_temp.groupby('MONTH').min())

# o toda la descripción
print('Principales métricas')
df_mes_temp.groupby('MONTH').describe()
```
