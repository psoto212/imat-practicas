## Distribuciones de probabilidad discretas

stats.rv_discrete, stats.rv_discrete.expect

```
# import required packages
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

### Sea la v.a. X indicada por sus posibles valores y probabilidades asociadas

```
# v.a. X de partida

# valores posibles de la v.a. (los pisos)
xk = [1, 2, 3, 4, 5, 6]

# probabilidad de parar en cada piso
pk = [0.15, 0.1, 0.15, 0.25, 0.25, 0.1]

# comprobacion sum(pk) = 1
print('Verificar que 1 = sum(pks)=',sum(pk))

# crea la v.a. utilizando los valores posibles y su probabilidad (xk,pk)
X = stats.rv_discrete(values=(xk, pk))
```

### a) Representa la distribución de la v.a. X e intenta imaginar el valor esperado de la v.a. observando las gráficas, sin realizar ningún cálculo.

```
# respuesta

# puntos en los que se evalua la función para representarla
xs = range(-2, 10)

# en formato tabla
df =  pd.DataFrame(list(zip(xs,  X.pmf(xs), X.cdf(xs))), columns=['xs', 'pmf(xs)', 'cdf(xs)'])
print(df)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, X.pmf(xs), basefmt='none')
plt.ylim(0, 1)
plt.ylabel('pmf(x)')

plt.subplot(2,1,2, sharex = ax1)
plt.xlim(xs[0], xs[-1])
plt.step(xs, X.cdf(xs), color='red', where='post')

plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()

plt.show()
```

### b) Calcula el valor esperado de la v.a. X y compáralo con tu estimación anterior.

```
# respuesta

media_X = X.mean()
print('La media de X vale', media_X)
```

### c) Calcula el valor esperado de la v.a. Y = 100 + 7 * X, usando únicamente las propiedades del operador esperanza matemática (sin utilizar expect)

```
# respuesta

print('Hay que calcular E[Y]')
print('E[Y]=100 + 7*E[X]')
print('Usando la E[X]=', media_X)
print('E[Y]=100 + 7*E[X] =', 100 + 7*media_X)
```

### d) Calcula el valor esperado de la v.a. Y = 100 + 7 * X, pero ahora utilizando la función expect. Compara con el cálculo exacto anterior.

```
# respuesta 

g_X = lambda x: 100 + 7 * x

# prueba de la función para verificar que es lo que queremos
a= 10
print('Prueba de mi_funcion(',a,')=', g_X(a))

# calcula el valor experado de g(X), siendo X la v.a.
val_esperado = X.expect(g_X)
print('Valor esperado de Y=', val_esperado)

# Compara el resultado con el apartado anterior
print('Se obtiene, en este caso, exactamente el mismo valor esperado que cuando '
      'se calcula de forma analítica con la esperanza matemática.')
```
