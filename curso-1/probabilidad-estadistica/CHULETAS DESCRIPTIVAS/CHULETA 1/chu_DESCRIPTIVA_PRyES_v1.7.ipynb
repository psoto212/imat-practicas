## Librerías disponibles

```
# Importar librerías
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```
# variables tipo array para pruebas de la chuleta
x = (np.abs(np.random.randn(20)))**1.7
y = np.random.randn(20)
z = x - 0.5*np.random.randn(20)

# variables tipo lista para pruebas de la chuleta
#x=[1,2,3,5,6,6,8,8,8,8,8,9,9,10,11,1]
#y=[6,7,8,5,6,6,6,3,3,3,3,2,2,1,1,10]
#z=[6,6,5,7,8,8,4,6,9,9,9,12,11,10,4,5]
```

## 1. Medidas estadísticas de una variable x

### 1.1 Medidas de centralización: media, mediana y moda

```
me = np.mean(x)    # media
md = np.median(x)  # mediana

mo = stats.mode(x) # moda

print('x:', x)
print('media(x):', me)
print('mediana(x):', md)
print('moda(x):', mo) # devuelve el valor de la moda y el nº de veces que se repite
```

### 1.2 Medidas de posición: mínimo, máximo, percentiles (cuantiles) y cuartiles

```
# mínimo y máximo
x_min = min(x)
x_max = max(x)

# percentiles y cuartiles (se puede obtener lo mismo usando percentile y quantile)
p = 75 # percentil entre 0 y 100
x_prctle = np.percentile(x, p)
#prctle2 = np.percentile(x,p, interpolation='lower')

q=0.25  # cuantil entre 0 y 1 (los cuartiles son 0.25, 0.5 y 0.75)
x_qtle = np.quantile(x, q)

print('x:', x)
print('mínimo de x:', x_min, ', máximo de x:', x_max)
print(f'percentil(x, {p}): { x_prctle}')
print(f'cuantil(x, {q}): {x_qtle}')

# representación visual de los valores x y las medidas de posición calculadas
plt.stem(x, np.ones(np.size(x)))
plt.plot(x_prctle, 0, 'ks', label='percentil')
plt.plot(x_qtle, 0, 'kd', label='cuantil')
plt.legend()
plt.title('Datos y medidas de posición')
plt.show()
```

### 1.3 Medidas de dispersión: rango, rango intercuartílico, varianza, desviación típica y coeficiente de variación

```
rg = max(x)-min(x)  # rango
iqr = stats.iqr(x) # eango intercuartílico

va = np.var(x) # varianza
st = np.std(x) # desviación típica

cvar  = stats.variation(x) # coeficiente de variación

print('x:', x)
print('Rango(x):', rg)
print('Rango intercuartílico IQR(x):', iqr)
print('Varianza(x):', va)
print('Desviación típica(x):', st)
print('Coeficiente de variación(x):', cvar)
```

### 1.4 Medidas de forma: coeficientes de apuntamiento y de simetría

```
# coef. apuntamiento (incluye corrección con el ajuste de -3)
kur = stats.kurtosis(x, fisher=True)
# coef. de simetría
ske = stats.skew(x)

print('Coef. de apuntamiento(x):', kur)
print('Coef. de simetría(x):', ske)
```

## 3. Medidas varias variables

Para trabajar con varias variables simultáneamente es necesario agruparlas en una "matriz". 
Existen varias alternativas dependiendo de tipo de dato que se tenga, pero quizás la más versatil sea crear un dataframe df de pandas en el que cada variable pasa a ser una columna y trabajar directamente con df. La creación de dicho dataframe puede ser curiosa, aquí usamos una bastante cómoda que vale tanto para variables tipo lista como tipo array. En algunas ocasiones ya puedes partir de un dataframe creado que tenga las variables listas para trabajar, en ese caso sobra la primera línea de código en la que se crea df a partir de las variables.

### 3.1 Medias, medianas, etc

```
# creamos dataframe para facilitar los cálculos
df =  pd.DataFrame(list(zip(x, y, z)), columns=['x', 'y', 'z'])
#print(df)

# métricas básicas
print(df.describe())

# alguna métrica concreta para todas las variables en df
medianas = df.median()
print('Mediana:')
print(medianas)

cvs = df.std() / df.mean()
print('Coeficiente de variación:')
print(cvs)

skew = df.skew()
print('Coeficiente de asimetría:')
print(skew)

kurtosis = df.kurtosis()
print('Coeficiente de apuntamiento:')
print(kurtosis)
```

### 3.2 Matrices de covarianzas y correlaciones para dos variables x e y

```
# matriz de covarianzas
mat_cov = np.cov(x,y)
print('Matriz covarianzas:')
print(np.round(mat_cov, 3)) # Redondeando a solo tres decimales para verlo mejor

# matriz de correlaciones
mat_corr = np.corrcoef(x,y)                
print('Matriz correlaciones:')
print(np.round(mat_corr, 3)) # Redondeando a solo tres decimales para verlo mejor

# Muestra la matriz de correlaciones visualmente
fig = plt.figure(1,figsize=(5, 5))
sns.heatmap(mat_corr, annot=True) # si no se ven los textos con demasiadas variables, poner annot=False
plt.title('Matriz de correlaciones')
plt.show()
```

### 3.3 Matrices de covarianzas y correlaciones (muchas variables)

```
# creamos dataframe para facilitar los cálculos y gráficos 
df =  pd.DataFrame(list(zip(x, y, z)), columns=['x', 'y', 'z'])
#print(df)

# matriz de covarianzas creada directamente con el dataframe (df)
mat_cov = df.cov()                
print('Matriz covarianzas:')
print(np.round(mat_cov, 2)) # Redondeando a solo dos decimales para verlo mejor

# matriz de correlaciones creada directamente con el dataframe (df)
mat_corr = df.corr()                
print('Matriz correlaciones:')
print(np.round(mat_corr, 2)) # Redondeando a solo dos decimales para verlo mejor

# Muestra la matriz de correlaciones visualmente
fig = plt.figure(1,figsize=(5, 5))
sns.heatmap(mat_corr, annot=True) # si no se ven los textos con demasiadas variables, poner annot=False
plt.title('Matriz de correlaciones')
plt.show()
```

## 4. Gráficos conjuntos para dos variables x e y

### 4.1 Nube de puntos

```
# scatterplot usando plt
fig = plt.figure(1, figsize=(6, 6))
plt.plot(x, y, '.', color='r', markersize = 20)
plt.title('Nube de puntos (x,y)')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()

# completando el scatter con los histogramas marginales
fig = plt.figure(2)
sns.jointplot(x=x, y=y)
plt.show()

# completando el scatter con los boxplots marginales
fig = plt.figure(3)
g = sns.JointGrid(x=x, y=y)
g.plot_joint(sns.scatterplot)
g.plot_marginals(sns.boxplot)
plt.show()
```

### 4.2 Histograma 2D

```
# histograma 2D
nbins_x_y = (40, 40) # número de intervalos histograma en ejes x e y, ajustar para buena visualización

fig = plt.figure(1, figsize=(8, 8))
ax=sns.displot(x=x, y=y, bins=nbins_x_y, cbar=True)
plt.title('Histograma de (x,y)')
plt.xlabel('x')
plt.ylabel('y')
plt.show()

# completando el hostograma 2D con los boxplots marginales
fig = plt.figure(2)
g = sns.JointGrid(x=x, y=y)
g.plot_joint(sns.histplot)   
g.plot_marginals(sns.boxplot)
plt.show()
```
