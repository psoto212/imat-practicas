## Librerías disponibles

```
# Importar librerías
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Variable aleatoria continua

### 1.1 Crear una v.a. continua

```
# Una v.a. continua no se puede crear definiendo probabilidades
# Se crea indicando la distribución de probabilidad que queremos usar

# Hay que indicar los parámetros que definen la distribución, que depende
# de la distribución concreta que se quiera usar

# Ejemplo de una exponencial
lmbda = 2
escalado = 1 / lmbda
VA = stats.expon(loc = 0, scale = escalado)
```

### 1.2 Cálculos con pdf y cdf

```
# Densidad de probabilidad en un punto concreto
x=3.0
densidad = VA.pdf(x)
print(f'f({x}) =', densidad)

# Función de probabilidad acumulada
x=3
cdf_va = VA.cdf(x)
print(f'F({x}) = P(X<={x}) =', cdf_va)

# Soporte de la v.a.
soporte_va = VA.support()
print('Soporte(X) =', soporte_va)
```

### 1.3 Medidas estadísticas fundamentales de la v.a.

```
# Se obtienen directamente a partir de la v.a. creada

# un conjunto básico
print('Media:', VA.mean())
print('Mediana:', VA.median())
print('Varianza:', VA.var())
print('Desviación estándar:', VA.std())

# percentiles (usando la inversa de la cdf)
q = 0.8
prctle_x = VA.ppf(q)
print(F'Quantil {q}:',  prctle_x)
```

### 1.4 Esperanza matemática de una función de una v.a.

```
# esperanza de una función de una v.a. VA

# definición de la función g(X)= 7 + 2 * X (combinación lineal)
# se usa una función anónima para crearla en línea
mi_funcion = lambda x: 7 + 2 * x

# prueba de la función para verificar que es lo que queremos
a= 10
print(F'Prueba de mi_funcion({a}) =', mi_funcion(a))

# calcula el valor experado de g(X), siendo X la v.a.
val_esperado = VA.expect(mi_funcion)
print('Valor esperado de g(X) = E[g(X)] =', val_esperado)
```

### 1.5 Gráfica de la función de densidad (pdf)

```
# Varias representaciones de la pdf, para elegir la más interesante

# Puntos en los que se evalua la función para representarla
# logicamente no pueden ser infinitos, con un nº razonable es
# suficiente para obtener una representación buena.
# Usando la función cuantíl es muy sencillo obtener el rango
# correcto para representar la zona de interés (percentiles 1% y 99%)
xs = np.linspace(VA.ppf(0.01), VA.ppf(0.99), 1000)

# usando una línea
fig = plt.figure(1, figsize=(6, 3))
plt.plot( xs, VA.pdf(xs), color="orange")
plt.xlabel('x')
plt.ylabel('pdf')
plt.grid()
plt.show()

# usando area para acentuar el hecho de que se integra para sacar cdf
fig = plt.figure(1, figsize=(6, 3))
plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, VA.pdf(xs), color="orange")
plt.xlabel('x')
plt.ylabel('pdf(x)')
plt.grid()
plt.show()
```

### 1.6 Gráfica de la función de distribución (cdf)

```
# varias representaciones de la cdf, para elegir la más interesante

# Puntos en los que se evalua la función para representarla
# logicamente no pueden ser infinitos, con un nº razonable es
# suficiente para obtener una representación buena.
# Usando la función cuantíl es muy sencillo obtener el rango
# correcto para representar la zona de interés (percentiles 1% y 99%)
xs = np.linspace(VA.ppf(0.01), VA.ppf(0.99), 1000)

# representa
fig = plt.figure(1, figsize=(6, 3))
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, VA.cdf(xs), color='blue')

plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()
plt.show()
```

### 1.7 Gráficas juntas de las funciones de densidad (pdf) y de distribución (cdf)

```
# Puntos en los que se evalua la función para representarla
# logicamente no pueden ser infinitos, con un nº razonable es
# suficiente para obtener una representación buena.
# Usando la función cuantíl es muy sencillo obtener el rango
# correcto para representar la zona de interés (percentiles 1% y 99%)
xs = np.linspace(VA.ppf(0.01), VA.ppf(0.99), 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, VA.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, VA.cdf(xs), color='blue', label = 'cdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()

plt.show()
```

## 2. Distribución Uniforme U(a, b)

```
# crea la v.a. "VA" que sigue una Uniforme

# Está definida en stats en su forma estándar U(0,1), es decir,
# es una uniforme en el rango [0,1]

# Es necesario utilizar los parámetros de localización (loc) y escalado (scale)
# para crear una VA con una distribución uniforme en el rango [a, b]
a= -2
b= 1.5

inicio = a
incremento = b - a
VA = stats.uniform(loc=inicio, scale=incremento)
```

```
# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(VA.ppf(0.01)-1, VA.ppf(0.99)+1, 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, VA.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, VA.cdf(xs), color='blue', label = 'cdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```

## 3. Distribución Triangular T(a,c,b)

```
# crea la v.a. "VA" que sigue una triangular

# Está definida en stats en su forma "estándar"

# Es necesario utilizar los parámetros de localización (loc), escalado (scale)
# y forma (shape) para crear una VA con una distribución deseada

a = 32 # min
c = 35.5 # moda 
b = 36 # max


loc = a
escalado = b-a
forma = (c-a)/escalado

VA = stats.triang(loc=loc, scale=escalado, c = forma)
# with an up-sloping line from loc to (loc + c*scale)
# and then downsloping for (loc + c*scale) to (loc + scale).
```

```
# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(VA.ppf(0.01)-1, VA.ppf(0.99)+1, 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, VA.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, VA.cdf(xs), color='blue', label = 'cdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```

## 3. Distribución Exponencial E($\lambda$)

```
# crea la v.a. "VA" que sigue una Exponencial

# Está definida en stats en su forma estándar E(1), es decir,
# es una exponencial de lambda = 1

# Es necesario utilizar los parámetros de localización (loc) y escalado (scale)
# para crear una VA con una distribución del lambda deseado
lmbda= 5

escalado = 1 / lmbda
VA = stats.expon(loc = 0, scale = escalado)
```

```
# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(VA.ppf(0.01), VA.ppf(0.99), 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, VA.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, VA.cdf(xs), color='blue', label = 'cdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```

## 3. Distribución Normal N($\mu$, $\sigma$)

```
# crea la v.a. "VA" que sigue una Normal

# Está definida en stats en su forma estándar N(0,1)

# Es necesario utilizar los parámetros de localización (loc) y escalado (scale)
# para crear una VA con una distribución Normal diferente

mu = 10
sigma = 0.5

VA = stats.norm(loc = mu, scale = sigma)
```

```
# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(VA.ppf(0.001), VA.ppf(0.999), 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, VA.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, VA.cdf(xs), color='blue', label = 'cdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```

## 3. Distribución Lognormal LN($\mu$, $\sigma$)

```
# crea la v.a. "VA" que sigue una LogNormal

# Está definida en stats en su forma estándar N(0,1)

# Es necesario utilizar los parámetros de forma (s) y escalado (scale)
# para crear una VA con una distribución Lognormal diferente

mu = 0.1
sigma = 0.4

s = sigma
escalado = np.exp(mu)

VA = stats.lognorm(s = s, scale = escalado)
```

```
# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(VA.ppf(0.001), VA.ppf(0.999), 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, VA.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, VA.cdf(xs), color='blue', label = 'cdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```
