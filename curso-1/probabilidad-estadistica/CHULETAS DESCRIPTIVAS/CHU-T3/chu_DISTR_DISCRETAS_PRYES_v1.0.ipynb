## Librerías disponibles

```
# Importar librerías
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

## 1. Variable aleatoria discreta

### 1.1 Crear una v.a. discreta

```
# Crea una v.a. a partir de la definición de sus probabilidades puntuales

# valores posibles de la v.a.
xk = [1, 2, 3, 4, 5]

# probabilidad de cada xk
pk = [0.1, 0.2, 0.4, 0.1, 0.2]

# comprobacion sum(pk) = 1
print('Verificar que 1 = sum(pks)=',sum(pk))

# crea la v.a. X utilizando los valores posibles y su probabilidad (xk,pk)
X = stats.rv_discrete(values=(xk, pk))
```

### 1.2 Cálculos con pmf y cdf

```
# Probabilidad puntual calculada usando la función de cuantia
k=3.0
pmf_va = X.pmf(k)
print(f'P(X={k}) =', pmf_va)

# Función de probabilidad acumulada
k=3
cdf_va = X.cdf(k)
print(f'P(X<={k}) =', cdf_va)

# soporte de la v.a.
soporte_va = X.support()
print('Soporte(X) =', soporte_va)
```

### 1.3 Medidas estadísticas fundamentales de la v.a.

```
# Se obtienen directamente a partir de la v.a. creada

# un conjunto básico
print('Media(X) =', X.mean())
print('Mediana(X) =', X.median())
print('Varianza(X) =', X.var())
print('Desviación estándar (X) =', X.std())

# percentiles (usando la inversa de la cdf)
q = 0.8
prctle_x = X.ppf(q)
print(f'Quantil {q} de X =',  prctle_x)
```

### 1.4 Esperanza matemática de una función de una v.a.

```
# esperanza de una función de una v.a. X

# definición de la función g(X)= 3 * X**2 (3 por X al cuadrado)
# se usa una función anónima para crearla en línea
mi_funcion = lambda x: 3 * x**2

# prueba de la función para verificar que es lo que queremos
a= 10
print(f'Prueba de mi_funcion({a}) =', mi_funcion(a))

# calcula el valor experado de g(X), siendo X la v.a.
val_esperado = X.expect(mi_funcion)
print('Valor esperado de g(X) = E[g(X)] =', val_esperado)
```

### 1.5 Gráfica de la función de cuantía (pmf)

```
# varias representaciones de la pmf, para elegir la más interesante

# puntos en los que se evalua la función para representarla
xs = range(-2, 7)

# usando "alfileres"
fig = plt.figure(1, figsize=(6, 3))
plt.stem(xs, X.pmf(xs), basefmt='none')
plt.ylim(0, 1)
plt.xlabel('x')
plt.ylabel('pmf(x)')
plt.show()

# usando barras
fig = plt.figure(1, figsize=(6, 3))
plt.bar(xs, X.pmf(xs), color='orange')
plt.ylim(0, 1)
plt.xlabel('x')
plt.ylabel('P(X=x)')
plt.show()
```

### 1.6 Gráfica de la función de distribución (cdf)

```
# varias representaciones de la cdf, para elegir la más interesante

# puntos en los que se evalua la función para representarla
xs = range(-2, 7)

# representa mediante escalones
fig = plt.figure(1, figsize=(6, 3))
plt.xlim(xs[0], xs[-1])
plt.step(xs, X.cdf(xs), color='red', where='post')

plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()
plt.show()
```

### 1.7 Gráficas juntas de las funciones de cuantía (pmf) y de distribución (cdf)

```
# puntos en los que se evalua la función para representarla
xs = range(-2, 7)

# en formato tabla
df =  pd.DataFrame(list(zip(xs,  X.pmf(xs), X.cdf(xs))), columns=['xs', 'pmf(xs)', 'cdf(xs)'])
print(df)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, X.pmf(xs), basefmt='none')
plt.ylim(0, 1)
plt.ylabel('pmf(x)')

plt.subplot(2,1,2, sharex = ax1)
plt.xlim(xs[0], xs[-1])
plt.step(xs, X.cdf(xs), color='red', where='post')

plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()

plt.show()
```

## 2. Distribución Binomial B(n, p)

```
# crea la v.a. "mi_va" que sigue una binomial
n= 20
p=0.05
X = stats.binom(n=n,p=p)
```

```
# puntos en los que se evalua las funciones para representarlas
xs = range(-2, 22)

# en formato tabla
df =  pd.DataFrame(list(zip(xs,  X.pmf(xs), X.cdf(xs))), columns=['xs', 'pmf(xs)', 'cdf(xs)'])
print(df)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(8, 6))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, X.pmf(xs), basefmt='none')
plt.ylim(0, 1)
plt.ylabel('pmf(x)')
plt.grid()
plt.title(f'Binomial(n={n}, p={round(p,4)})')

plt.subplot(2,1,2, sharex = ax1)
plt.xlim(xs[0], xs[-1])
plt.step(xs, X.cdf(xs), color='red', where='post')
plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()
plt.show()
```

## 3. Distribución Poisson P($\lambda$)

```
# crea la v.a. "mi_va" que sigue una Poisson
lmda= 4

X = stats.poisson(lmda)
```

```
# puntos en los que se evalua las funciones para representarlas
xs = range(-2, 20)

# en formato tabla
df =  pd.DataFrame(list(zip(xs,  X.pmf(xs), X.cdf(xs))), columns=['xs', 'pmf(xs)', 'cdf(xs)'])
print(df)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(8, 6))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, X.pmf(xs), basefmt='none')
#plt.ylim(0, 1)
plt.ylabel('pmf(x)')
plt.grid()
plt.title(f'Poisson(lambda={lmda})')

plt.subplot(2,1,2, sharex = ax1)
plt.xlim(xs[0], xs[-1])
plt.step(xs, X.cdf(xs), color='red', where='post')
plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()
plt.show()
```
