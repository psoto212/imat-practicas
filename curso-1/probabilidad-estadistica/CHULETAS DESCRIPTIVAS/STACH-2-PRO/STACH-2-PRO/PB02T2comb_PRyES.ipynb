### Los algoritmos son el objeto de estudio de la algoritmia. En concreto, el análisis de un algoritmo trata de encontrar la complejidad computacional del mismo, con el objetivo de realizar estimaciones teóricas de los recursos que necesita (tiempo y/o almacenamiento). Esto permite, por ejemplo, comparar teóricamente las necesidades de varios algoritmos para elegir el que mejor se adapte a los recursos disponibles.

### Se dispone del siguiente implementación como una función python de un algoritmo que muestra una determinada secuencia de números, construida mediante tres bucles anidados:

```
# implementación del algoritmo
def foo(n_c, n_d, n_u, mostrar_secuencia):

    num_numeros_generados = 0
    
    for i in range(0,n_c):
        for j in range(0,n_d):
            for k in range(0,n_u):
                if mostrar_secuencia:
                    print(i,j,k)
                num_numeros_generados += 1
                
    return num_numeros_generados
```

```
# ejemplo de ejecución
num_numeros = foo(4, 2, 3, mostrar_secuencia = True)
print(f'Se han generado {num_numeros} números.')
```

### Se ha encontrado una implementación de un segundo algoritmo que muestra una determinada secuencia de números, construida también mediante tres bucles anidados, pero menos obvia:

```
# implementación del segundo algoritmo
def foo_2(n_c, mostrar_secuencia):

    num_numeros_generados = 0
    
    for i in range(0,n_c+1):
        for j in range(0,i+1):
            for k in range(0,j+1):
                if mostrar_secuencia:
                    print(i,j,k)
                num_numeros_generados += 1
                
    return num_numeros_generados
```

```
# ejemplos de ejecución
print ('Ejecutando foo_2(2):')
num_numeros = foo_2(2, mostrar_secuencia = True)
print(f'Se han generado {num_numeros} números.')

print ('Ejecutando foo_2(3):')
num_numeros = foo_2(3, mostrar_secuencia = True)
print(f'Se han generado {num_numeros} números.')
```
