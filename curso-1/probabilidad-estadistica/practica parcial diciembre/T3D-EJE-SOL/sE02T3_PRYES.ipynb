```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import pandas as pd
```

### a) Crea una v.a. X que represente el nº de caras que se pueden obtener cuando se lanza 3 veces consecutivas una moneda normal (sin trucar). No utilices la distribución binomial para resolverlo.

```
# respuesta

print('Necesitamos determinar las probabilidades puntuales de los posibles valores de la v.a. X: 0, 1, 2 y 3. Es independiente lo '
      'que pase en cada tirada, las intersecciones son productos.')
print('  P(X=0) = P(X y X y X) = P(X) * P(X) * P(X) = (1/2)**3 = 1/8') 
print('  P(X=3) = P(C y C y C) = P(C) * P(C) * P(C) = (1/2)**3 = 1/8') 
print('  P(X=1) = P((CyXyX) o (XyCyX)  o (XyXyC)) = P(CyXyX) + P(XyCyX) + P(XyXyC) = 3 * ((1/2)**3 )  = 3/8') 
print('  P(X=2) = P((CyCyX) o (CyXyC)  o (XyCyC)) = P(CyCyX) + P(CyXyC) + P(XyCyC) = 3 * ((1/2)**3 )  = 3/8') 

# valores posibles de la v.a.
xk = [0, 1, 2, 3]

# probabilidad de cada xk
pk = [1/8, 3/8, 3/8, 1/8]

# comprobacion sum(pk) = 1
print('Verificar que 1 = sum(pks)=',sum(pk))

# crea la v.a. X utilizando los valores posibles y su probabilidad (xk,pk)
X = stats.rv_discrete(values=(xk, pk))
```

### b) Representa las funciones de masa de probabilidad y la función de distribución de probabilidad de la v.a. X

```
# respuesta

# puntos en los que se evalua la función para representarla
xs = range(-1, 5)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, X.pmf(xs), basefmt='none')
plt.ylabel('pmf(x)')

plt.subplot(2,1,2, sharex = ax1)
plt.xlim(xs[0], xs[-1])
plt.step(xs, X.cdf(xs), color='red', where='post')

plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()

plt.show()
```

### c) Utiliza la v.a. X para determinar la probabilidad de que siempre salga cruz

```
# respuesta

print('Se pide calcular P(X=3), directo de calcular a partir de la función de masa de probabilidad (pmf):')

prob = X.pmf(3)
print(f'P(X=3) = cdf(3) = {prob}')
```

### d) Calcula la probabilidad de que X sea mayor o igual que 1

```
# respuesta

print('Se pide calcular P(X>=1):')

prob = 1 - X.cdf(0)
print(f'P(X>=1) = 1 - P(X<1) = 1 - P(X<=0) = 1 - cdf(0) = {prob}')
print(f'Lógicamente, también se puede calcular P(X>=1) = 1 - P(X==0) = 1 - pmf(0) = {1 - X.pmf(0)}')
```
