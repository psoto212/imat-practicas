## Distribuciones de probabilidad discretas

stats.poisson

```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

### a) Sea X una v.a. que represente el nº de píxeles defectuosos en una pantalla de móvil de 5 pulgadas de un determinado fabricante SUMSAG, cuando se sabe que su proceso de fabricación tiene una tasa constante de produccir píxeles defectuosos de 2 píxeles por pulgada. Representa la distribución de dicha variable.

```
# Representa la distribución

print('X se distribuye como una Poisson de media 2 ppp x 5 p = 10 píxeles en una pantalla')

lmda= 10
X = stats.poisson(lmda)

# representa la distribución

# puntos en los que se evalua las funciones para representarlas
xs = range(0, 30)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(8, 6))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, X.pmf(xs), basefmt='none')
#plt.ylim(0, 1)
plt.ylabel('pmf(x)')
plt.grid()
plt.title(f'Poisson(lambda={lmda})')

plt.subplot(2,1,2, sharex = ax1)
plt.xlim(xs[0], xs[-1])
plt.step(xs, X.cdf(xs), color='red', where='post')
plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()
plt.show()
```

### b) Calcula el valor esperado del nº de píxeles defectuosos en un móvil de 5 pulgadas. Utiliza para ello la esperanza matemática de la v.a. ¿Es más o menos lo que esperabas?

```
# Respuesta

X_esp = X.mean()
print(f'El valor esperado E[X] coincide con el parámetro lambda (10 píxeles/pantalla), la media de X: {X_esp}')

# cálculo con expect

# definición de la función (es la identidad)
mi_funcion = lambda x: x

# prueba de la función para verificar que es lo que queremos
a= 2
print(f'Prueba de mi_funcion({a}) =', mi_funcion(a))

# calcula el valor experado de g(X), siendo X la v.a.
val_esperado = X.expect(mi_funcion)
print('Valor esperado usando expect: de X = E[X] =', val_esperado)

print('Lógicamente, el cálculo con expect es menos preciso al realizarse sumando numéricamente.')
```

### c) Calcula la probabilidad de que el nº de píxeles defectuosos en una pantalla de 5 pulgadas supere los 2 píxeles.

```
# Respuesta

print('Se pide calcular P(X>2), es decir 1-P(X<=2)')

prob = 1 - X.cdf(2)
print(f'Usando la cdf se obtiene fácilmente: P(X>2) = 1-P(X<=2) = 1 - cdf(2) = {prob}')
print('Lógico que sea una probabilidad alta ya que el valor esperado es 10.')
```

### d) Calcula la probabilidad de que el nº de píxeles defectuosos en una pantalla de 5 pulgadas sea exactamente cero.

```
# Respuesta 

print('Se pide calcular P(X=0).')

prob = X.pmf(0)
print(f'Usando la pmf se obtiene directamente: P(X=0) = pmf(0) = {prob}')
```

### e) Sabiendo que se dispone de un móvil con una pantalla de 5 pulgadas fabricada por SUMSAG que tiene un millón de píxeles, indica la probabilidad de que un píxel de dicha pantalla esté defectuoso.

```
# Respuesta

print('Hay un millón de píxeles en la pantalla, cada uno puede estar defectuoso o no.')
print('La tasa de defectuosos es de 10 píxeles por pantalla, es decir, 10 = n * p, siendo '
      'n el nº de píxeles n = 1e6, y p la probabilidad de que un píxel esté defectuoso.' )

p = 10 / 1e6
print(f'Por tanto, la probabilidad de que un píxel esté defectuoso es: p = 10 / n = {p}.')
```
