## Distribuciones de probabilidad discretas

stats.binom

```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import pandas as pd
```

### a) Crea una v.a. X que represente el nº de respuestas correctas a una pregunta de test realizada a un conjunto de 100 alumnos, sabiendo que la pregunta ha sido respondida al azar por todos los alumnos, con una probabilidad de 1/3 de marcar la correcta. Representa la distribución de dicha variable.

```
# Respuesta

# param de la binomial
p = 1/3 
n = 100

# v.a. nº respuestas correctas en 100 alum
X = stats.binom(n=n,p=p)
```

```
# Representa la distribución

# puntos en los que se evalua las funciones para representarlas
xs = range(0, 100)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(8, 6))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, X.pmf(xs), basefmt='none')
plt.ylabel('pmf(x)')
plt.grid()
plt.title(f'Binomial(n={n}, p={round(p,4)})')

plt.subplot(2,1,2, sharex = ax1)
plt.xlim(xs[0], xs[-1])
plt.step(xs, X.cdf(xs), color='red', where='post')
plt.xlabel('x')
plt.ylabel('cdf(x)')
plt.grid()
plt.show()
```

### b) Calcula el valor esperado del nº de respuestas correctas. ¿Es más o menos lo que esperabas?

```
# Respuesta

X_esp = X.mean()

print(f'El valor esperado de X es {X_esp}, que coincide lógicamente con la expresión analítica que permite calcularlo: ')
print(f'E[X] = n * p = {n} *{p} = {n*p}')

print(f'Es lógico ya que se hacen 100 intentos (hay 100 alumnos).')
```

### c) Calcula la probabilidad de que el nº de alumnos que respondan correctamente supere los 60.

```
# Respuesta

print('Se pide calcular P(X>60), es decir 1-P(X<=60)')

prob = 1 - X.cdf(60)
print(f'Usando la cdf se obtiene fácilmente:  P(X>60) = 1-P(X<=60) = 1 - {X.cdf(60)} = {prob}')
```

### d) Calcula la probabilidad de que el nº de respuestas correctas sea superior o igual al 25% y menor que el 75% del total de alumnos

```
# Respuesta 

print('Se pide calcular P(X>=25 & X<75), se obtiene usando la cdf y la pmf así: F(75) - F(25) - p(75) + p(25), '
      'o usando solo la cdf así  F(74) - F(24)')

prob1 = X.cdf(75) - X.cdf(25) - X.pmf(75) + X.pmf(25)
prob2 = X.cdf(74) - X.cdf(24)
print(f'Por tanto, P(X>=25 & X<75) = P(X<=75) - P(X<=25) - P(X==75) + P(X==25) = {prob1}')
print(f'Por tanto, P(X>=25 & X<75) = P(X<=74) - P(X<=24) = {prob2}')
```

### e) Calcula la probabilidad de que el nº de respuestas incorrectas sea igual o superior a las correctas

```
# Respuesta

print('X es v.a. que representa el nº de respuestas correctas, la v.a. que representa el nº de incorrectas será Y = 100 - X, ya que hay 100 alumnos.')
print('Se pide entonces P(X<=Y) = P(X <= 100-X) = P(X<=50). Se calcula fácilmente con la cdf:')

prob = X.cdf(50)
print(f'P(X<=Y) = P(X<=50) = cdf(50) = {prob}')
```
