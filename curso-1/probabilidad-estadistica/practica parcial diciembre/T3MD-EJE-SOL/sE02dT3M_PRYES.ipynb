```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### Se tiene un vector aleatorio [X, Y] que sigue la distribución bivariante indicada mediante su función de masa de probabilidad conjunta pmfXY:

```
# v.a. bidimensional de partida

# valores posibles de la v.a. X e Y
xk = [1, 2, 3]
yk = [10, 20, 30, 40]

# probabilidad de cada pareja (xk, yk), xk: columnas, yk: filas
pmfXY = np.matrix([
    [2/12, 1/12, 2/12],
    [2/12, 0/12, 0/12],
    [0/12, 3/12, 0/12],
    [1/12, 0/12, 1/12]
    ])
```

### a) Representa la función de masa de probabilidad conjunta pmfXY en una gráfica 3D.

```
# Representación de la pmf conjunta

# muestra SUPERFICIE 3D de la pmf
X, Y = np.meshgrid(xk,yk)

# evalua la pdf en cada punto (x_i,y_j)
Z = np.zeros(X.shape)
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
         Z[i,j] = pmfXY[i,j]

fig = plt.figure(2, figsize=(8, 8))
ax = plt.axes(projection='3d')
ax.set_xticks(xk)
ax.set_yticks(yk)
ax.invert_xaxis()
ax.stem(X.flatten(), Y.flatten(), Z.flatten(), basefmt='none')
plt.xlabel('x')
plt.ylabel('y')
plt.title('pmf conjunta')

# cambia orientación para que se vea bien
ax.azim = -50
ax.dist = 10
ax.elev = 20
plt.show()
```

### b) Calcula la función de distribución acumulada conjunta de (X,Y) y represéntala gráficamente.

```
# Cálculo

# función que crea la cdfXY (es una matriz con los valores acumulados)
def crea_matriz_cdfXY(pmfXY): # crea una matriz a partir de la info de la V.A.
    # sumas acumuladas horizontales y luego las acumula verticalmente
    jcdf = np.cumsum(np.cumsum(pmfXY,axis=1),axis=0)
    return jcdf

# crea la función de probabilidad conjunta como una matriz
cdfXY = crea_matriz_cdfXY(pmfXY)

# en un dataframe para visualizar fácilmente
print('cdf conjunta cdfXY:')
dfcdfXY = pd.DataFrame(cdfXY, index = yk, columns = xk)
print(dfcdfXY)
```

```
# Gráficamente

# obtiene los puntos de cambio primero
X, Y = np.meshgrid(xk,yk)
# evalua la cdf en cada punto (x_i,y_j)
Z = np.zeros(X.shape)
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
         Z[i,j] = cdfXY[i,j]

# ahora los puntos para la superficie
xk1 = np.append(xk, xk[-1] + (xk[-1]-xk[-2]) )
yk1 = np.append(yk, yk[-1] + (yk[-1]-yk[-2]) )

xxs = np.linspace(xk[0]-0.5*(xk[1]-xk[0]), xk1[-1]-0.01, 500)
yys = np.linspace(yk[0]-0.5*(yk[1]-yk[0]), yk1[-1]-0.01, 500)
XX, YY = np.meshgrid(xxs,yys)
ZZ = np.zeros(XX.shape)

for i in range(len(xk1)):
    for j in range(len(yk1)):
        ind_xxs = (XX < xk1[i]) & (XX >= xk1[i-1])
        ind_yys = (YY < yk1[j]) & (YY >= yk1[j-1])
        ind_zzs = ind_xxs & ind_yys
        ZZ[ind_zzs] = cdfXY[j-1,i-1]
        
# representa
fig = plt.figure(1, figsize=(16, 8))
ax = fig.add_subplot(121,projection='3d')   
ax.plot_surface(XX, YY, ZZ, alpha=0.95, rstride=2, cstride=2, cmap='viridis', edgecolor='none')
ax.stem(X.flatten(), Y.flatten(), Z.flatten(), basefmt='none')

ax.set_xticks(xk)
ax.set_yticks(yk)
ax.invert_xaxis()
plt.xlabel("X")
plt.ylabel("Y")
plt.title('cdf conjunta')
ax.azim = -50
ax.dist = 10
ax.elev = 40

# representa NIVELES en el otro eje
ax = fig.add_subplot(122)   
ax.set_xticks(xk)
ax.set_yticks(yk)
ax.invert_xaxis()
plt.contourf(XX, YY, ZZ, cmap='viridis', alpha = 1) # alpha para controlar la transparencia y poder pintar varias

plt.xlabel("X")
plt.ylabel("Y")
plt.title('cdf conjunta (niveles)')
plt.grid()
plt.colorbar()
plt.show()
```

### c) Calcula la probabilidad de que X sea igual a 2.

```
# Respuesta

print('Para calcular P(X=2) se necesita la pmf de la marginal X.')

pmfX = np.sum(pmfXY, 0).getA1()
VA_X = stats.rv_discrete(values=(xk, pmfX))
print('Verifica que 1 = suma de Marginal pmfX = ', pmfX.sum())
print('Marginal pmfX:', pmfX)

# usamos la v.a. creada
prob = VA_X.pmf(2)
print(f'Usando la v.a. X se obtiene: P(X=2) = X.pmf(2) = {prob}')
```

### d) Calcula la probabilidad de que Y sea menor o igual que 24.

```
# Respuesta

print('Para calcular P(Y<=24) se necesita la cdf de la marginal Y.')

# Marginal Y a partir de la pmf de (X,Y) discreta bivariante
pmfY = np.sum(pmfXY, 1).getA1()
VA_Y = stats.rv_discrete(values=(yk, pmfY))
print('Verifica que 1 = suma de Marginal pmfY = ', pmfY.sum())
print('Marginal pmfY:', pmfY)

# usamos la v.a. creada
prob = VA_Y.cdf(24)
print(f'Usando la v.a. Y se obtiene: P(Y<=24) = Y.cdf(24) = {prob}')
```

### e) Calcula la probabilidad de que Y sea menor o igual que 24 sabiendo que X = 2.

```
# Respuesta

print('Para calcular P(Y<=24|X=2) se necesita la cdf de la v.a Y|X=2.')

x = 2
colum_valor_x = xk.index(x)
print(f'Posición del valor x={x} en xk: {colum_valor_x}')

pmfYcX = (pmfXY[:, colum_valor_x] / pmfX[colum_valor_x]).getA1()
print(f'Verifica que 1 = suma de pmfY|X={x} = ', pmfYcX.sum())
print(f'pmfY|X={x} = {np.transpose(pmfXY[:, colum_valor_x])}/{pmfX[colum_valor_x]} = {pmfYcX}')
VA_YcondX = stats.rv_discrete(values=(yk, pmfYcX))

# usamos la v.a. creada
prob = VA_YcondX.cdf(24)
print(f'Usando la v.a. Y|X=2 se obtiene: P(Y<=24|X=2) = Y|X=2.cdf(24) = {prob}')
```

### f) Calcula, utilizando la cdf conjunta, la probabilidad de que, al mismo tiempo, X sea mayor que 2 y que Y sea mayor que 30. Comprueba el resultado usando la pmf conjunta.

```
# Respuesta

infinito_positivo = np.PINF

print('Se pide P(X>2,Y>30), que se obtiene a partir de la cdf conjunta de la v.a. (X,Y).')
print('Se cumple que P(X>2,Y>30) = 1 - P(X<=2,Y<=INF) - P(X<=INF,Y<=30) +  P(X<=2,Y<=30)'
      'P(X>2,Y>30) = 1 - XY.cdf(2,40) - XY.cdf(3,30) + XY.cdf(2,30)')

# XY.cdf(2,INF)
x_p = 2
y_p = infinito_positivo
ind_xk = np.where(np.array(xk) <= x_p)
ind_yk = np.where(np.array(yk) <= y_p)

if (len(ind_xk[0])==0) | (len(ind_yk[0])==0):
    prob = 0.0
else:
    x_cdf = xk[ind_xk[0][-1]]
    y_cdf = yk[ind_yk[0][-1]]
    prob_2_INF = cdfXY[yk.index(y_cdf), xk.index(x_cdf)]
       
print(f'cdfXY(x={x_p},y={y_p}) = P(X<={x_p} and Y<={y_p} ) = {prob_2_INF}')

# XY.cdf(INF,30)
x_p = infinito_positivo
y_p = 30
ind_xk = np.where(np.array(xk) <= x_p)
ind_yk = np.where(np.array(yk) <= y_p)

if (len(ind_xk[0])==0) | (len(ind_yk[0])==0):
    prob = 0.0
else:
    x_cdf = xk[ind_xk[0][-1]]
    y_cdf = yk[ind_yk[0][-1]]
    prob_INF_30 = cdfXY[yk.index(y_cdf), xk.index(x_cdf)]
       
print(f'cdfXY(x={x_p},y={y_p}) = P(X<={x_p} and Y<={y_p} ) = {prob_INF_30}')


# XY.cdf(2,30)
x_p = 2
y_p = 30
ind_xk = np.where(np.array(xk) <= x_p)
ind_yk = np.where(np.array(yk) <= y_p)

if (len(ind_xk[0])==0) | (len(ind_yk[0])==0):
    prob = 0.0
else:
    x_cdf = xk[ind_xk[0][-1]]
    y_cdf = yk[ind_yk[0][-1]]
    prob_2_30 = cdfXY[yk.index(y_cdf), xk.index(x_cdf)]
       
print(f'cdfXY(x={x_p},y={y_p}) = P(X<={x_p} and Y<={y_p} ) = {prob_2_30}')

# prob pedida
prob = 1 - prob_2_INF - prob_INF_30 + prob_2_30
print(f'Por tanto, P(X>2,Y>30) = 1 - P(X<=2,Y<=INF) - P(X<=INF,Y<=30) +  P(X<=2,Y<=30) = 1 - {prob_2_INF} - {prob_INF_30} + {prob_2_30} = {prob}')

# Usando la pmf, eligiendo a mano lo que hay que sumar

# en un dataframe para visualizar fácilmente
print('Observando la pmf conjunta pmfXY:')
dfpmfXY = pd.DataFrame(pmfXY, index = yk, columns = xk)
print(dfpmfXY)

prob_2 = dfpmfXY.iloc[-1,-1]
print(f'Sumando pmfs: jpmf(3, 40) = {prob_2}, mismo resultado, claro.')
```
