```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import pandas as pd
```

### Se tiene un vector aleatorio [X, Y] que sigue la distribución bivariante indicada mediante su función de masa de probabilidad conjunta pmfXY:

```
# v.a. bidimensional de partida

# valores posibles de la v.a. X e Y
xk = [10, 20, 30]
yk = [1, 2, 3, 4, 5]

# probabilidad de cada pareja (xk, yk), xk: columnas, yk: filas
pmfXY = np.matrix([
    [1/15, 1/15, 1/15],
    [1/15, 1/15, 1/15],
    [1/15, 1/15, 1/15],
    [1/15, 1/15, 1/15],
    [1/15, 1/15, 1/15]
    ])
```

### a) Representa la función de masa de probabilidad conjunta pmfXY en una gráfica 3D.

```
# Representación de la pmf conjunta

# muestra SUPERFICIE 3D de la pmf
X, Y = np.meshgrid(xk,yk)

# evalua la pdf en cada punto (x_i,y_j)
Z = np.zeros(X.shape)
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
         Z[i,j] = pmfXY[i,j]

fig = plt.figure(2, figsize=(8, 8))
ax = plt.axes(projection='3d')
ax.set_xticks(xk)
ax.set_yticks(yk)
ax.invert_xaxis()
ax.stem(X.flatten(), Y.flatten(), Z.flatten(), basefmt='none')
plt.xlabel('x')
plt.ylabel('y')
plt.title('pmf conjunta')

# cambia orientación para que se vea bien
ax.azim = -50
ax.dist = 10
ax.elev = 20
plt.show()
```

### b) Observando únicamente la gráfica anterior, indica cómo serán las distribuciones marginales de masa de probabilidad de X e Y. Verifica tu intuición calculando y representando ambas distribuciones marginales.

```
# Respuesta

print('Cada distribución marginal será una distribución uniforme discreta, es decir, '
      'todos los valores de cada variable son igual de probables.')

# Marginal X a partir de la pmf y cdf de (X,Y) discreta bivariante
pmfX = np.sum(pmfXY, 0).getA1()
print('Marginal pmfX:', pmfX)

# Marginal Y a partir de la pmf y cdf de (X,Y) discreta bivariante
pmfY = np.sum(pmfXY, 1).getA1()
print('Marginal pmfY:', pmfY)

# comprueba que son pmfs válidas
print('Verifica que 1 = suma de Marginal pmfX = ', pmfX.sum())
print('Verifica que 1 = suma de Marginal pmfY = ', pmfY.sum())


# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(10, 4))

plt.subplot(1,2,1)
plt.stem(xk, pmfX, basefmt='none')
plt.ylim(0, 1)
plt.xlabel('x')
plt.ylabel('P(X=x)')
plt.title('Marginal X');plt.grid()

plt.subplot(1,2,2)
plt.stem(yk,pmfY, basefmt='none')
plt.ylim(0, 1)
plt.xlabel('y')
plt.ylabel('P(Y=y)')
plt.title('Marginal Y');plt.grid()

plt.show()
```

### c) Calcula la función de distribución acumulada conjunta del vector aleatorio dado y muestra la matriz que la representa. Observando únicamente dicha matriz, indica cuál es la probabilidad de que, al mismo tiempo, X sea inferior a 22 e Y sea inferior a 3.2.

```
# Respuesta

# función que crea la cdfXY (es una matriz con los valores acumulados)
def crea_matriz_cdfXY(pmfXY): # crea una matriz a partir de la info de la V.A.
    # sumas acumuladas horizontales y luego las acumula verticalmente
    jcdf = np.cumsum(np.cumsum(pmfXY,axis=1),axis=0)
    return jcdf

# crea la función de probabilidad conjunta como una matriz
cdfXY = crea_matriz_cdfXY(pmfXY)

# en un dataframe para visualizar fácilmente
print('cdf conjunta cdfXY:')
dfcdfXY = pd.DataFrame(cdfXY, index = yk, columns = xk)
print(dfcdfXY)

# calcula la probabilidad solicitada
print('\nSe pide calcular P(X<22 y Y<3.2) que equivale a P(X<20 y Y<3). Observando la matriz '
      'que representa la cdf conjunta, es sencillo ver que P(X<22 y Y<3.2) = 0.4 ')
```

### d) Representa gráficamente la cdf conjunta de la v.a. (X,Y). Añade además los puntos (X=22, Y<3.2) y (X=20, Y=3). ¿Están estos puntos en planos diferentes de la gráfica?

```
# Respuesta

# ojo, puede tardar un poco ...

# obtiene los puntos de cambio primero
X, Y = np.meshgrid(xk,yk)
# evalua la cdf en cada punto (x_i,y_j)
Z = np.zeros(X.shape)
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
         Z[i,j] = cdfXY[i,j]

# ahora los puntos para la superficie
xk1 = np.append(xk, xk[-1] + (xk[-1]-xk[-2]) )
yk1 = np.append(yk, yk[-1] + (yk[-1]-yk[-2]) )

xxs = np.linspace(xk[0]-0.5*(xk[1]-xk[0]), xk1[-1]-0.01, 500)
yys = np.linspace(yk[0]-0.5*(yk[1]-yk[0]), yk1[-1]-0.01, 500)
XX, YY = np.meshgrid(xxs,yys)
ZZ = np.zeros(XX.shape)

for i in range(len(xk1)):
    for j in range(len(yk1)):
        ind_xxs = (XX < xk1[i]) & (XX >= xk1[i-1])
        ind_yys = (YY < yk1[j]) & (YY >= yk1[j-1])
        ind_zzs = ind_xxs & ind_yys
        ZZ[ind_zzs] = cdfXY[j-1,i-1]
        
# representa
fig = plt.figure(1, figsize=(16, 8))
ax = fig.add_subplot(121,projection='3d')   
ax.plot_surface(XX, YY, ZZ, alpha=0.95, rstride=2, cstride=2, cmap='viridis', edgecolor='none')
ax.stem(X.flatten(), Y.flatten(), Z.flatten(), basefmt='none')

# añade los puntos pedidos
cdf_puntos = 0.4
plt.plot(22, 3.2, cdf_puntos, 'sr', label = 'cdf(X=22, Y=3.2)=0.4')
plt.plot(20, 3, cdf_puntos, 'sk', label = 'cdf(X=20, Y=3)=0.4')
plt.legend()

ax.set_xticks(xk)
ax.set_yticks(yk)
ax.invert_xaxis()
plt.xlabel("X")
plt.ylabel("Y")
plt.title('cdf conjunta')
ax.azim = -50
ax.dist = 10
ax.elev = 40

# representa NIVELES en el otro eje
ax = fig.add_subplot(122)   
ax.set_xticks(xk)
ax.set_yticks(yk)
ax.invert_xaxis()
plt.contourf(XX, YY, ZZ, cmap='viridis', alpha = 1) # alpha para controlar la transparencia y poder pintar varias

# añade los puntos pedidos
plt.plot(22, 3.2, 'sr', label = 'cdf(X=22, Y=3.2)=0.4')
plt.plot(20, 3, 'sk', label = 'cdf(X=20, Y=3)=0.4')
plt.legend()
    
plt.xlabel("X")
plt.ylabel("Y")
plt.title('cdf conjunta (niveles)')
plt.grid()
plt.colorbar()
plt.show()

print('Lógicamente, estos dos puntos están en el mismo plano, tienen el mismo valor de cdf ya que no '
      'se acumula probabilidad entre el punto (20, 3) y el (22, 3.2). Esto es lo que se ha utilizado en el '
      'apartado anterior para calcular fácilmente P(X<22 y Y<3.2).')
```
