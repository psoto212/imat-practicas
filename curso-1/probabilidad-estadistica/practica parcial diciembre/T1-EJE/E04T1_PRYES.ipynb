### c) Calcula el coeficiente de asimetría y el de exceso de curtosis de los datos almacenados en las variables x1, x2, x3, x4 y x5. Estudia los resultados obtenidos, comparándolos con un histograma de estas variables. ¿Qué significa un coeficiente de exceso curtosis negativo?¿Y un coeficiente de asimetría negativo?

```
# datos
x1 = [1,2,2, 3,3,3,3,3,3, 4,4,5] # symmetrical and meso
x2 = [1,2,2, 3,3,3,3,3,3,3,3,3,3, 4,4,5] # symmetrical and lepto
x3 = [1,2,3,4,5,6,7,8,9] # symmetrical and flat (platyk)

x4 = [1,1,1,1,2,2,3,4,9] # asymmetrical (positive skew)
x5 = -1 * np.array(x4) # asymmetrical (negative skew)

print('x1: ', x1)
print('x2: ', x2)
print('x3: ', x3)
print('x4: ', x4)
print('x5: ', x5)
```

```
# respuesta

fig = plt.figure(1, figsize=(8, 20))  #mayor alto para que se vean bien las 5 gráficas

# x1
# calculamos ambos índices
kur = stats.kurtosis(x1, fisher = True) # <--- correction -3
ske = stats.skew(x1)

ax1= plt.subplot(5,1,1)
sns.histplot(data=x1)
sns.rugplot(data=x1, color='r')
plt.title(f'Histograma x1: Exceso curtosis= {kur},  Asimetría= {ske}')

# x2
# calculamos ambos índices
kur = stats.kurtosis(x2, fisher = True) # <--- correction -3
ske = stats.skew(x2)

ax1= plt.subplot(5,1,2)
sns.histplot(data=x2)
sns.rugplot(data=x2, color='r')
plt.title(f'Histograma x2: Exceso curtosis= {kur},  Asimetría= {ske}')

# x3
# calculamos ambos índices
kur = stats.kurtosis(x3, fisher = True) # <--- correction -3
ske = stats.skew(x3)

ax1= plt.subplot(5,1,3)
sns.histplot(data=x3)
sns.rugplot(data=x3, color='r')
plt.title(f'Histograma x3: Exceso curtosis= {kur},  Asimetría= {ske}')

# x4
# calculamos ambos índices
kur = stats.kurtosis(x4, fisher = True) # <--- correction -3
ske = stats.skew(x4)

ax1= plt.subplot(5,1,4)
sns.histplot(data=x4)
sns.rugplot(data=x4, color='r')
plt.title(f'Histograma x4: Exceso curtosis= {kur},  Asimetría= {ske}')

# x5
# calculamos ambos índices
kur = stats.kurtosis(x5, fisher = True) # <--- correction -3
ske = stats.skew(x5)

ax1= plt.subplot(5,1,5)
sns.histplot(data=x5)
sns.rugplot(data=x5, color='r')
plt.title(f'Histograma x5: Exceso curtosis= {kur},  Asimetría= {ske}')

plt.show()

print('Un exceso negativo de curtosis, como en el caso de x3, implica que la distribución es menos apuntada de lo normal. '
      'Una asimetría negativa implica que la cola de la distribución apunta a la izquierda, como en el caso de la variable x5.')
```
