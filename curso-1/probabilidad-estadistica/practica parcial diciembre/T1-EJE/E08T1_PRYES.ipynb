### a) Se dispone de los datos almacenados en las variables x e y. Calcula las matrices de covarianzas y de correlaciones lineales. ¿Qué variable tiene una varianza mayor? ¿Existe una relación lineal entre las variables? Confirma lo reportado por estas medidas con lo que puedas observar en un gráfico de nube de puntos.

```
# datos
x=[1,2,3,5,6,6,8,8,8,8,8,9,9,10,11,1]
y=[6,7,8,5,6,6,6,3,3,3,3,2,2,1,1,10]

print('x: ', str(x))
print('y: ', str(y))
```

```
# respuesta

# cálculo de la covarianza y de la correlación
mat_cov = np.cov(x,y)
mat_corr = np.corrcoef(x,y)
print('Matriz de covarianzas: \n', mat_cov)
print('Matriz de correlaciones: \n', mat_corr)

print('Según la matriz de covarianzas, la variable x tiene mayor varianza que la y.')
print(f'El coeficiente de correlación lineal {mat_corr[0][1]} indica que parece existir una relación lineal inversa relevante.')

# scatterplot
sns.relplot(x=x, y=y, color='green')
plt.title('Nube de puntos (x,y)')
plt.xlabel('x')
plt.ylabel('y')
plt.grid()
plt.show()

print('Se confirma que existe esa relación lineal inversa.')
```

### b) Se dispone de los datos almacenados en las variables x, y y z. Calcula las matrices de covarianzas y de correlaciones lineales. ¿Qué pareja de variables tiene mayor correlación lineal? Representa la matriz de gráficos de dispersión para tener más información sobre la posible relación entre las variables.

```
# datos
x=[1,2,3,5,6,6,8,8,8,8,8,9,9,10,11,1]
y=[6,7,8,5,6,6,6,3,3,3,3,2,2,1,1,10]
z=[6,6,5,7,8,8,4,6,9,9,9,12,11,10,4,5]

print('x: ', x)
print('y: ', y)
print('z: ', z)
```

```
# respuesta

# crea dataframe con las variables
df =  pd.DataFrame(list(zip(x, y, z)), columns=['x', 'y', 'z'])
#print(df)

# matriz de covarianzas creada directamente con el dataframe (df)
mat_cov = df.cov()                
print('Matriz covarianzas:')
print(np.round(mat_cov, 3)) # Redondeando a solo tres decimales para verlo mejor

# matriz de correlaciones creada directamente con el dataframe (df)
mat_corr = df.corr()                
print('Matriz correlaciones:')
print(np.round(mat_corr, 3)) # Redondeando a solo tres decimales para verlo mejor


print(f'Las variables más correlacionadas son (x,y) ya que tienen una correlación lineal de {mat_corr.iloc[0][1]}, '
      'indicando que existe una correlación lineal inversa importante entre ambas variables.')

# representa nubes de puntos
sns.pairplot(df, height=2)
plt.show()
```
