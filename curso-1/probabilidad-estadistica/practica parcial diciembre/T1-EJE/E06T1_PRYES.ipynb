```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

### a) Representa el gráfico de violín de los datos almacenados en la variable x. No fijes el ancho de banda, deja que se decida automáticamente. ¿Cómo es la asimetría de los datos?

```
# datos
x=[1,1,2,5,6,6,8,8,8,8,8,9,9,10,11]
```

```
# respuesta

fig = plt.figure(1, figsize=(8, 4)) # permite indicar el nº de la figura y las dimensiones (ancho y alto)
sns.violinplot(data=x, orient='horizontal')
plt.xlabel('x')
plt.show()

print('Se observa una clara asimetría a la izquierda.')
```

### b) Representa en la misma figura el histograma, el gráfico de cajas y el gráfico de violín de los datos almacenados en la variable x. Asegúrate de que los dos gráficos compartan el eje x para una correcta interpretación. ¿En qué gráfico es más sencillo ver el mínimo y máximo de los datos?¿Y la media?

```
# datos
x=[1,1,2,5,6,6,8,8,8,8,8,9,9,10,11]
print('x=',x) 
print('Median(x)=', np.median(x), '| Mean(x)=', np.mean(x))
```

```
# respuesta

# dibuja en un mismo gráfico el histograma, el boxplot y el gráfico de violín
fig = plt.figure(1, figsize=(8, 10)) 

# histograma
ax1= plt.subplot(3,1,1)
sns.histplot(data=x)
plt.title('Histograma')

# cajas
ax2=plt.subplot(3,1,2, sharex=ax1)
sns.boxplot(data=x, orient='horizontal')
plt.title('Diagrama de cajas')

# violin
ax3=plt.subplot(3,1,3, sharex=ax1)
sns.violinplot(data=x, orient='horizontal')
sns.rugplot(data=x, color='r')
plt.xlabel('x')
plt.title('Gráfico de violín')

plt.show()

print('El mínimo y máximo de los datos se observa muy bien en el diagrama de cajas, lógicamente también en la versión compacta del mismo que aparece en el diagrama violín.')
print('La media no se observa directamente en ninguno de los tres gráficos.')
```
