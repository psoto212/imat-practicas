```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

### a) Crea una v.a. X que siga una distribución Uniforme U[10, 20], representa su función de densidad y observa el valor máximo que alcanza. ¿Coincide con el valor teórico?

```
# Respuesta 

# crea la v.a. "VA" que sigue una Uniforme
a= 10
b= 20

inicio = a
incremento = b - a
X = stats.uniform(loc=inicio, scale=incremento)

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(X.ppf(0.01)-5, X.ppf(0.99)+5, 1000)

# pinta
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X.pdf(xs), color="orange", label = 'pdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
plt.show()

# máximo, calculado a partir de un punto cualquiera dentro de [a, b]

max_pdf = X.pdf(12)
max_teo = 1 /(b-a)
print(f'El valor máximo que alcanza es {max_pdf}, que coincide con el valor teórico 1/(b-a) = {max_teo}.')
```

### b) Crea una nueva v.a. Y que siga una distribución uniforme con parámetro a = 10 y parámetro b tal que el máximo de la función de densidad sea 10. Representa la pdf para confirmar que es correcto.

```
# Respuesta

a= 10
b= a + 1/10
print('Para que el área bajo la función de densidad de una uniforme sea 1 tiene que ocurrir que: ')
print('1 = base * altura = (b-a) * maximo = (b-10) * 10')
print('Es decir, la función de densidad vale 1/(b-a) en el intervalo [a, b], si a=10, para que '
      f'el maximo sea 10, b ha de ser {b}')

# representa para comprobar
inicio = a
incremento = b - a
Y = stats.uniform(loc=inicio, scale=incremento)

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(Y.ppf(0.01)-0.1, Y.ppf(0.99)+0.1, 1000)

fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, Y.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, Y.pdf(xs), color="orange", label = 'pdf(y)')

plt.xlabel('y')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```

### d) Prueba a crear una v.a. distribuida uniformemente entre 10 y 10+0.000001. ¿Qué altura esperas que alcance la pdf de la v.a.?

```
# Respuesta
a=10
b=10+0.000001
print('Esperamos un altura muy grande para que el área bajo la curva sea 1.')
print('El valor máximo será 1/(b-a) =', np.round(1/(b-a),2))

# creamos la variable y representamos la pdf para verificarlo
inicio = a
incremento = b - a
X = stats.uniform(loc=inicio, scale=incremento)

# puntos en los que se evalua la pdf para representarla
# hay que tener cuidado para tener puntos dentro del intervalo
xs = np.linspace(X.ppf(0.01)-0.0000001, X.ppf(0.99)+0.000001, 10000)

fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X.pdf(xs), color="orange", label = 'pdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('Efectivamente, el máximo de la pdf es 1 millón.')
```

### e) ¿Qué piensas que puede ocurrir si creas una v.a. distribuida según una uniforme con el parámetro b = a = 10? Crea la v.a. y evalúa la pdf en 10. ¿Qué resultado obtienes?

```
# Respuesta

print('Esperamos un altura muy grande en la función de densidad para que el área bajo la curva sea 1.')
a= 10
b= 10

inicio = a
incremento = b - a
VA = stats.uniform(loc=inicio, scale=incremento)

x=10
densidad = VA.pdf(x)
print(f'f({x}) =', densidad)

print('La pdf devuelve un nan, indicando que no está bien definida.')
```
