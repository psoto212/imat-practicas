```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

### a) Crea una v.a. X que siga una distribución Normal de media 100 y varianza 2, representa sus funciones de densidad y de distribución de probabilidad.

```
# Respuesta 

mu = 100
sigma = np.sqrt(2)

X = stats.norm(loc = mu, scale = sigma)

# representa la pdf y cdf
xs = np.linspace(X.ppf(0.001), X.ppf(0.999), 1000)
fig = plt.figure(1, figsize=(12, 4))

plt.fill_between( xs, X.pdf(xs), color="green", alpha=0.3)
plt.plot(xs, X.pdf(xs), color="green", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, X.cdf(xs), color='blue', label = 'cdf(x)')

plt.xlabel('x')
plt.legend()
plt.grid()
plt.show()
```

### b) Calcula el valor esperado y la varianza de la v.a. X. Confirma que el resultado es el que esperabas.

```
# Respuesta

print('Media:', X.mean())
print('Varianza:', X.var())
print('Es lo esperado, son los parámetros con los que se ha creado la v.a.')
```

### c) Calcula la probabilidad de que 95<X<=101

```
# Respuesta

prob = X.cdf(101)-X.cdf(95)
print('Se pide calcular P(X<16 y X>=10) = P(10<=X<16). Se calcula usando la función acumulada:')
print('P(95<X<=101) = P(95<=X<=101) = Prob(95<=X<101) = Prob(95<X<101) = F(101)-F(95) =', prob)
```

### d) Crea una v.a. Z que se distribuya según una normal estándar. A partir de Z, calcula la probabilidad de que X>100. Recuerda que X sigue una distribución Normal de media 100 y varianza 2. Confirma el resultado comparando con el cálculo directo usando X.

```
# Respuesta

# creamos la v.a. Z
mu_z = 0
sigma_z = np.sqrt(1)
Z = stats.norm(loc = mu_z, scale = sigma_z)

# prob pedida desde Z
print('Se pide calcular P(X>100) a partir de la v.a. Z.')
print('Z es la versión tipificada de X, es decir, Z = (X - mean(X)) /std(X) = (X - 100) / sqrt(2).')
print('Por tanto tipificando a ambos lados: P(X>100) = P((X - 100)/(sqrt(2)  > (100 - 100)/sqrt(2)) = P(Z > 0)' )
      
prob = 1- Z.cdf(0)
print(f'P(X>100) = P(Z > 0) = 1 - P(Z<=0) = {prob}')

# prob usando X directamente

prob = 1- X.cdf(100)
print(f'Con el cálculo directo, tenemos que P(X>100) = 1 - P(X<=100) = {prob}')

print('Hemos obtenido, como era de esperar, el mismo valor.')
```

### e) Calcula la probabilidad de que la v.a. Y=(X/2)-100 sea positiva.

```
# Respuesta

print('Se pide calcular P(Y>0), siendo la v.a. Y = (X/2)-100.')
print('La v.a. Y se distribuye como una normal ya que se ha obtenido mediante una transformación lineal de X, '
      'que se distribuye como una normal de media 100 y varianza 2. Creamos dicha variable Y para luego '
      'calcular cómodamente la probabilidad pedida.')

# cálculo parámetros nueva distribución
media_Y = X.mean()/2 - 100
varianza_Y = X.var()/4
print('La media de la nueva variable Y es E[Y] =', media_Y)
print('La varianza de la nueva variable Y es Var(Y) =', varianza_Y)

# crea la nueva v.a. Y
mu = media_Y
sigma = np.sqrt(varianza_Y)
Y = stats.norm(loc = mu, scale = sigma)

# cálculo de la probabilidad usando la cdf de la v.a. Y
prob = 1 - Y.cdf(0)
print(f'\nPor tanto, la probabilidad pedida es P(Y>0) = 1-P(Y<=0) = 1 - F_Y(0) = {prob}')

# representación gráfica de la distribución de Y
xs = np.linspace(Y.ppf(0.0001), Y.ppf(0.9999), 1000)

fig = plt.figure(1, figsize=(6, 3))
plt.plot(xs, Y.pdf(xs), color="orange")
plt.xlabel('y')
plt.ylabel('pdf')
plt.grid()
plt.show()
```

### f) Indica la probabilidad de que la v.a. X tome valores entre a+b y a+2b, siendo a la media de X y b la desviación típica de X.

```
# Respuesta

print('Se pide calcular P(a+b<X<a+2b), siendo a = mean(X) y b = std(X) y X la v.a. creada previamente.')

a= X.mean()
b= X.std()

lim_inf=a+b
lim_sup=a+2*b
prob = X.cdf(lim_sup)-X.cdf(lim_inf)
print(f'La probabilidad solicitada es P(a+b<X<a+2b) = P({lim_inf}<X<{lim_sup}) = {prob}')

# gráficamente
print('Esta probabilidad se corresponde con el siguiente área:')
xs = np.linspace(X.ppf(0.0000001), X.ppf(0.9999999), 1000)
xarea = np.linspace(lim_inf, lim_sup, 1000)
# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(12, 4))

plt.fill_between(xarea, X.pdf(xarea), color="red", alpha=0.3, label = f'P({lim_inf}<X<{lim_sup})')
plt.plot(xs, X.pdf(xs), color="orange", label = 'pdf(x)')

plt.xlabel('x')
plt.legend()
plt.grid()
plt.show()
```
