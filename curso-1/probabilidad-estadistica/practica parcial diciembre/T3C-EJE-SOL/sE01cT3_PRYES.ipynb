```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

### a) Crea una v.a. X que siga una distribución Uniforme U[10, 20], representa su función de densidad y determina visualmente el valor esperado de X.

```
# Respuesta 

# crea la v.a. "VA" que sigue una Uniforme
a= 10
b= 20

inicio = a
incremento = b - a
X = stats.uniform(loc=inicio, scale=incremento)

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(X.ppf(0.01)-5, X.ppf(0.99)+5, 1000)

# pinta
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X.pdf(xs), color="orange", label = 'pdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()

print('El valor medio de esta distribución es el valor central, que es 15.')
print('Vemos que se corresponde con la media calculada con la fórmula:')
print('Media:', X.mean())
```

### b) Calcula la probabilidad de que la v.a. X sea igual a 15. Confirma que el resultado es lo que debería salir según la teoría.

```
# Respuesta

print('La probabilidad en un punto concreto representa el área bajo la función de densidad evaluada en ese punto concreto y es cero.')
print('La probabilidad de que X=15 equivale a calcular la probabilidad de que X esté en el intervalo [15, 15+dx], siendo dx un diferencial de x. '
      'Esta probabilidad se obtiene a partir de la función de densidad: P(x<X<x+dx)= f(x)*dx, que para dx=1 vale P(15<X<16) = f(15). ')

# Densidad de probabilidad en un punto concreto
x=15
densidad = X.pdf(x)
print(f'f({x}) =', densidad)
print('Por tanto, el valor de la densidad en ese punto hay que multiplicarlo por dx, que si tiende a cero, la probabilidad también tiende a cero.')
```

### c) Calcula la probabilidad de que la v.a. X sea menor o igual a 15. Confirma que el resultado es lo que esperabas.

```
# Respuesta

# Función de probabilidad acumulada
x=15
cdf_va = X.cdf(x)
print(f'La P(X<={x}) se calcula fácilmente con la función acumulada: F({x}) =', cdf_va)

# Soporte de la v.a.
soporte_va = X.support()
print('Soporte(X) =', soporte_va)

print('Es lo que esperábamos porque la distribución es simétrica con lo que la media es igual a la mediana (valor 15). '
      'La P(X<=15) tiene que ser 0.5, por definición de la mediana.')
```

### d) Calcula la probabilidad de que la v.a. X sea menor que 15.

```
# Respuesta

x=15
cdf_va = X.cdf(x)
print(f'La probabilidad pedida es P(X<{x}) = P(X<={x}) = F({x}) =', cdf_va)
```

### e) Calcula la probabilidad de que la v.a. X sea mayor que 15

```
# Respuesta

x=15
cdf_va = X.cdf(x)
print(f'La probabilidad pedida es P(X>{x}) = 1 - P(X<{x}) = 1 - P(X<={x}) = 1 - F({x}) =', 1-cdf_va)
```

### f) Representa la función de distribución de probabilidad (cdf) de la v.a. X. Utilizando únicamente dicha gráfica, indica la probabilidad de que X sea menor o igual que 15.

```
# Respuesta

xs = np.linspace(X.ppf(0.001)-5, X.ppf(0.999)+5, 1000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, X.cdf(xs), color='blue', label = 'cdf(x)')

# añade punto x = 15, cdf = 0.5
plt.plot(15,0.5,'sr', label = '0.5 = cdf(x=15)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()

plt.show()

print('Se ve claramente que dicha probabilidad es 0.5')
```
