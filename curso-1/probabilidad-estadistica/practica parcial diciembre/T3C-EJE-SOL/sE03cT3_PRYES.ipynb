```
# import required packages
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

### a) Crea una v.a. X que siga una distribución Exponencial cuyo valor esperado sea 20, representa sus funciones de densidad y de distribución de probabilidad.

```
# Respuesta 

lmbda= 1/20
print('Si el valor esperado es 20, significa que el parámetro de la Exponencial lamnda vale 1/20 =', lmbda)

# crea la v.a.
escalado = 1 / lmbda
X = stats.expon(loc = 0, scale = escalado)

# Representa la pdf y la cdf
xs = np.linspace(X.ppf(0.01)-1, X.ppf(0.99), 1000)

# pinta sincronizadamente ambas funciones
fig = plt.figure(1, figsize=(12, 6))
ax1 = plt.subplot(2,1,1)
plt.plot(xs, X.cdf(xs), color='blue', label = 'cdf(x)')
plt.ylabel('cdf(x)')
plt.legend()
plt.grid()

plt.subplot(2,1,2, sharex = ax1)
plt.fill_between( xs, X.pdf(xs), color="orange", alpha=0.3)
plt.plot( xs, X.pdf(xs), color="orange", label = 'pdf(x)')
plt.xlabel('x')
plt.ylabel('pdf(x)')
plt.legend()
plt.grid()
plt.show()
```

### b) Calcula el valor esperado y la varianza de la v.a. X. Confirma que el resultado es el que esperabas.

```
# Respuesta

print('Calculamos ambas medidas a partir de la v.a. X:')
print('  Media:', X.mean())
print('  Varianza:', X.var())

print('Es lo esperado ya que la v.a. se ha creado para que su valor esperado sea 20 ')
print('y teóricamente se sabe que la varianza de una exponencial es el cuadrado de su media:', 20**2)
```

### c) Calcula la probabilidad de que la v.a. X sea, al mismo tiempo, menor que 16 y mayor o igual que 10.

```
# Respuesta

prob = X.cdf(16)-X.cdf(10)
print('Se pide calcular P(X<16 y X>=10) = P(10<=X<16). Se calcula usando la función acumulada:')
print('P(10<X<16) = P(10<=X<=16) = P(10<X<=16) = P(10<=X<16) = F(16)-F(10) = ', prob)
```

### d) Calcula la probabilidad de que la v.a. X sea menor que 15.

```
# Respuesta

prob = X.cdf(15)
print('Se pide calcular P(X<15), al ser X una v.a. continua:')
print('P(X<15) = P(X<=15) = F(15) =', prob)
```

### e) Calcula la probabilidad de que la v.a. X sea mayor que 15

```
# Respuesta

prob = 1 - X.cdf(15)
print('Se pide calcular P(X>15):')
print('P(X>15) = 1 - P(X<=15) = 1-F(15) = ', prob)
```

### f) Sea Y una v.a. obtenida a partir de la expresión Y = 2+3*X. Indica el valor medio y la desviación típica de Y. Confirma el resultado utilizando la función "expect".

```
# Respuesta

# cálculo exacto usando los operadores esperanza matemática y varianza
e_y = 2 + 3 * X.mean()
var_y = 9 * X.var()
std_y = 3 * X.std()

print('Cálculo EXACTO:')
print('El valor esperado de Y es E[Y] = 2 + 3*E[X] = ', e_y)
print('La varianza de Y es Var(Y) = 9 * Var(X) = ', var_y)
print('La desviación típica de Y es Std(Y) = 3 * Std(X) = ', std_y)

# usando expect
print('\nCálculo APROXIMADO usando la función expect:')

print('Para calcular E[Y] se integra con expect la función g(X) = 2 + 3 * X')
funcion_g = lambda x: 2 + 3 * x
# prueba de la función para verificar que es lo que queremos
a= 10
print(f' - Prueba de funcion_g({a}) =', funcion_g(a))

# calcula el valor experado de g(X), siendo X la v.a.
val_esperado_g = X.expect(funcion_g)
print(' - El valor esperado de g(X) = 2+3X vale: E[g(X)] =', val_esperado_g)

print('\nPara calcular Var(Y) se integra con expect la función h(X) = (g(X) - E[g(X)])^2 = (2 +3*x - val_esperado_g)**2')
funcion_h = lambda x: (2 +3*x - val_esperado_g)**2
# prueba de la función para verificar que es lo que queremos
a= 10
print(f' - Prueba de funcion_h({a}) =', funcion_h(a))

# calcula el valor experado de h(X) que será la var(X), siendo X la v.a.
val_esperado_h = X.expect(funcion_h)
print(f' - El valor esperado de h(X) vale: E[h(X)] =, {val_esperado_h}, es decir, Var(X) = {val_esperado_h}')
print(f' - Por tanto, la desviación típica de X vale std(X) = sqrt(Var(X)) = {np.sqrt(val_esperado_h)}')


print('\nLa Var(g(X)) también se puede calcular así: Var(g(X)) = E(g(X)^2) - E(g(X))^2. Para calcular E(g(X)^2) integramos la función '
      'g2(X) = g(X)^2 =(2 +3*x)**2')
funcion_g2 = lambda x: (2 +3*x)**2
# prueba de la función para verificar que es lo que queremos
a= 10
print(f' - Prueba de funcion_g2({a}) =', funcion_g2(a))

# calcula el valor experado de h(X) que será la var(X), siendo X la v.a.
val_esperado_g2 = X.expect(funcion_g2)
var_calculo_2 = val_esperado_g2 - val_esperado_g**2
print(f' - El valor esperado de g2(X) vale: E[g2(X)] = {val_esperado_g2}')
print(f' - Por tanto, Var(g(X)) = E(g(X)^2) - E(g(X))^2 = {val_esperado_g2} - {val_esperado_g**2} = {var_calculo_2}')
```
