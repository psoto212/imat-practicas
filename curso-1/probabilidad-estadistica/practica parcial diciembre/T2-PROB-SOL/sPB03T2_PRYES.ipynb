### Ana y Carlos son amigos y estudiantes de iMAT. Al llegar por la mañana muy temprano a la biblioteca de ICAI observan que solo hay una persona al fondo de la sala. Está sentada de espaldas a ellos y solo pueden apreciar que está leyendo el Financial Times (FT).

#### - Ana: ¿qué opinas, es un nuevo profesor o es un estudiante?
#### - Carlos: creo que es un profesor
#### - Ana: ¿seguro? ¿por qué piensas eso?
#### - Carlos: porque está leyendo el FT
#### - Ana: creo que tu opinión puede mejorarse basándonos en la teoría de probabilidad que estudiamos en la asignatura de Probabilidad y Estadística
####  - Carlos: Vamos a secretaría y verás por qué tengo razón 

### Ya en secretaría de dirección les informan de dos hechos:
#### - En una encuesta realizada recientemente en ICAI sobre las publicaciones de economía y empresa más leídas, se obtuvo que el 50% de los estudiantes y el 90% de los profesores leen el FT. Al escuchar esto Carlos mira a Ana y sonríe pensando "lo ves ...". 
#### - Este curso ICAI tiene 1500 estudiantes y 110 profesores. Ante esto Ana le pregunta a Carlos "¿sigues en tus trece?". Carlos se pone serio y dice "pues claro, es evidente que es un profesor, estaba leyendo el FT".

### a) En base a la información recopilada por Ana y Carlos, y asumiendo que en el edificio de ICAI solo hay estudiantes y profesores, ¿Está justificada la postura de Carlos? Razona la respuesta.

```
# respuesta

# Sucesos
print('Definimos los siguientes sucesos:')
print('  E: la persona es estudiante')
print('  P: la persona es profesor')
print('  FT: la persona lee Financial Times')

# datos de partida
Pr_E = 1500/(1500+110)  # probabilidad a priori
Pr_P = 110/(1500+110)   # probabilidad a priori
Pr_FT_cond_E = 0.5
Pr_FT_cond_P = 0.9

print(f'Se tienen las probabilidades a priori P(E)={Pr_E}, P(P) = {Pr_P}, y las condicionadas P(FT|E)={Pr_FT_cond_E} y P(FT|P) = {Pr_FT_cond_P}.')

print('\nQueremos mejorar las probabilidades a priori incluyendo la información adicional de que sabemos que la persona está leyendo el Financial Times. '
      'Podemos calcular las probabilidades condicionadas de ser estudiante y profesor sabiendo que la persona lee el FT y ver qué es más probable:')
print('    P(E|FT) = P(E y FT) / P(FT) =  P(FT|E) P(E) / P(FT)')
print('    P(P|FT) = P(P y FT) / P(FT) =  P(FT|P) P(P) / P(FT)')

print('\nComo E y P forman una partición disjunta del espacio muestral (la persona que está en el hall es E o P), podemos usar la Regla de Bayes para obtener '
      'P(E|FT): la probabilidad de que sea estudiante sabiendo que lee FT, y P(P|FT): la de ser profesor sabiendo que lee FT P(P|FT). Para ello primero hay que calcular '
      ' P(FT): la probabilidad de que la persona lea el FT. Usando el Teorema de la Probabilidad Total es sencillo:')
     
Pr_FT = Pr_E * Pr_FT_cond_E + Pr_P * Pr_FT_cond_P

print(f'P(FT) =  P(FT|E) P(E) + P(FT|P) P(P) = {Pr_E * Pr_FT_cond_E} + {Pr_P * Pr_FT_cond_P} = {Pr_FT}')

Pr_E_cond_FT = (Pr_FT_cond_E * Pr_E) / Pr_FT
Pr_P_cond_FT = (Pr_FT_cond_P * Pr_P) / Pr_FT

print('\nPor tanto:')
print(f'  - La probabilidad de ser estudiante sabiendo que lee FT es:  P(E|FT) = P(FT|E) P(E) / P(FT) = {Pr_FT_cond_E * Pr_E}/{Pr_FT} = {Pr_E_cond_FT} ')
print(f'  - La probabilidad de ser profesor sabiendo que lee FT es:    P(P|FT) = P(FT|P) P(P) / P(FT) = {Pr_FT_cond_P * Pr_P}/{Pr_FT} = {Pr_P_cond_FT} ')
print(f'    La probabilidad P(P|FT) también se puede calcular como 1- P(E|FT) = 1 - {Pr_E_cond_FT} = {1 - Pr_E_cond_FT}')

print(f'\nLa probabilidad a priori de que una persona sentada en la biblioteca sea un profesor era del {np.round(100*Pr_P,4)}%. '
      f'Sin embargo, si usamos la información adicional de que está leyendo el FT, entonces la probabilidad de que sea profesor aumenta al {np.round(100*Pr_P_cond_FT,4)}%.')

print('\nViendo las probabilidades, parece que Ana tenía razones fundadas para pensar que el juicio a priori de Carlos podía ser prematuro. '
      'Lo más probable es que sea un estudiante, aunque esté leyendo el Financial Times.')
```

### b) ¿Qué probabilidad de leer el Financial Times tendría que tener, como máximo, un estudiante para que Carlos tuviese razón? Justifica la respuesta.

```
# respuesta

print('Se pide determinar el valor de P(FT|E) para que Carlos tenga razón, es decir el valor a partir del cual siempre ocurre que P(E|FT) < P(P|FT). '
      'Según lo explicado en el apartado anterior:')
print('  P(P|FT) = P(FT|P) P(P) / P(FT)')
print('  P(E|FT) = P(FT|E) P(E) / P(FT)')
print('  P(FT) =  P(FT|E) P(E) + P(FT|P) P(P)')
      
print('\nEntonces se busca el valor P(FT|E) para el que se cumple la desigualdad:')
print('P(FT|E) P(E) / P(FT) < P(FT|P) P(P) / P(FT)   -->  P(FT|E) P(E)  <  P(FT|P) P(P)')
print(f'       P(FT|E) P(E)  <  P(FT|P) P(P)          -->  P(FT|E) * {Pr_E}  <  {Pr_FT_cond_P} * {Pr_P}   -->  P(FT|E) * {Pr_E}  <  {Pr_FT_cond_P * Pr_P}')

valor_max = (Pr_FT_cond_P * Pr_P) / Pr_E

print(f'\nEs decir, P(FT|E) tendría que ser menor que {valor_max}.')

print('\nConfirmamos repitiendo los cálculos anteriores pero con la probabilidad calculada:')

# datos de partida
Pr_E = 1500/(1500+110)  # probabilidad a priori
Pr_P = 110/(1500+110)   # probabilidad a priori
Pr_FT_cond_E = valor_max # <-----------
Pr_FT_cond_P = 0.9   

Pr_FT = Pr_E * Pr_FT_cond_E + Pr_P * Pr_FT_cond_P
Pr_E_cond_FT = (Pr_FT_cond_E * Pr_E) / Pr_FT
Pr_P_cond_FT = (Pr_FT_cond_P * Pr_P) / Pr_FT

print('\nLos nuevos resultados:')
print(f'  - La probabilidad de ser estudiante sabiendo que está leyendo FT es:  P(E|FT) = P(FT|E) P(E) / P(FT) = {Pr_FT_cond_E * Pr_E}/{Pr_FT} = {Pr_E_cond_FT} ')
print(f'  - La probabilidad de ser profesor sabiendo que está leyendo es:    P(P|FT) = P(FT|P) P(P) / P(FT) = {Pr_FT_cond_P * Pr_P}/{Pr_FT} = {Pr_P_cond_FT} ')
print(f'Se observa que justo el valor P(FT|E)={valor_max} hace que ambas probabilidades sean iguales. Un valor un poco menor ya hace que ocurra lo esperado.')
print(f'Esto significa que los estudiantes tendrían que leer muy poco el FT (menos de un {np.round(100*valor_max,4)}%) para que si la persona que se observa está leyendo el FT, '
      'entonces tenga sentido afirmar que lo más probable es que sea un profesor, aunque haya muy pocos profesores en comparación con el nº de estudiantes.')
```
