```
# importamos los paquetes necesarios
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
```

### En una pequeña oficina hay únicamente dos redes inalámbricas disponibles W1 y W2. Entre las dos se pretende que haya cobertura wifi en toda la oficina para acceso a Internet. Aunque todos los dispositivos consiguen conectarse, por su localización y potencia, el 70% de las veces los dispositivos se conectan a la red W1.
### Recientemente se ha comprobado que la red W1 tiene más problemas que la W2. En concreto, el 10% de los dispositivos conectados a W1 pierden el acceso a Internet, mientras que de los que se conectan a W2 únicamente un 1% pierde el acceso.

### a) Calcula la probabilidad de que un dispositivo, conectado usando una de las redes de la oficina, pierda el acceso a Internet. Justifica la respuesta.

```
# respuesta

# datos de partida

# Sucesos:
print('Definimos los sucesos:')
print('  W1:  El dispositivo se conecta a la red W1')
print('  W2:  El dispositivo se conecta a la red W2')
print('  D:   El dispositivo se desconecta, pierde acceso a la red')

# Probabilidades proporcionadas
Pr_W1 = 0.7 # la probabilidad de que un dispositivo se conecta a la red 1
Pr_DcondW1 = 0.1
Pr_DcondW2 = 0.01

Pr_W2 = 1 - Pr_W1

print('\nSe tienen las probabilidades:')
print(f'  Prob. de concectarse a W1: P(W1) = {Pr_W1}')
print(f'  Prob. de concectarse a W2: P(W2) = 1-{ Pr_W1} = {Pr_W2}')
print(f'  Prob. de perder acceso estando conectado a W1: P(D|W1) = {Pr_DcondW1}')
print(f'  Prob. de perder acceso estando conectado a W1: P(D|W2) = {Pr_DcondW2}')

print('\nSe pide P(D), la probabilidad de que un dispositivo que esté conectado usando cualquiera de las dos redes de la oficina, pierda acceso.')
print('Dado que W1 y W2 forman una partición disjunta, se puede usar el teorema de la probabilidad total:')
print('  P(D) = P(D|W1) * P(W1) + P(D|W2) * P(W2)')

Pr_D = Pr_DcondW1 * Pr_W1 + Pr_DcondW2 * Pr_W2

print('\nPor tanto, dado que todas las probabilidades son conocidas, es sencillo calcular la probabilidad de perder acceso a internet:')
print(f'P(D) = P(D|W1) * P(W1) + P(D|W2) * P(W2) = {Pr_DcondW1} * {Pr_W1} + {Pr_DcondW2} * {Pr_W2} = {Pr_D}')
```

### b) Si nuestro dispositivo no tiene acceso a Internet en la oficina ¿Qué probabilidad hay de que estemos conectados a través de la red W2?

```
# respuesta

print('Lo que nos están pidiendo es la probabilidad de W2 sabiendo que ha perdido el acceso a la red, es decir, P(W2|D).')
print('Esta probabilidad condicionada se puede calcular de manera sencilla usando el Teorema de Bayes:')
print('P(W2|D) = P(W2 y D) / P(D) = P(D|W2)*P(W2) / P(D)')

# Se trata de una aplicación del Teorema de Bayes: PrW2condP = PrW2yP / PrP = (PrPcondW2 * PrW2) / PrP
Pr_W2condD = (Pr_DcondW2 * Pr_W2) / Pr_D

print('\nLa probabilidad de que estemos conectados a la red W2 si el dispositivo no tiene acceso a intenet entonces es:')
print(f'P(W2|D) = P(D|W2)*P(W2) / P(D) = {Pr_DcondW2}*{Pr_W2} / {Pr_D} = {Pr_W2condD}')
```

### c) Si en un momento dado en la oficina hay 15 dispositivos conectados a alguna de las dos redes wifi disponibles.  Considerando que el acceso a Internet de los dispositivos se realiza de forma independiente y que también pierden la conexión a intenet de forma independiente, calcular la probabilidad de que más de 2 dispositivos pierdan el acceso a Internet.

```
# respuesta

print('La probabilidad pedida implica calcular P(15F o 14F o ... 3F), es decir, la unión de que fallen 15, fallen 14, fallen 13, ... fallen 3.')
print('Se puede interpretar como un espacio muestral formado por 16 sucesos elementales: 15F, 14F, ..., 3F, 2F, 1F y 0F. El suceso 0F representa que '
      'ningún dispositivo pierda la conexión. Lógicamente, todos estos sucesos son disjuntos, si ocurre uno, no pueden ocurrir los otros.')

print('\nHay que considerar el suceso compuesto "15F o 14F o ... 3F", que es lo mismo que "Omega - (2F o 1F o 0F)". Trabajar con el complementario '
      'en este caso es mucho más sencillo. Por tanto, lo que se pide calcular se puede plantear así:')
print('P(15F o 14F o ... 3F) = P(Omega - (2F o 1F o 0F)) = 1 - P(2F o 1F o 0F)')

print('\nComo los sucesos 2F, 1F y 0F son disjuntos, la probabilidad de su unión es directamente la suma de sus probabilidades, no hay intersecciones:')
print('P(15F o 14F o ... 3F) = 1 - P(2F o 1F o 0F) = 1 - P(2F) - P(1F) - P(0F)')

print('\nLa probabilidad de que no falle ninguno se calcula como la probabilidad de que no falle el dispositivo 1 y que no falle el disp. 2, ... '
      'y que no falle el disp. 15. Como el fallo de cada dispositivo es independiente, la probabilidad de la intersección es fácil de calcular:')

Pr_0Fallos = (1-Pr_D) ** 15 # ojo, elevado es ** en python
print(f'P(0F) = P(D1 y D2 y ... D15) = P(D1) * P(D2) * ... * P(D14) * P(D15) = P(D)*P(D)* ...*P(D) = P(D)^15 = {Pr_D}^15 = {Pr_0Fallos}')

print('\nLa probabilidad de que falle 1 dispositivo cualesquiera equivale a que solo falle d1, o que falle solo d2, ..., o que solo falle d15. '
      'Es importante notar que "falle solo d2" significa que "falla d2 y el resto funciona", no es lo mismo que "falla d2". '
      'La probabilidad de que un dispositivo falle y el resto siga funcionando, al ser independientes, se puede calcular con el producto, '
      'teniendo claro que la probabilidad de que el resto (ahora 14) siga funcionando es (1-P(D))^14, siguiendo un razonamiento similar como en P(0F):')

Pr_1Dfalle_y_RestoOK = Pr_D * (1-Pr_D)**14

print(f'P(1dFyRok) = P(1dF) * P(Rok) = P(D) * (1 - P(RestoFalla)) = P(D) * (1 - P(D))^14 = {Pr_D} * {(1-Pr_D)**14} = {Pr_1Dfalle_y_RestoOK}')
      
print('\nPor tanto, como son independientes, la probabilidad de la unión es la suma directamente de 15 casos idénticos:')

Pr_1Fallo = 15 * Pr_1Dfalle_y_RestoOK 
print(f'P(1F) = P(D1yRok o ... D15yRok) = P(D1yRok) + ... + P(D15yRok) = P(1dFyRok) + ...+ P(1dFyRok) = 15 * P(1dFyRok) = 15*{Pr_1Dfalle_y_RestoOK} = {Pr_1Fallo}')
      
print(f'\nLa probabilidad de que fallen 2 dispositivos cualesquiera es la prob. de que "fallen D1 y D2 y el resto siga funcionando", o que "fallen D1 y D3 y resto siga '
     'funcionando", ..., es decir, todas combinaciones de dos parejas fallando y el resto funcionando. Son muchos casos, exactamente son las combinaciones de 15 elementos '
     'tomados de 2 en 2 sin repetición, según esta fórmula:  15! / 2!(15-2)! = 15 * 14 * 13! / 2 * 13! = 15 * 14/2 = 105')

print('La probabilidad de que "dos dispositivos fallen y el resto siga funcionando", al ser independientes, se puede calcular con el producto, '
      'de forma similar al caso de un dispositivo:')

Pr_2Dfallen_y_RestoOK = Pr_D * Pr_D * (1-Pr_D)**13
print(f'P(2dFyRok) = P(2dF) * P(Rok) = P(D) * P(D) * (1 - P(RestoFalla)) = P(D)^2 * (1 - P(D))^13 = {Pr_D}^2 * {(1-Pr_D)**13} = {Pr_2Dfallen_y_RestoOK}')

num_casos_2disp_fallan = np.math.factorial(15)  / (2 *  np.math.factorial(15-2))
Pr_2Fallos = num_casos_2disp_fallan * Pr_2Dfallen_y_RestoOK 
print(f'P(2F) = P(D1yD2yRok o ... D1yD15yRok) = P(D1yD2yRok) + ... + P(D1yD15yRok) = '
      f'P(2dFyRok) + ...+ P(2dFyRok) = 105 * P(2dFyRok) = {num_casos_2disp_fallan}*{Pr_2Dfallen_y_RestoOK} = {Pr_2Fallos}')

Pr_Masde2 = 1 - Pr_2Fallos - Pr_1Fallo - Pr_0Fallos

print('\nPor tanto, la probabilidad pedida es:')
print(f'P(15F o 14F o ... 3F) = 1 - P(2F) - P(1F) - P(0F) = 1 - {Pr_2Fallos} - {Pr_1Fallo} - {Pr_0Fallos} = {Pr_Masde2}')

print('\n-----------------------------------------------------------------------------')
print('Nota: Existe una forma muy elegante y sencilla de calcular este tipo de probabilidades, se verá más adelente cuando se trate el tema '
      'de las distribuciones de probabilidad discretas. Así:')

va = stats.binom(n=15, p=Pr_D)
prob = 1 - va.cdf(2)
print(f'P(nº de fallos>2) = {prob}')
```
