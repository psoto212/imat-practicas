### La empresa de reparto de comida a domicilio RICORICO considera que en cualquier envío se pueden producir los siguientes errores posibles:
#### A:	Se envía un paquete equivocado.
#### B:	El paquete enviado se pierde en el trayecto.
#### C:	El paquete enviado se daña en el trayecto. 

### Supongamos que el suceso A es independiente tanto de B como de C y que los sucesos B y C son incompatibles. 
### Además, se sabe de los registros de la empresa sobre envíos anteriores que P(A) = 0.02, P(B) = 0.01 y que P(C) = 0.04.

```
# datos de partida

# Sucesos:
# A:	Se envía un paquete equivocado.
# B:	El paquete enviado se pierde en el trayecto.
# C:	El paquete enviado se daña en el trayecto.

# probabilidades dadas
Pr_A = 0.02
Pr_B = 0.01
Pr_C = 0.04
```

### a)	Calcula el porcentaje de envíos que realiza esta empresa con algún error. Justifica la respuesta.

```
# respuesta

print('Se pide calcular la probabilidad de que ocurra alguno de los tres errores posibles, es decir '
      'P(A U B U C) o lo que es lo mismo P(A o B o C).')

# otra información porporcionada
print('\nAdemás de las probabilidades de los sucesos elementales, se puede suponer que:')
print('A y B son independientes, con lo que P(A y B) = P(A) * P(B), P(A|B) = P(A) y P(B|A) = P(B)')
print('B y C son incompatibles, con lo que P(B y C) = 0')
  
print('\nGráficamente la situación puede ser algo así (B y C incompatibles):')
print('                     |-------------  A  --------------|')
print('      |---------  B  ---------|')
print('                                          |--------  C  --------|')
    
print('La fórmula que tenemos que aplicar es:')
print('P(A o B o C) = P(A) + P(B) + P(C) - P(A y B) - P(A y C) - P(B y C) + P(A y B y C)')

print('\nPara calcular dicha probabilidad es necesario conocer las probabilidades de varias intersecciones: ')

# cálculos intermedios
Pr_AyB = Pr_A * Pr_B    # A y B indep, es el producto
Pr_AyC = Pr_A * Pr_C    # A y C indep, es el producto
Pr_ByC = 0              # B y C incompatibles
Pr_AyByC = 0            # como ByC=conjunto vacio, AyByC=vacio vacio también y su probabilidad es cero

print(f'La P(A y B) = P(A)*P(B) = {Pr_A} * {Pr_B} = {Pr_AyB}, ya que son sucesos independientes.')
print(f'La P(A y C) = P(A)*P(C) = {Pr_A} * {Pr_C} = {Pr_AyC}, ya que también son sucesos independientes.')
print(f'La P(B y C) = {Pr_ByC}, al ser sucesos incompatibles.')
print(f'La P(A y B y C) = {Pr_AyByC}, al ser sucesos incompatibles B y C.')

prob =  Pr_A + Pr_B + Pr_C - Pr_AyB - Pr_AyC - Pr_ByC + Pr_AyByC
print('\nLa probabilidad solicitada es entonces:', prob)
```
