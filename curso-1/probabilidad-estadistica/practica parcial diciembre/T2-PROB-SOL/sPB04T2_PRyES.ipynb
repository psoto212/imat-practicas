El precio horario de la electricidad es muy dependiente del mix de tecnologías que se emplee para generar la electricidad. Si la producción con la tecnología solar y eólica es alta, entonces el precio puede ser bajo.

Se quiere estudiar los casos en los que el precio es bajo, según las producciones de solar y eólica sean altas o normales. La probabilidad de que la producción solar sea alta es 0.2, mientras que la probabilidad de que la eólica sea alta es de 0.6. Además, se puede considerar que la producción solar es independiente de la eólica.

La probabilidad de que el precio sea bajo, sabiendo que la solar y la eólica han sido altas al mismo tiempo es de 0.8. Además, la probabilidad de que el precio sea bajo es nula tanto si la producción eólica es normal como cuando la solar no es alta.

### a) Calcular la probabilidad de que la producción solar y eólica sean altas al mismo tiempo. Justifica la respuesta (2 puntos)

```
# respuesta

# Sucesos:
print('Definimos los sucesos:')
print('  PB: precio electricidad bajo')
print('  SA: producción solar alta (noSA: solar normal)')
print('  EA: producción eólica alta (noEA: eólica normal)')

# datos de partida
Pr_SA = 0.2
Pr_EA = 0.6

# como son independientes
Pr_SAyEA = Pr_SA * Pr_EA

print('\nSe tienen las probabilidades:')
print(f'  Prob. de la solar sea alta: P(SA) = {Pr_SA}')
print(f'  Prob. de la eólica sea alta: P(EA) = {Pr_EA}')

print('\nSA y EA son sucesos independientes')
print('la probabilidad de que la producción solar y eólica sean altas al mismo tiempo se obtiene como la probabilidad de la intersección')
print('Como son independientes SA y EA, la prob. de la intersección es el producto')
print('Por tanto, la probabilidad pedida es PrSAyEA = PrSA * PrEA = ',  Pr_SAyEA)
```

### b) Calcular la probabilidad, justificando la respuesta, de que el precio horario de la electricidad sea bajo (4 puntos)

```
# respuesta

#hay que calcular PrPB, usamos el teorema de la probabilidad total

print('Los sucesos SAyEA, SAynoEA, noSAyEA y noSAynoEA forman una partición')

#La probabilidad de que el precio sea bajo, sabiendo que la solar y la eólica han sido altas al mismo tiempo es de 0.8.
Pr_PB_cond_SAyEA = 0.8

#Además, la probabilidad de que el precio sea bajo es nula tanto si la producción eólica es normal como cuando 
#la solar no es alta.
Pr_PB_cond_noEA = 0.
Pr_PB_cond_noSA = 0.

# por tanto:
Pr_PB_cond_SAynoEA = 0.
Pr_PB_cond_noSAyEA = 0.
Pr_PB_cond_noSAynoEA = 0.

print('Usamos: PrPB = PrPBcond_SAyEA * PrSAyEA + PrPBcond_SAynoEA * PrSAynoEA + PrPBcond_noSAyEA * PrnoSAyEA + PrPBcond_noSAynoEA * PrnoSAynoEA')

print('Las condicionadas de los casos con noEA y noSA son nulas, no es necesario considerarlos')
# como tres términos son nulos entonces:
Pr_PB = Pr_PB_cond_SAyEA * Pr_SAyEA

print('La probabilidad de que el precio sea bajo es:', Pr_PB)
```
