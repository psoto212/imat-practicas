```
# importamos paquetes aquí
import numpy as np
from scipy import stats
import pandas as pd
import matplotlib.pyplot as plt
```

#### La compañía de teléfonos móviles MAGENTA está revisando la prima anual que debe cobrar  por un seguro a todo riesgo para las pantallas de los móviles de alta gama que ofrece a sus clientes. El seguro cubre la reparación de la pantalla, lo que supone un coste de 230€ para la compañia. La aseguradora ha determinado, a partir de su base de datos, que la tasa de rotura anual de las pantallas de estos móviles es de 50 por 1000.

#### a) Se quiere calcular la prima anual del seguro que tiene que cobrar por cada móvil asegurado para que, de media, la compañía no gane ni pierda dinero ofreciendo este servicio a sus clientes.

```
# Respuesta
print('Por cada 1000 móviles asegurados la empresa ingresará 1000 * prima')
print('Por cada 1000 móviles asegurados, es de esperar que haya que reparar 50')
print('Reparar 50 móviles supone un coste de 50 * 230 € para la empresa')
print('Para que no le cueste dinero a MAGENTA tiene que ocurrir que: 1000 * prima = 50 * 230 ')

prima = 50 * 230 /1000
print('Por tanto, la prima será', prima, '€')
```

#### b) En realidad el coste que supone la reparación de una pantalla no es exactamente 230€. Preguntando al responsable se ha sacado en claro que este coste de reparación depende del modelo exacto de móvil, del tiempo invertido en repararlo, del daño concreto, etc. En cualquier caso, como mínimo es 200€, nunca superior a 300€, y lo más probable es que valga 230€. Según esta información, ¿Cúal es la probabilidad de que el coste de reparación de una pantalla sea superior a 230€? Representa la función de densidad y de distribución de probabilidad de la v.a. empleada para resolverlo.

```
# Respuesta

print('El coste se puede modelar como una v.a. C ~ Triangular (a=200,c=230,d=300)')
print('Se pide P(C>230)')

a = 200 # min
c = 230 # moda 
b = 300 # max

loc = a
escalado = b-a
forma = (c-a)/escalado
VA = stats.triang(loc=loc, scale=escalado, c = forma)

# calculamos la probabilidad
prob = 1-VA.cdf(230)

print('Por tanto P(C>230) = 1 - P(C<=230) = 1 - cdf(230) =', prob)

# representamos la pdf y la cdf

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(a-10, b+10, 1000)
xs_pedido = np.linspace(c, b+10, 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))

ax1 = plt.subplot(2,1,1)
plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.fill_between( xs_pedido, VA.pdf(xs_pedido), color="black", alpha=0.3, label = 'P(C>230)')
plt.plot( xs, VA.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.legend()

plt.subplot(2,1,2, sharex= ax1)
plt.plot(xs, VA.cdf(xs), color='blue', label = 'cdf(x)')

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```

#### c) Teniendo en cuenta la información adicional proporcionada por el responsable en el apartado anterior, vuelve a calcular la prima anual que tiene que cobrar para que de media la compañía no pierda ni gane dinero ofreciendo este servicio a sus clientes.

```
# Respuesta

print('Ahora el coste de la reparación es una v.a. C')

print('Para que no le cueste dinero de MEDIA a MAGENTA tiene que ocurrir que: 1000 * prima = 50 * media_C')

media_VA = VA.mean()
prima = 50 * media_VA/1000
print('La media de la v.a. es:', media_VA, '€')
print('Por tanto, la prima será ahora', prima, '€')
```
