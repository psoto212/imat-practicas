#### Se ha definido la v.a. X que se distribuye según una binomial de parámetros n = 20 y p = 0.2. Con esta variable se está modelando el nº de días que un alumno perezoso llega tarde a la clase de las 8 de la mañana en un mes típico de 4 semanas con 5 días lectivos cada una.

```
# importamos paquetes aquí
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

#### a) Indica la probabilidad de que X valga exactamente 40, sin realizar ningún cálculo. Justifica la respuesta.

```
# Respuesta

print('X: v.a. nº días alumno perezoso llega tarde en 1 mes (4x5 días lectivos)')
print('X~B(n=20, p=0.2)')

print('La P(X=40) es nula, ya que es imposible que X alcance ese valor, como mucho puede valer 20')

# comprobación
X = stats.binom(n=20,p=0.2)
prob = X.pmf(40)
print('Como tenía que ser, P(X=40) =', prob)
```

#### b) Siguiendo la misma filosofía de modelado, indica la distribución que sigue una v.a. que representa el nº de días que llega tarde a la clase de las 8 de la mañana un alumno responsable en un trimestre típico de 12 semanas, todas ellas de 5 días lectivos. Considera que la probabilidad de que un alumno responsable llegue tarde un día es 0.01. Justifica la respuesta.

```
# Respuesta

print('Sea Y la v.a. que representa ese nº de días. Entonces sigue una binomial, Y~B(n=12*5, p=0.01)')

# creamos la v.a.
n = 12*5 
p = 0.01
Y = stats.binom(n=n,p=p)
```

#### c) Calcula, utilizando la distribución del apartado anterior, la probabilidad de que el alumno responsable llegue como máximo 4 días tarde en un trimestre típico.

```
# Respuesta
print('Hay que calcular P(Y<=4)')

prob = Y.cdf(4)
print('P(Y<=4) = cdf(4) = ', prob)
```

#### d)	Calcula la probabilidad que tendría que tener un alumno muy responsable de llegar tarde un día cualquiera para que la probabilidad de llegar a la hora durante 56 días de los 60 días del trimestre típico fuese igual o superior a un 98%

```
# Respuesta

print('Sea Z: v.a. nº de días que el alumno llega tarde en 60 días')
print('La probabilidad de llegar a la hora 56/60 días es igual que la probabilidad de llegar tarde 4/60 días')
print('Piden la probabilidad de tener éxito (llegar tarde) p en Z~B(n=60, p) tal que P(Z<=4) = 1 - 0.98 (o superior)')

# se podría plantear la ecuación de la cdf=0.98 y despejar p, puff

# Por fuerza bruta, un barrido en valores de p
num_valores_p = 30
p_min = 0.1646010 # despues de iterar un poco ...
p_max = 0.1689
lista_p= np.linspace(p_min, p_max, num_valores_p)

n = 12*5

for p in lista_p:
    Z = stats.binom(n=n,p=p)
    prob = Z.cdf(4)
    print('p=',p,' -> P(Z<=4)=', prob, '  >> Valor buscado:', 0.02)

# revisando resultados elegimos valor de p
p_elegida = 0.166972 # valor aproximado ...
Z_def = stats.binom(n=n,p=p_elegida)
prob = Z_def.cdf(4)
print('Se puede ver que con p =',p_elegida,' P(Z<=4)=', prob)
print('Es decir, que un alumno muy responsable podría tener una probabilidad diaria de llegar tarde hasta del ', np.round(100*p_elegida,4),'%')
print('y la probabilidad de llegar a la hora 56 de los 60 días sería igual o superior al 98%')


# visualmente chequeamos que está bien

xs = range(-1,7)
fig = plt.figure(1, figsize=(8, 6))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, Z_def.pmf(xs))
plt.ylabel('pmf(x)')
plt.xlim(xs[0], xs[-1])
plt.title('Binomial(n='+ str(n)+', p='+ str(round(p_elegida,4))+')')

plt.subplot(2,1,2, sharex = ax1)
plt.step(xs, Z_def.cdf(xs), color='red', where='post')
plt.plot(4,Z_def.cdf(4), 'sk')
plt.ylabel('cdf(x)')
plt.xlabel('Z: nº de días llega tarde')
plt.grid()
plt.show()
```
