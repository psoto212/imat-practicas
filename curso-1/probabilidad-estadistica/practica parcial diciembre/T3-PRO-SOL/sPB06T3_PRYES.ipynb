#### Un parque eólico dispone de 40 aerogeneradores similares. Según el fabricante, la probabilidad de que falle uno de estos aerogeneradores en una semana cualquiera vale 0.001.

```
# importamos paquetes aquí
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

#### a) A partir de la información proporcionada por el fabricante, ¿Cuál es la probabilidad de que en una semana se averíen más de 2 aerogeneradores?

```
# respuesta

print('n = 40 máquinas, prob. de fallo p = 0.001');
print('Sea X: v.a. nº de generadores que fallan semanalmente de los 40')
print('X~Binomial(n=40, p=0.001)')
print('Se pide P(X>2)')
n= 40
p=0.001
X = stats.binom(n=n,p=p)
prob = 1 - X.cdf(2)
print('La probabilidad pedida P(X>2) = 1 - P(X<=2) = 1 - cdf(2) =', prob)
```

#### b) A partir de la información proporcionada por el fabricante, ¿Cuál es la probabilidad de que en una semana se averíen al menos 2 aerogeneradores?¶

```
# respuesta

print('Al menos 2 significa que en una misma semana se averíen 2 o más')
prob = 1 - X.cdf(2) + X.pmf(2)

print('Se pide P(X>=2) = P(X>2) + P(X=2) = 1 - P(X<=2) + P(X=2) = 1 - cdf(2) + pmf(2) =', prob)
```
