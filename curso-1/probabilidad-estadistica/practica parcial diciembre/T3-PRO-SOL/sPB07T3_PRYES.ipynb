# Solución Problema PB07  |  Tema 3 - PRyES

#### Un taller de reparación de drones emplea en cada reparación un tiempo, en horas, tal que la duración media de las reparaciones es de 4 horas y la desviación típica es de 1.15 horas. Además, el taller factura cada hora a 32€. Calcular razonadamente la probabilidad de que la recaudación por la reparación de 50 drones sea superior a 6500€.

```
# importamos paquetes aquí
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

```
# respuesta

print('Se pide P(R>4300€), siendo R: v.a. recaudación de 50 drones')
print('La recaudación de un dron es 32€/hora * T, siendo T una v.a. que representa el tiempo de reparación')
print('De la v.a. T se conoce la media 4h y la std 1.15h')
print('Considerando que el tiempo de reparacíon de todos los drones es i.i.d., ')
print('la distribución de R (suma de 50) se puede modelar como una normal')
print('R = suma_50(32 * T) = 32 * suma_50(T) = 32 * S, donde S ~ Normal(mu_S,sigma_S)')
print('R= 32*S es una transf. lineal de una Normal, por tanto R es una normal')
print('R ~ Normal(mu = 32*mu_S, sigma = 32 * sigma_S)')
print('mu_S = 50 * 4 horas y sigma_S = sqrt(50) * 1.15 horas')

mu = 32 * 50 * 4
sigma = 32 * np.sqrt(50) * 1.15
print('R ~ Normal(mu=', mu,',', 'sigma=', sigma,')')

VA = stats.norm(loc = mu, scale = sigma)
prob = 1-VA.cdf(6500)
print('Por tanto, la probabilidad de que la recaudación por la reparación ' 
      'de 50 drones sea superior a 6500 € es P(R>6500)=1-P(R<=6500)=', prob)
```
