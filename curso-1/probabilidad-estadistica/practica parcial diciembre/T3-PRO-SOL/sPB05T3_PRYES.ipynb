# Solución Problema PB05  |  Tema 3 - PRyES

#### En un edificio de oficinas se dispone de dos conjuntos de ascensores que utilizan los empleados de forma masiva para subir a las plantas en las que trabajan. Los 4 ascensores de la izquierda son más rápidos que los de la derecha, un poco más antiguos y que se averían más veces. A primera hora de la mañana siempre hay cola para coger el ascensor, una cola para los de la izquierda y otra para los de la derecha. Cuando solo están funcionando los ascensores de la izquierda, el tiempo que se espera en la cola de la planta baja a primera hora es una variable aleatoria con distribución exponencial de media 5 minutos. 

```
# importamos paquetes aquí
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
```

#### a) Calcular la probabilidad de tener que esperar más de 3 minutos en la cola de la planta baja para coger el ascensor cuando solo están funcionando los ascensores de la izquierda.

```
# respuesta

print('Hay que calcular P(T>3) siendo T: v.a. tiempo de espera en la cola de la izq., una exponencial de media 5 min')

media = 5

lmbda= 1/media
escalado = 1 / lmbda
VA = stats.expon(loc = 0, scale = escalado)

prob = 1-VA.cdf(3)
print('La probabilidad P(T<3)=1-cdf(3)=', prob)
```

#### b) Calcular la probabilidad de que el tiempo de espera en la cola de la planta baja esté entre 3 y 5 minutos, cuando solo están funcionando los ascensores de la izquierda.

```
# respuesta

print('Hay que calcular P(T>3 & T<5) siendo T: v.a. tiempo de espera cola de la izquierda, una exponencial de media 5 min')
prob = VA.cdf(5) - VA.cdf(3)
print('P(T>3 & T<5) = cdf(5) - cdf(3) =', prob)
```

#### c) Si solo están funcionando los ascensores de la izquierda y una persona lleva esperando en la cola 3 minutos, calcular la probabilidad de que tenga que esperar en total más de 5 minutos.

```
# respuesta

# usando la propiedad de ausencia de memoria de la exponencial
print('Cálculo basado en la ausencia de memoria:')
print('P(T>5 | T>3) = P(T>2) ya que saber que no ha llegado en 3 minutos es como poner el contador del tiempo de espera ahí')
p_mayor2 = 1 - VA.cdf(2)
print('P(T>5 | T>3) = P(T>2) = 1 -  P(T<=2) = ', p_mayor2)

print('Confirmamos el resultado utilizando la probabilidad condicionada')
print('Hay que calcular P(T>5 | T>3) = 1 - P(T<=5 | T>3)')
print('P(T<=5 | T>3) = P(T<=5 & T>3) / P(T>3) = P(T<5 & T>3) / P(T>3)')

p_mayor3 = 1 - VA.cdf(3)
p_mayor3ymenor5 =  VA.cdf(5) - VA.cdf(3)
prob = 1 - (p_mayor3ymenor5 / p_mayor3)
print('P(T>5 | T>3) = 1 - P(T<=5 | T>3) = 1 - P(T<5 & T>3) / P(T>3) = 1 -', p_mayor3ymenor5,'/', p_mayor3, '=',prob)
print('Lógicamente se llega al mismo resultado')

print('\nOperando la probabilidad condicionada de otra forma:')
print('P(T>5 | T>3) =  P(T>5 & T>3) / P(T>3)')
print('P(T>5 & T>3) = P(T>5)')

p_mayor5 = 1 - VA.cdf(5)
prob = p_mayor5 / p_mayor3
print(f'P(T>5 | T>3) =  P(T>5 & T>3) / P(T>3) =  P(T>5) / P(T>3) = {p_mayor5}/{p_mayor3} = {prob}')
print('Todo encaja si se opera bien')
```

#### d) Los 3 ascensores de la derecha son más lentos que los de la izquierda, el tiempo de espera en la cola de la planta baja a primera hora para coger alguno de estos 3 ascensores es de 8 minutos. Un empleado veterano se coloca en una de las dos colas en función del nº de personas que ya hay esperando en cada una y según su percepción del lado más rápido, situándose en la cola de la izquierda el 70% de las veces. Si ha tardado menos de 4 minutos en coger el ascensor, ¿cuál es la probabilidad de que haya utilizado un ascensor de la derecha?

```
# respuesta

print('Se pide P(colaDE | T<=4) = P(colaDE & T<=4) / P(T<=4)')
print('Donde colaDE: esperar en la cola de la derecha, idem para colaIZ')
print('P(colaDE) = 0.3 y P(colaIZ) = 0.7') 
print('Usado Bayes se puede resolver fácilmente:')
print('P(colaDE & T<=4) = P(T<=4 | colaDE) * P(colaDE)')

# P(colaDE & T<=4)
print('Para calcular P(T<=4 | colaDE) usamos la exponencial de 8 minutos')
media = 8
lmbda= 1/media
escalado = 1 / lmbda
VAdrcha = stats.expon(loc = 0, scale = escalado)

p_menos4min_colaDE  = VAdrcha.cdf(4)
print('P(T<=4 | colaDE) =', p_menos4min_colaDE)

# teorema prob total para obtener el denominador
print('Para calcular P(T<=4) usamos la partición dada por colaDE, colaIZ')
p_colaDE = 0.3
P_colaIZ = 0.7
p_menos4min_colaIZ  = VA.cdf(4)
print('P(T<=4 | colaIZ) =', p_menos4min_colaIZ)
p_menos4 = p_menos4min_colaDE * p_colaDE + p_menos4min_colaIZ * P_colaIZ
print('P(T<=4) = P(colaDE & T<=4) + P(colaIZ & T<=4) = P(T<=4 | colaDE) * P(colaDE) + P(T<=4 | colaIZ) * P(colaIZ) = ', p_menos4)

prob = p_menos4min_colaDE * p_colaDE / p_menos4
print('Por tanto la probabilidad de que haya utilizado un ascensor de la derecha es P(colaDE | T<=4) = P(colaDE & T<=4) / P(T<=4) =', prob)
```
