# Solución Problema PB03  |  Tema 3 - PRyES

#### Calcula la probabiliad de que un alumno apruebe un determinado examen (nota >=5), sabiendo que el alumno piensa que la nota más probable que va a obtener en dicho examen es un 7. Las notas pueden variar entre 0 y 10.

```
# importamos paquetes aquí
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```
# Respuesta

print('Información disponible: Nota entre 0 y 10 (acotada) y la moda=7')
print('Definimos la v.a. X: nota del alumno')
print('X ~ triangular(a=0, c=7, b= 10)')

# creamos la v.a. continua

a = 0 # min
c = 7 # moda 
b = 10 # max

loc = a
escalado = b-a
forma = (c-a)/escalado
VA = stats.triang(loc=loc, scale=escalado, c = forma)

# calculamos la probabilidad
prob = 1-VA.cdf(5)
print('La probabilidad de que apruebe es P(nota>=5)=', prob)
```

```
# verificamos visualmente

# puntos en los que se evalua las funciones para representarlas
xs = np.linspace(VA.ppf(0.01)-1, VA.ppf(0.99)+1, 1000)
xs_pedido = np.linspace(5, VA.ppf(0.99)+1, 1000)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(6, 4))

plt.fill_between( xs, VA.pdf(xs), color="orange", alpha=0.3)
plt.fill_between( xs_pedido, VA.pdf(xs_pedido), color="black", alpha=0.3, label = 'P(nota>=5)')
plt.plot( xs, VA.pdf(xs), color="orange", label = 'pdf(x)')
#plt.xlim(xs[0], xs[-1])
plt.plot(xs, VA.cdf(xs), color='blue', label = 'cdf(x)')

# marco el valor x=5
plt.vlines(x=5, ymin=0, ymax=VA.cdf(5), color='k')
plt.hlines(y=VA.cdf(5), xmin=0, xmax=5, color='k')
plt.plot(5, VA.cdf(5),'sk')

plt.legend()

plt.xlabel('x')
plt.ylabel('p')
plt.legend()
plt.grid()
plt.show()
```
