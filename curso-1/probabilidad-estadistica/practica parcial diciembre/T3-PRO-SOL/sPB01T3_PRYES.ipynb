#### Se lanzan simultáneamente 20 veces 3 dados normales pero de distinto color (uno blanco, uno rojo y otro azul).  

```
# importamos paquetes aquí
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

#### a) Indica, justificando la respuesta, qué es más probable: Que el nº de cincos obtenidos con el dado blanco sea igual a 10 o que el nº de cincos obtenidos entre el dado rojo y el azul sea igual a 10. Confirma tu respuesta calculando ambos resultados.

```
# Respuesta
print('Se pide comparar P(B=10) con P(RyA=10)')
print('siendo B: nº de 5s con el dado Blanco, R: nº de 5s con el rojo y A: nº de 5s con el dado Azul.')

print('Los tres dados se tiran el mismo nº de veces (20), es como si un dado rojiazul se tirase 40 veces, por lo que el nº de cincos '
      'obtenidos con ese dado de dos colores será mayor que con el blanco, que solo se tira 20 veces.')

# calculamos las probabilidades
print('B: v.a. nº de 5 en 20 tiradas con dado blanco, B~Binomial(n=20,p=1/6)')
print('De la misma forma R~Binomial(n=20,p=1/6) y A~Binomial(n=20,p=1/6)')
print('El nº de 5 entre R y A se puede modelar así RyA~Binomial(n=2*20,p=1/6), es como si fuese un único dado que se tira 40 veces.')

# creamos las 3 v.a
B = stats.binom(n=20,p=1/6)
R = stats.binom(n=20,p=1/6)
A = stats.binom(n=20,p=1/6)
RyA = stats.binom(n=2*20,p=1/6)

pBigual10 = B.pmf(10)
pRyAigual10 = RyA.pmf(10)

print('La P(B=10) = ', pBigual10)
print('La P(RyA=10) = ', pRyAigual10)
print('Se confirma que P(RyA=10) es más probable que P(B=10)')

# visualmente para entender qué pasa con ambas distribuciones y que ocurre en 10
xs = range(-2, 20)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(8, 6))
ax1 = plt.subplot(2,1,1)
plt.bar(xs, R.pmf(xs), color = 'orange', label='dado Blanco')
plt.stem(xs, RyA.pmf(xs), basefmt='none', label='dado Rojiazul')
plt.ylabel('pmf(x)')
plt.grid()
plt.legend()
plt.title('Binomiales del dado Blanco y el Rojiazul')
plt.show()

print('Ojo, si nos hubíesen preguntado por la prob. de obtener exactamente 3 cincos en lugar de  10, la respuesta sería la contraria ...')
```

```
# Razonamiento alternativo

print('Sabiendo que la Binomial es reproductiva, y que: ')
print('  Dado rojo: R~Binomial(n=20,p=1/6)')
print('  Dado azul: A~Binomial(n=20,p=1/6)')
print('La v.a. X: "nº de cincos entre los dados rojo y azul en 20 tiradas" es exactamente X=R+A, '
      'como R y A tienen la misma p, entonces se cumple que X ~ Binomial(n = 20 + 20, p=1/6), la misma '
      'distribución a la que se había llegado de forma intuitiva.')
```

#### b) Calcular la probabilidad de que, al mismo tiempo, el nº de cincos obtenidos con el dado blanco sea mayor que 3 y que el nº de cincos obtenidos con el dado azul sea igual a 10. Justifica la respuesta.

```
# Respuesta
print('Se pide P(B>3 & A=10)')
print('Son sucesos independientes, por lo que  P(B>3 & A=10) =  P(B>3) * P(A=10)')

pBmayor3 = 1 - B.cdf(3)
pAigual10 = A.pmf(10)

print('P(B>3) = 1 - P(B<=3) = 1 - cdf(3) = ', pBmayor3)
print('P(A=10) = pmf(10) = ', pAigual10)
print('Por tanto, P(B>3 & A=10) =  P(B>3) * P(A=10) = ', pBmayor3*pAigual10)
```

#### c) Sea X la v.a. número total de cincos que aparecen en los 20 lanzamientos considerando los 3 dados. Razónese la ley de probabilidad que sigue la v.a. X.

```
# Respuesta
print('La v.a. X  = B + R + A (la suma del nº de 5 obtenidos con los 3 dados)')
print('Las v.a. B, R y A siguen la misma B(n=20, p=1/6), con la misma p')
print('La distribución binomial es reproductiva, la suma B + R + A sigue una binomial')
print('Luego X ~ B(n=3*20, p=1/6)')

n = 3*20
p = 1/6
X = stats.binom(n=n,p=p)

# visualmente es así
# puntos en los que se evalua las funciones para representarlas
xs = range(-2, 62)

# pinta sincronizadametne ambas funciones
fig = plt.figure(1, figsize=(8, 6))
ax1 = plt.subplot(2,1,1)
plt.stem(xs, X.pmf(xs), basefmt='none')
plt.ylabel('pmf(x)')
plt.grid()
plt.title(f'Binomial(n={n}, p={round(p,4)})')
plt.show()
```

#### d)	Calcular la probabilidad de que se obtengan al menos 7 cincos entre los tres dados.

```
# Respuesta
print('Al menos 7 cincos con los 3 dados en los 20 lanzamientos implica calcular P(X>=7)')

prob = 1 - (X.cdf(7) - X.pmf(7))
print('P(X>=7) = 1 - P(X<7) = 1 - [cdf(7) - pmf(7)] =', prob)
```
